## Notebooks

Notebooks are interactive work environments: you can run code and display text in them. They're very handy to:

1. Iterate quickly on code and data while prototyping a study or a project
1. Present the results of your work, with all the benefits of executable code powering charts / visualizations.

A notebook is a sequence of cells. You can delete or add cells at any time, then run them (if you have edit rights).

### Text Cells

A text cell displays text, images or tables.

The syntax is Markdown, a very simple language to describe formatted text. An [official guide](https://colab.research.google.com/notebooks/markdown_guide.ipynb) is available for all the details (you can also expand this cell to get details in this notebook).


A link is created with its text between square brackets, followed by its address between parentheses:

```markdown

```

```markdown
![xkcd on Machine Learning](https://imgs.xkcd.com/comics/machine_learning.png)
```

### List Comprehensions and Generators

You can define lists very quickly with Python's *list comprehensions*. They use the syntax

```python
[result for variable in list if condition]
```
where `result` and `condition` use `variable`.

````
squares = [k ** 2 for k in range(10)]
print(squares)

even_squares = [k for k in squares if k % 2 == 0]
print(even_squares)

for i, j in [(k, k ** 2) for k in range(10)]:
  print(i, j)
````

A generator is defined like a list comprehension, but with parentheses instead of square brackets.

The difference is that the list defined by the generator is only computed when it's used, and can only be used once (it isn't stored in memory but computed on the fly).

````
def make_even_squares_generator():
  return (k ** 2 for k in range(10) if k % 2 == 0)

even_squares = make_even_squares_generator()

for square in even_squares:
  print(square)

print("***")

# Does nothing: the generator has already been used
for square in even_squares:
  print(square)

print("***")

# The generator must be recreated to iterate again
for square in make_even_squares_generator():
  print(square)
````

### File Handling

To handle files, we use a `with` block, which makes sure the resource is properly closed, even when exceptions occur.

````
from pathlib import Path

with Path("sample_data/README.md").open(encoding="utf8") as fh:
  content = fh.read()
  



lines = content.splitlines()
print(lines[3])
````

To write, the mechanism is the same, but we specify the *writing* mode `w`:

````
from pathlib import Path

with Path("output.txt").open(encoding="utf8", mode="w") as fh:
  fh.write("Hello World!")

print(Path("output.txt").read_text())
````

### List and Dictionary Unpacking

The `*` operator packs scattered elements into a list, or unpacks lists into scattered elements; the `**` operator is its equivalent for dictionaries.

````
def square(x, y):
  return x ** 2 + y ** 2

square(*[2, 4])

# *args packs "Greetings:" and "Hello" into ["Greetings:", "Hello"]
# **kwargs packs whom="World!" into {"whom": "World!"}
def main(*args, **kwargs):
  # *args unpacks ["Greetings:", "Hello"] into "Greetings:", "Hello"
  print(*args, kwargs["whom"])

main("Greetings:", "Hello", whom="World!")
````
