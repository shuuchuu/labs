Pour chaque mot de passe, mettez quatre booléens à `False` (un par sorte de caractère), passez-les à `True` en parcourant les caractères, puis ajoutez à une liste `problemes` un message par règle non respectée. ### Conseils supplémentaires

Un motif `case ["ajouter", article]:` reconnaît une liste de deux mots dont le premier est `"ajouter"`, et range le second dans `article` ; `case _:` attrape tout le reste, y compris `ajouter du thé` (trois mots). ```python


## Retrouver la version fautive par dichotomie

Un logiciel a connu 200 versions. C'est ce que fait `git bisect` sur l'historique d'un dépôt.

### Conseils supplémentaires

Parcours dans l'ordre : `enumerate`, et `break` à la première version qui échoue. Dichotomie : gardez deux bornes `bas` et `haut` telles que la première version boguée soit toujours entre les deux (au départ, la première et la dernière version, qui est boguée). Tant que `bas < haut`, consultez `milieu = (bas + haut) // 2` : si ses tests passent, le bogue est après, sinon il est en `milieu` ou avant. Vous devez trouver la version 137 dans les deux cas, en 138 consultations, puis en 7.

```python
nb_consultations = 0
for version, ok in enumerate(tests_ok):
  nb_consultations += 1
  if ...:
    premiere_boguee = version
    break
print(f"Dans l'ordre : version {premiere_boguee}, {nb_consultations} consultations")

bas = 0
haut = len(tests_ok) - 1
nb_consultations = 0
while bas < haut:
  milieu = (bas + haut) // 2
  nb_consultations += 1
  if tests_ok[milieu]:
    bas = ...
  else:
    haut = ...
print(f"Par dichotomie : version {bas}, {nb_consultations} consultations")
```

### Solution

````python
nb_consultations = 0
for version, ok in enumerate(tests_ok):
  nb_consultations += 1
  if not ok:
    premiere_boguee = version
    break
print(f"Dans l'ordre : version {premiere_boguee}, {nb_consultations} consultations")

bas = 0
haut = len(tests_ok) - 1
nb_consultations = 0
while bas < haut:
  milieu = (bas + haut) // 2
  nb_consultations += 1
  if tests_ok[milieu]:
    bas = milieu + 1
  else:
    haut = milieu
print(f"Par dichotomie : version {bas}, {nb_consultations} consultations")

# Chaque consultation coupe l'intervalle en deux : 200 versions demandent au
# plus 8 consultations (2 ** 8 = 256), contre 138 ici dans l'ordre.
````
