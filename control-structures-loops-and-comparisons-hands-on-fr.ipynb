### Conseils supplémentaires

Dans la boucle, testez d'abord la fin de transmission (`break`), puis le relevé manquant (comptez-le, puis `continue` passe au suivant) ; le reste de la boucle ne voit plus que des relevés valides. Vous devez obtenir 6 relevés valides, 2 manquants et une moyenne de `3.33` mm. Pour la question, essayez `not 0.0`.

```python
total = 0

nb_manquants = 0
for releve in pluie_mm:
  if releve == -1:
    ...
  if releve is None:
    nb_manquants += 1
    ...
  total += releve
  nb_valides += 1
print(f"{nb_valides} relevés valides, {nb_manquants} manquants")
print(f"Moyenne : {...:.2f} mm")
```

### Solution

````python
total = 0
nb_valides = 0
nb_manquants = 0
for releve in pluie_mm:
  if releve == -1:
    break
  if releve is None:
    nb_manquants += 1
    continue
  total += releve
  nb_valides += 1
print(f"{nb_valides} relevés valides, {nb_manquants} manquants")
print(f"Moyenne : {total / nb_valides:.2f} mm")

# `not releve` est vrai aussi pour 0.0, un jour sans pluie : un relevé valide.
# Le test `not releve` écarterait les deux jours secs, et la moyenne deviendrait
# 20.0 / 4 = 5.00 mm. `is None` ne vise que le relevé manquant.
````

### Conseils supplémentaires

`zip(altitudes, altitudes[1:])` associe chaque relevé au suivant : `(1020, 1085)`, `(1085, 1160)`… Pour chaque couple, l'écart `apres - avant` est une montée s'il est positif, une descente sinon. Pour la plus forte montée, gardez la plus grande valeur vue jusqu'ici. Vous devez obtenir 385 m de montée, 125 m de descente et une plus forte montée de 95 m.

```python
montee = 0
descente = 0
plus_forte_montee = 0
for avant, apres in zip(...):
  ecart = apres - avant
  if ecart > 0:
    montee += ecart
    if ...:
      plus_forte_montee = ecart
  else:
    descente += ...
print(f"D+ : {montee} m, D- : {descente} m")
print(f"Plus forte montée : {plus_forte_montee} m")
```

### Solution

````python
montee = 0
descente = 0
plus_forte_montee = 0
for avant, apres in zip(altitudes, altitudes[1:]):
  ecart = apres - avant
  if ecart > 0:
    montee += ecart
    if ecart > plus_forte_montee:
      plus_forte_montee = ecart
  else:
    descente += -ecart
print(f"D+ : {montee} m, D- : {descente} m")
print(f"Plus forte montée : {plus_forte_montee} m")

# itertools.pairwise(altitudes) donne les mêmes couples que ce zip.
# Le dénivelé positif moins le dénivelé négatif donne l'arrivée moins le départ :
print(montee - descente, altitudes[-1] - altitudes[0])  # 260 260
````

### Conseils supplémentaires

Deux boucles : la première, sur `visites.values()`, trouve le record, la seconde, sur `visites.items()`, affiche. Une chaîne multipliée par un entier se répète : `"#" * 3` vaut `"###"`, et `95 // 10` vaut `9`. Le format `:<10` aligne une chaîne à gauche sur 10 caractères.

```python
record = 0
for nombre in visites.values():
  if ...:
    record = nombre

for jour, nombre in visites.items():
  ligne = f"{jour:<10}{...} {nombre}"
  if ...:
    ligne += " <- record"
  print(ligne)
```

### Solution

````python
record = 0
for nombre in visites.values():
  if nombre > record:
    record = nombre

for jour, nombre in visites.items():
  ligne = f"{jour:<10}{'#' * (nombre // 10)} {nombre}"
  if nombre == record:
    ligne += " <- record"
  print(ligne)
````

Pour ne voir chaque paire qu'une fois, associez chaque équipe aux seules équipes qui la suivent dans la liste : avec `enumerate`, `equipes[i + 1:]` donne celles qui suivent l'équipe d'indice `i`. En aller-retour, chaque équipe rencontre toutes les autres, et il suffit d'écarter le cas où une équipe se rencontrerait elle-même. Avec 5 équipes, vous devez obtenir 10 matchs, puis 20.


### Solution

````python
numero = 0
for i, equipe_1 in enumerate(equipes):
  for equipe_2 in equipes[i + 1 :]:
    numero += 1
    print(f"Match {numero} : {equipe_1} – {equipe_2}")

nb_aller_retour = 0
for domicile in equipes:
  for exterieur in equipes:
    if domicile != exterieur:
      nb_aller_retour += 1
print(nb_aller_retour, "matchs en aller-retour")

# Chacune des n équipes reçoit les n - 1 autres : n(n - 1) matchs en
# aller-retour, et deux fois moins, n(n - 1) / 2, en matchs simples.
````

### Solution

````python
if mensualite <= capital * taux_mensuel:
  print("Mensualité trop faible : le prêt ne serait jamais remboursé")
else:
  reste_du = capital
  nb_mois = 0
  total_interets = 0
  while reste_du > 0:
    interets = reste_du * taux_mensuel
    reste_du += interets
    total_interets += interets
    if reste_du < mensualite:
      reste_du = 0
    else:
      reste_du -= mensualite
    nb_mois += 1
  print(f"Remboursé en {nb_mois} mois, intérêts : {total_interets:.2f} €")

# Avec 40 €, la mensualité paie tout juste les intérêts du mois (10 000 × 0,004) :
# le reste dû ne baisse jamais, et la boucle ne s'arrêterait pas.
````

### Conseils supplémentaires

Parcourez les heures possibles avec `range` ; `heure not in occupees` teste si une heure est libre. La clause `else` d'une boucle ne s'exécute que si la boucle est allée jusqu'au bout sans `break` : c'est là qu'on sait qu'aucune heure ne convenait. Une réunion de deux heures commence au plus tard à 15 h. Vous devez trouver 12 h pour une heure, et aucun créneau pour deux heures.

```python
for heure in range(...):
  if heure not in occupees:
    print(f"Première heure libre : {heure} h")
    ...
else:
  print("Aucun créneau libre")

for heure in range(...):
  if ...:
    print(f"Premier créneau de deux heures : {heure} h")
    break
else:
  print("Aucun créneau de deux heures")
```

### Solution

````python
for heure in range(9, 17):
  if heure not in occupees:
    print(f"Première heure libre : {heure} h")
    break
else:
  print("Aucun créneau libre")

for heure in range(9, 16):
  if heure not in occupees and heure + 1 not in occupees:
    print(f"Premier créneau de deux heures : {heure} h")
    break
else:
  print("Aucun créneau de deux heures")
````
