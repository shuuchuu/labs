### Conseils supplémentaires

`float("vingt")` lève une `ValueError` : c'est la classe à nommer dans l'`except`, et `as e` donne accès à l'erreur, que `{e!r}` affiche avec sa classe. ```python
def en_nombre(texte: str) -> float | None:
  try:
    return ...
  except ... as e:
    print(f"Valeur ignorée : {...}")

### Solution

### Conseils supplémentaires

Convertir les deux chaînes peut lever une `ValueError`, diviser par une valeur nulle une `ZeroDivisionError` : deux clauses `except`, chacune avec son message. `print(e)`, ou `{e}` dans une f-string, affiche le message de l'erreur `e`. Le `else` ne s'exécute que si le `try` est allé jusqu'au bout. Dans une f-string, la spécification `+.1f` affiche le signe et une décimale. Les instructions du `try` s'exécutent dans l'ordre : la première qui échoue choisit la clause `except`.

```python
def variation(avant: str, apres: str) -> None:
  try:
    debut = float(avant)
    fin = ...
    taux = ...
  except ValueError as e:
    print(f"Valeur non numérique : {e}")
  except ...:
    print(...)
  else:
    print(f"Variation : {taux:...} %")


variation("80", "92")
variation("80", "n/a")
variation("0", "5")
variation("0", "abc")  # Prédiction : ...
```

### Solution

````python
def variation(avant: str, apres: str) -> None:
  try:
    debut = float(avant)
    fin = float(apres)
    taux = (fin - debut) / debut * 100
  except ValueError as e:
    print(f"Valeur non numérique : {e}")
  except ZeroDivisionError:
    print("Variation indéfinie : valeur de départ nulle")
  else:
    print(f"Variation : {taux:+.1f} %")


variation("80", "92")
variation("80", "n/a")
variation("0", "5")
# float("abc") échoue avant la division : « Valeur non numérique »
variation("0", "abc")
````

### Conseils supplémentaires

`open` lève une `FileNotFoundError` quand le fichier n'existe pas. `fichier.readlines()` renvoie la liste des lignes, dont `len` donne le nombre. Le `finally` s'exécute même quand le `try` ou l'`except` sort de la fonction par un `return` : juste avant que la fonction ne rende sa valeur. Vous devez trouver 3 lignes dans `journal.txt`.

```python
def compter_lignes(chemin: str) -> int:
  try:
    with open(chemin, encoding="utf-8") as fichier:
      return ...
  except ...:
    print(...)
    return 0
  finally:
    print(...)


print(compter_lignes("journal.txt"))
print(compter_lignes("absent.txt"))
# Prédiction : ...
```

### Solution

````python
def compter_lignes(chemin: str) -> int:
  try:
    with open(chemin, encoding="utf-8") as fichier:
      return len(fichier.readlines())
  except FileNotFoundError:
    print(f"Fichier introuvable : {chemin}")
    return 0
  finally:
    print(f"Fin de la lecture de {chemin}")


# Le finally affiche son message avant que print n'affiche le résultat :
# Fin de la lecture de journal.txt, 3, Fichier introuvable : absent.txt,
# Fin de la lecture de absent.txt, 0
print(compter_lignes("journal.txt"))
print(compter_lignes("absent.txt"))
````

### Conseils supplémentaires

`ligne.split(";")` découpe la ligne à chaque `;` et renvoie la liste des champs. `raise ValueError(f"...")` lève l'erreur avec son message, et sort de la fonction. Pour l'essai, un `try` autour de l'appel, l'affichage de la mesure dans le `else`. Vous devez obtenir une mesure (`salon`) et trois lignes rejetées, chacune pour une raison différente.

```python
def lire_mesure(ligne: str) -> tuple[str, float]:
  champs = ligne.split(";")
  if ...:
    raise ValueError(f"{...} champs au lieu de 2")
  capteur, texte = champs
  valeur = float(texte)
  if ...:
    raise ...
  return capteur, valeur


for ligne in lignes_test:
  try:
    capteur, valeur = ...
  except ValueError as e:
    print(f"{ligne!r} rejetée : {e}")
  else:
    print(f"{capteur} : {valeur} °C")
```

### Solution

````python
def lire_mesure(ligne: str) -> tuple[str, float]:
  champs = ligne.split(";")
  if len(champs) != 2:
    raise ValueError(f"{len(champs)} champs au lieu de 2")
  capteur, texte = champs
  valeur = float(texte)  # lève ValueError si texte n'est pas un nombre
  if valeur < -40 or valeur > 60:
    raise ValueError(f"valeur hors limites : {valeur}")
  return capteur, valeur


for ligne in lignes_test:
  try:
    capteur, valeur = lire_mesure(ligne)
  except ValueError as e:
    print(f"{ligne!r} rejetée : {e}")
  else:
    print(f"{capteur} : {valeur} °C")
````

### Conseils supplémentaires

Une erreur personnalisée est une classe vide qui hérite d'`Exception` : sa docstring suffit comme corps. Dans `lire_mesure_verifiee`, un `try` autour de l'appel, et dans l'`except ValueError as e`, `raise ... from e` lève la nouvelle erreur avec `e` pour cause. `MesureInvalideError` hérite d'`Exception`, pas de `ValueError` : regardez la hiérarchie du cours.

```python
class MesureInvalideError(...):
  """..."""


def lire_mesure_verifiee(ligne: str) -> tuple[str, float]:
  try:
    return ...
  except ValueError as e:
    raise MesureInvalideError(f"ligne {ligne!r} invalide ({e})") from ...


try:
  lire_mesure_verifiee("cave;12;3")
except ... as e:
  print(f"Mesure refusée : {e}")

# Prédiction : ...
try:
  lire_mesure_verifiee("grenier;85")
except ValueError:
  print("Attrapée par except ValueError")
except ...:
  print(...)

# lire_mesure_verifiee("cave;12;3")
```

### Solution

````python
class MesureInvalideError(Exception):
  """Ligne du fichier de mesures que l'application ne peut pas lire."""


def lire_mesure_verifiee(ligne: str) -> tuple[str, float]:
  try:
    return lire_mesure(ligne)
  except ValueError as e:
    raise MesureInvalideError(f"ligne {ligne!r} invalide ({e})") from e


# 1. L'erreur de l'application, avec le message de sa cause
try:
  lire_mesure_verifiee("cave;12;3")
except MesureInvalideError as e:
  print(f"Mesure refusée : {e}")

# 2. MesureInvalideError hérite d'Exception, pas de ValueError :
# except ValueError ne l'attrape pas
try:
  lire_mesure_verifiee("grenier;85")
except ValueError:
  print("Attrapée par except ValueError")
except MesureInvalideError:
  print("Attrapée par except MesureInvalideError")

# 3. La trace montre la ValueError, puis la MesureInvalideError qu'elle a causée
# lire_mesure_verifiee("cave;12;3")
````
