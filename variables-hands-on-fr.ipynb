### Conseils supplémentaires

Les appels et les règles à utiliser :

- `nom.isidentifier()` renvoie `True` si la chaîne `nom` a la forme d'un nom : des lettres, des chiffres et des tirets bas, sans chiffre en tête. Elle renvoie aussi `True` pour un mot-clef comme `class`
- `keyword.iskeyword(nom)` (après `import keyword`) renvoie `True` pour un mot-clef : réservé par le langage, il ne peut pas nommer une variable
- `ma-variable` se lit `ma - variable` : une soustraction
- PEP 8 écrit les noms de variables en minuscules, les mots séparés par des tirets bas (`ma_variable`) ; un tiret bas en tête (`_x`) signale un nom à usage interne
- Python accepte les lettres accentuées, mais PEP 8 recommande des noms ASCII pour du code partagé

### Solution

````python
import keyword

# 2x : invalide, un nom ne commence pas par un chiffre
nom = "2x"
print(nom, nom.isidentifier() and not keyword.iskeyword(nom))
# _x : valide, et conforme à PEP 8 (le tiret bas en tête signale un nom interne)
nom = "_x"
print(nom, nom.isidentifier() and not keyword.iskeyword(nom))
# ma-variable : invalide, Python y lit la soustraction ma - variable
nom = "ma-variable"
print(nom, nom.isidentifier() and not keyword.iskeyword(nom))
# maVariable : valide, mais PEP 8 écrit ma_variable (minuscules, tirets bas)
nom = "maVariable"
print(nom, nom.isidentifier() and not keyword.iskeyword(nom))
# class : invalide, c'est un mot-clef (isidentifier seul ne le voit pas)
nom = "class"
print(nom, nom.isidentifier() and not keyword.iskeyword(nom))
# été : valide, mais PEP 8 recommande des noms ASCII pour du code partagé
nom = "été"
print(nom, nom.isidentifier() and not keyword.iskeyword(nom))
````

Les règles à appliquer :

- les priorités : `**` d'abord, puis `*`, `/`, `//` et `%`, puis `+` et `-` ; les parenthèses passent avant tout
- `//` arrondit vers le bas (vers moins l'infini), pas vers zéro, et le reste `%` a le signe du diviseur : `a == (a // b) * b + a % b`

### Conseils supplémentaires

Les règles à appliquer :

- les comparaisons s'enchaînent : `a < b < c` veut dire `a < b and b < c`
- deux chaînes se comparent caractère par caractère, selon leur code Unicode : les majuscules passent avant les minuscules (`"Z"` vaut 90, `"a"` 97), et les chiffres d'une chaîne se comparent comme des caractères, pas comme des nombres
- un `float` est stocké en binaire, où `0.1` n'a pas d'écriture exacte : les calculs gardent un minuscule écart. `math.isclose(a, b)` (après `import math`) compare avec une tolérance
- `or` renvoie son premier opérande vrai (le dernier s'il n'y en a aucun), `and` son premier opérande faux (le dernier s'ils sont tous vrais) : le résultat n'est pas forcément un `bool`. Sont faux : `0`, `0.0`, `""`, `None` et `False`
- `and` et `or` s'arrêtent dès que le résultat est connu : le second opérande n'est alors pas calculé
- `&` est le « et » bit à bit : 5 s'écrit `101` en binaire, 6 `110`

Un squelette à copier dans la cellule de l'exercice, en remplaçant chaque `...` :

```python
import math

print(1 < 2 < 3)
print(...)
...
print(math.isclose(..., ...))
```

### Solution

````python
import math

print(1 < 2 < 3)  # True : 1 < 2 and 2 < 3
print("abc" < "abd")  # True : "c" < "d"
print("Zebra" < "apple")  # True : "Z" (90) < "a" (97)
print("10" < "9")  # True : "1" < "9", des caractères, pas des nombres
print(0.1 + 0.2 == 0.3)  # False : 0.1 + 0.2 vaut 0.30000000000000004
print(math.isclose(0.1 + 0.2, 0.3))  # True
print("" or "défaut")  # défaut : "" est faux, or renvoie l'autre opérande
print("Alice" or "défaut")  # Alice : le premier opérande vrai
print(5 and 6)  # 6 : 5 est vrai, and renvoie le second opérande
print(5 & 6)  # 4 : 101 et 110 bit à bit donnent 100
print(0 and 1 / 0)  # 0 : and s'arrête sur 0, 1 / 0 n'est jamais calculé
````
