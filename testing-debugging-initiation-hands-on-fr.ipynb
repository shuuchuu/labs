# Déboguer et tester ses fonctions

Ce TP met en pratique le débogueur et les premiers tests vus en cours. Les trois premiers exercices donnent chacun une fonction qui contient une erreur de logique : le code s'exécute, mais il renvoie un mauvais résultat ou s'arrête sur une erreur. Le débogueur vous montre où le programme s'écarte de ce que vous attendiez, puis vous corrigez la fonction. Les deux exercices suivants écrivent des tests pour ces fonctions et les lancent avec pytest. Le dernier exercice est facultatif.

Dans Colab, le débogueur s'ouvre sous la cellule : tapez une commande dans le champ de saisie qui apparaît, puis Entrée. Les commandes du cours :

- `n` exécute la ligne suivante ;
- `s` entre dans la fonction que la ligne appelle ;
- `c` continue jusqu'au prochain arrêt ;
- `p expression` affiche la valeur d'une expression, par exemple `p i` ;
- `q` quitte le débogueur et arrête la cellule.

Tant que le débogueur attend une commande, la cellule tourne, et aucune autre cellule ne s'exécute : terminez toujours par `c` jusqu'à la fin, ou par `q`. Après `q`, une trace `BdbQuit` s'affiche sous la cellule : elle est normale, ce n'est pas une erreur de votre code ; `c` finit la cellule sans elle. Avant de passer à l'exercice suivant, remettez le `#` devant chaque `breakpoint()`, pour que « Tout exécuter » ne s'y arrête pas.

## Entrer dans une fonction appelée

Une papeterie calcule le prix d'un panier avec deux fonctions : `prix_ligne` donne le prix d'un article commandé en plusieurs exemplaires, et `prix_total` additionne les lignes du panier. La règle : à partir de 10 exemplaires d'un même article, chacun coûte 1 € de moins. Le panier de la cellule suivante, 10 stylos à 2 € et 2 cahiers à 3 €, doit donc coûter 10 × 1 + 2 × 3 = 16 €. Exécutez la cellule : elle affiche autre chose.

1. Retirez le `#` devant `breakpoint()`, au début de `prix_total`, puis exécutez la cellule. Avancez avec `n` jusqu'à la ligne qui appelle `prix_ligne`. Là, `n` exécuterait `prix_ligne` d'un coup : entrez-y plutôt avec `s`. Avancez dans `prix_ligne` avec `n`, en affichant `quantite` et `prix_unitaire` avec `p` : la remise est-elle appliquée ? Quittez avec `q`, et notez en commentaire la ligne fautive.
2. Remettez le `#` devant `breakpoint()`, puis écrivez dans la cellule d'exercice une version corrigée de `prix_ligne`, et vérifiez que `prix_total` affiche 16.

````python
tarifs = {"stylo": 2, "cahier": 3, "classeur": 5}


def prix_ligne(article: str, quantite: int, tarifs: dict[str, int]) -> int:
  """Renvoie le prix de quantite exemplaires de article."""
  prix_unitaire = tarifs[article]
  if quantite > 10:
    prix_unitaire = prix_unitaire - 1
  return prix_unitaire * quantite


def prix_total(panier: dict[str, int], tarifs: dict[str, int]) -> int:
  """Renvoie le prix du panier, qui associe à chaque article sa quantité."""
  # breakpoint()
  total = 0
  for article, quantite in panier.items():
    total = total + prix_ligne(article, quantite, tarifs)
  return total


print(prix_total({"stylo": 10, "cahier": 2}, tarifs))  # attendu : 16
````

````python
%%writefile test_serie.py
def plus_longue_serie(resultats: list[str]) -> int:
  """Renvoie la longueur de la plus longue suite de victoires ("V") de resultats."""
  meilleure = 0
  serie = 0
  for resultat in resultats:
    if resultat == "V":
      serie = serie + 1
      if serie > meilleure:
        meilleure = serie
    else:
      serie = 0
  return meilleure


def test_plus_longue_serie_exemple():
  assert plus_longue_serie(["V", "V", "D", "V", "N"]) == 2


def test_plus_longue_serie_liste_vide():
  assert plus_longue_serie([]) == 0


def test_plus_longue_serie_a_la_fin():
  assert plus_longue_serie(["D", "V", "V", "V"]) == 3
````
