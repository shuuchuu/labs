## Set Manipulation

Given the following sets of programming language features (you can consult the [help page](https://docs.python.org/3/library/stdtypes.html#set-types-set-frozenset) for sets during the exercise):

- Find the common features between JavaScript and Python
- Find the features exclusive to Java compared to Python
- What is the complete set of features?
- What are the features exclusive to Python compared to all other languages?

```
js = {"dynamic typing", "interpreted", "native to the web", "prototype-based"}
python = {"dynamic typing", "interpreted", "scientific computing", "object-oriented"}
java = {"compiled", "object-oriented", "static typing"}

languages = [js, python, java]
```

```
# Your code here
```

### Solution

```
print("Common features between JS and Python:",
      js & python)
print("Features exclusive to Java compared to Python:",
      java - python)
print("Complete set of features:",
      set().union(*languages))
print("Features exclusive to Python compared to others:",
      python.difference(js, java))
```

## Which Person to Choose?

You want to hire someone for a mission in C++ and Python. You have the following information available (in the form of a dictionary with a name as the key and skills as the value).

Write the function `filter_people` that returns the names of the competent people.

```
skills = {'Alice': {'SAS', 'Python', 'R'},
          'Bob': {'HTML', 'JavaScript', 'CSS'},
          'Laura': {'C++', 'ASM', 'Python', 'CUDA'}}
mission = {'C++', 'Python'}


def filter_people(skills_directory: dict[str, set[str]],
                  mission: set[str]
                  ) -> set[str]:
  pass  # Your code here
```

### Solution

```
def filter_people(skills_directory: dict[str, set[str]],
                  mission: set[str]
                  ) -> set[str]:
  return {name
          for name, skills in skills_directory.items()
          if mission <= skills}


filter_people(skills, mission)
```

## Comparing DNA Strands with Hamming Distance

The [Hamming distance](https://en.wikipedia.org/wiki/Hamming_distance) is the total number of differences in two sequences of the same length:

$$d(a, b) = \#\big\{i : a_i \neq b_i\big\}$$

Implement a function to calculate this distance between the two given DNA strands.

```
from collections.abc import Sequence


strand_a = ["A", "C", "G", "A", "A", "T", "G", "A"]
strand_b = ["T", "C", "A", "A", "G", "C", "T", "A"]


def hamming_distance(strand_a: Sequence[str], strand_b: Sequence[str]) -> int:
  pass  # Your code here
```

### Solution

```
from collections.abc import Sequence


strand_a = ["A", "C", "G", "A", "A", "T", "G", "A"]
strand_b = ["T", "C", "A", "A", "G", "C", "T", "A"]


def hamming_distance(a: Sequence[str], b: Sequence[str]) -> int:
  if len(a) != len(b):
    raise ValueError("The two sequences are not equal in length.")
  return sum(1 for i, j in zip(a, b) if i != j)


hamming_distance(strand_a, strand_b)
```
