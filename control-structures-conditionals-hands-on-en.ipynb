## Complementary Sequence of a DNA Strand

The list below represents the sequence of a DNA strand:

```python
["A","C","G","T","T","A","G","C","T","A","A","C","G"]
```

Write a script that turns this sequence into its complementary sequence.

The complementary sequence is obtained by replacing `A` with `T`, `T` with `A`, `C` with `G` and `G` with `C`.

````
# Your code here
````

### Solution

````
dna_strand = ["A","C","G","T","T","A","G","C","T","A","A","C","G"]

complementary_dna_strand = []
for base in dna_strand:
    if base == "A":
        complementary_dna_strand.append("T")
    elif base == "T":
        complementary_dna_strand.append("A")
    elif base == "C":
        complementary_dna_strand.append("G")
    else:
        complementary_dna_strand.append("C")

print(dna_strand)
print(complementary_dna_strand)
````

````
# With a dictionary, without if
replacement = {
    "A": "T",
    "T": "A",
    "C": "G",
    "G": "C"
}
complementary_dna_strand = [replacement[x] for x in dna_strand]
print(dna_strand)
print(complementary_dna_strand)
````

## Minimum of a List

````
# Your code here
````

### Solution

````
values = [8, 4, 6, 1, 5]

minimum = None
for element in values:
    if minimum is None:
        minimum = element
    elif element < minimum:
        minimum = element

print(minimum)
````

## Amino Acid Frequency

The list below represents a sequence of amino acids:

```python
["A","R","A","W","W","A","W","A","R","W","W","R","A","G"]
```

Compute the frequency of the amino acids alanine (`A`), arginine (`R`), tryptophan (`W`) and glycine (`G`) in this sequence.

````
# Your code here
````

### Solution

````
amino_acids = ["A","R","A","W","W","A","W","A","R","W","W","R","A","G"]

count_a = 0
count_r = 0
count_w = 0
count_g = 0

for amino_acid in amino_acids:
    if amino_acid == "A":
        count_a += 1
    elif amino_acid == "R":
        count_r += 1
    elif amino_acid == "W":
        count_w += 1
    elif amino_acid == "G":
        count_g += 1

print("Alanine frequency:", count_a / len(amino_acids))
print("Arginine frequency:", count_r / len(amino_acids))
print("Tryptophan frequency:", count_w / len(amino_acids))
print("Glycine frequency:", count_g / len(amino_acids))

# We could also use the lists' count method:
print()
print("With .count:")
print("Alanine frequency:",
      amino_acids.count("A") / len(amino_acids))
print("Arginine frequency:",
      amino_acids.count("R") / len(amino_acids))
print("Tryptophan frequency:",
      amino_acids.count("W") / len(amino_acids))
print("Glycine frequency:",
      amino_acids.count("G") / len(amino_acids))
````

````
# With a defaultdict
from collections import defaultdict

name= {
        "A":"Alanine" ,
        "R":"Arginine" ,
        "W":"Tryptophan" ,
        "G":"Glycine"
        }

counts = defaultdict(lambda : 0)
for amino_acid in amino_acids:
    counts[amino_acid] += 1
for k, v in counts.items():
    print(f"{name[k]} frequency:", v / len(amino_acids))
````

## A Student's Grades and Honors

Here are a student's grades (out of 20): 14, 9, 13, 15 and 12. Write a script that displays the highest grade (use the [`max`](https://docs.python.org/3/library/functions.html#max) function), the lowest grade (use the [`min`](https://docs.python.org/3/library/functions.html#min) function) and computes the average.

Display the average with two decimals. Also display the honors obtained, knowing that the honors are "pass" if the average is between 10 inclusive and 12 exclusive, "fairly good" between 12 inclusive and 14 exclusive, and "good" from 14 up.

````
# Your code here
````

### Solution

````
grades = [14, 9, 13, 15, 12]

average = sum(grades) / len(grades)

if 10 <= average < 12:
    honors = "pass"
elif 12 <= average < 14:
    honors = "fairly good"
elif 14 <= average:
    honors = "good"
else:
    honors = "none"

print("Minimum:", min(grades))
print("Maximum:", max(grades))
print("Average: {:.2f}".format(average))
print("Honors:", honors)
````

## The Collatz Conjecture

The [Collatz conjecture](https://en.wikipedia.org/wiki/Collatz_conjecture) (also known as the Syracuse conjecture) is a mathematical conjecture that remains unproven to this day, defined as follows.

Take a positive integer $n$. If $n$ is even, divide it by 2. If it's odd, multiply it by 3 and add 1. By repeating this procedure, the sequence of numbers reaches the value 1, then continues indefinitely with a sequence of three trivial values called the trivial cycle.

So far, the Collatz conjecture, which states that from any positive integer the sequence reaches 1, has never been disproven.

For instance, the first elements of the sequence starting from 10 are: 10, 5, 16, 8, 4, 2, 1...

Write a script that, starting from a positive integer $n$ (for instance 10 or 20), builds a list of the numbers of the Collatz sequence. With different starting points (that is, with different values of $n$), does the conjecture always hold? Which numbers make up the trivial cycle?

Notes:

1. For this exercise, you need an unknown number of iterations for the sequence to reach 1 and then start its trivial cycle. You can test your algorithm with an arbitrary number of iterations, typically 20 or 100, depending on your starting number n.
1. A number is even when the remainder of its integer division (modulo operator `%`) by 2 is zero.

````
# Your code here
````

### Solution

````
n = 10
max_iterations = 1000

ns = [n]
iterations = 0
while True:
    if n == 1:
         print(f"n = {n}: convergence in {iterations} iterations")
         break
    if iterations == max_iterations:
        print(f"n = {n}: no convergence in {max_iterations} iterations")
        break
    if n % 2 == 0:
        n //= 2
    else:
        n = n * 3 + 1
    ns.append(n)
    iterations += 1
print(ns)
````

## Finding the Prime Numbers Below 100

Here's an excerpt from the [article on prime numbers](https://en.wikipedia.org/wiki/Prime_number) of the online encyclopedia Wikipedia.

A prime number is a natural number that has exactly two distinct positive integer divisors (which are then $1$ and itself). This definition excludes $1$, which has only one positive integer divisor. Conversely, a non-zero number that is the product of two integers other than $1$ is called composite. For instance, $6 = 2 \times 3$ is composite, as is $21 = 3 \times 7$, but $11$ is prime since $1$ and $11$ are the only divisors of $11$. The numbers $0$ and $1$ are neither prime nor composite.

Find the prime numbers below $100$. How many prime numbers are there between $0$ and $100$? To help you, we suggest several methods.

#### Method 1 (not very efficient but fairly intuitive)

For each number from $2$ to $100$, compute the remainder of the integer division (with the modulo operator `%`) by every number from $2$ up to itself minus $1$. If it's a prime number, no number will give a remainder of $0$.

#### Method 2 (more efficient and faster, but a bit more complicated)

Go through all the numbers from $2$ to $100$ and check whether they are composite, that is, divisible by a prime number. In practice, this means checking whether the remainder of the integer division (modulo operator `%`) of the number by each prime number found so far is zero. If so, the number isn't prime. Careful: for this method, you'll need to initialize the list of prime numbers with the first prime number (so $2$!).

````
# Your code here
````

### Solution

````
# Method 1
primes = []
for n in range(2, 101):
    prime = True
    for m in range(2, n):
        if n % m == 0:
            prime = False
            break
    if prime:
        primes.append(n)
print(primes)

# Method 1 with the for ... else ... form
primes = []
for n in range(2, 101):
    for m in range(2, n):
        if n % m == 0:
            break
    else:
        primes.append(n)
print(primes)

# Method 2
primes = []
for n in range(2, 101):
    for prime in primes:
        if n % prime == 0:
            break
    else:
        primes.append(n)
print(primes)
````

## Finding a Number by Bisection

A [bisection](https://en.wikipedia.org/wiki/Bisection_method) search is a method that consists in dividing a problem (usually into equal parts) to find its solution. As an example, here's a conversation between Peter and Patrick in which Peter tries to guess the number (between 1 and 100 inclusive) Patrick thought of.

- [Patrick] "Alright, I've thought of a number between 1 and 100."
- [Peter] "OK, I'll try to guess it. Is your number smaller or greater than 50?"
- [Patrick] "Greater."
- [Peter] "Is your number smaller than, greater than or equal to 75?"
- [Patrick] "Greater."
- [Peter] "Is your number smaller than, greater than or equal to 88?"
- [Patrick] "Smaller."
- [Peter] "Is your number smaller than, greater than or equal to 81?"
- [Patrick] "Smaller."
- [Peter] "Is your number smaller than, greater than or equal to 78?"
- [Patrick] "Greater."
- [Peter] "Is your number smaller than, greater than or equal to 79?"
- [Patrick] "Equal. That's the number I thought of. Well done!"

To guess the number quickly, the trick is to take half of the interval containing the number each time. Here are the details of the different steps:

- the number is between 1 and 100, we suggest 50 (1 + (100 - 1) // 2).
- the number is greater than 50, so it's between 51 and 100, we suggest 75 (51 + (100 - 51) // 2).
- the number is between 76 and 100, we suggest 88 (76 + (100 - 76) // 2).
- the number is smaller than 88, so it's between 76 and 87, we suggest 81 (76 + (87 - 76) // 2).
- the number is between 76 and 80, we suggest 78 (76 + (80 - 76) // 2).
- the number is between 79 and 80, we suggest 79 (79 + (80 - 79) // 2).

Write a script that reproduces this guessing game. You think of a number between 1 and 100 and the computer tries to guess it by bisection, asking you questions.

Your program will use the [`input`](https://docs.python.org/3/library/functions.html#input) function to interact with the user. Here's an example of how it works:

```python
>>> letter = input("Enter a letter: ")
Enter a letter: P
>>> print(letter)
P
```

To guide you, here's what the program would give with the previous conversation:

```default
Think of a number between 1 and 100.
Is your number greater than, smaller than or equal to 50? [+/-/=] +
Is your number greater than, smaller than or equal to 75? [+/-/=] +
Is your number greater than, smaller than or equal to 88? [+/-/=] -
Is your number greater than, smaller than or equal to 81? [+/-/=] -
Is your number greater than, smaller than or equal to 78? [+/-/=] +
Is your number greater than, smaller than or equal to 79? [+/-/=] =
Found it in 6 questions!
```

The characters `[+/-/=]` tell the user how to interact with the computer: enter either the `+` character if the chosen number is greater than the one the computer suggested, the `-` character if the chosen number is smaller than the one the computer suggested, or the `=` character if the chosen number is the one the computer suggested (then press *Enter*).

````
# Your code here
````

### Solution

````
print("Think of a number between 1 and 100.")

low = 1
high = 100
iterations = 1
while True:
    middle = low + (high - low) // 2
    hint = input("Is your number greater than, smaller than or equal "
                 f"to {middle}? [+/-/=] ")
    if hint == "+":
        low = middle + 1
    elif hint == "-":
        high = middle - 1
    elif hint == "=":
        break
    iterations += 1

print(f"Found it in {iterations} questions!")
````
