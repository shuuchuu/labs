### Additional hints

A `case ["add", item]:` pattern matches a list of two words whose first one is `"add"`, and stores the second one in `item`; `case _:` catches everything else, including `add some tea` (three words). In a `case`, `break` leaves the `for` loop. `list.remove(x)` removes the first occurrence of `x`. The first `show` should print `['bread', 'bread']`, the second one and the final cart `['coffee']`.

```python
cart = []
for command in commands:
  match command.split():
    case ["add", item]:
      cart.append(item)
    case ["remove", item]:
      if ...:
        ...
      else:
        print(f"{item} is not in the cart")
    case ["show"]:
      print(cart)
    case ["clear"]:
      ...
    case ["quit"]:
      ...
    case _:
      ...
print("Final cart:", cart)
```

### Solution

````python
cart = []
for command in commands:
  match command.split():
    case ["add", item]:
      cart.append(item)
    case ["remove", item]:
      if item in cart:
        cart.remove(item)
      else:
        print(f"{item} is not in the cart")
    case ["show"]:
      print(cart)
    case ["clear"]:
      cart = []
    case ["quit"]:
      break
    case _:
      print(f"Unknown command: {command}")
print("Final cart:", cart)

# The remove case can also be written with a guard, an `if` in the pattern:
#     case ["remove", item] if item in cart:
#       cart.remove(item)
#     case ["remove", item]:
#       print(f"{item} is not in the cart")
````

## Finding the Faulty Version by Bisection

A piece of software has had 200 versions. A bug appeared in one of them and has never been fixed since: the `tests_pass` list gives, for each version (from the oldest, at index 0, to the most recent), `True` if its tests pass. Running a version's tests takes ten minutes: we want to run as few of them as possible.

1. Find the index of the first buggy version by going through the versions in order, and count how many versions you checked.
2. Find it by bisection, with a `while` loop: since all the buggy versions come after all the others, checking the version in the middle of a range tells which half to search. Count the checked versions again.

This is what `git bisect` does on a repository's history.

````python
tests_pass = [True] * 137 + [False] * 63
````

````python
# Your code here
````

### Additional hints

In order: `enumerate`, and `break` at the first failing version. Bisection: keep two bounds `low` and `high` such that the first buggy version is always between them (at first, the first and the last version, which is buggy). As long as `low < high`, check `middle = (low + high) // 2`: if its tests pass, the bug is after it, otherwise it is at `middle` or before. You should find version 137 both ways, in 138 checks, then in 7.

```python
n_checks = 0
for version, ok in enumerate(tests_pass):
  n_checks += 1
  if ...:
    first_buggy = version
    break
print(f"In order: version {first_buggy}, {n_checks} checks")

low = 0
high = len(tests_pass) - 1
n_checks = 0
while low < high:
  middle = (low + high) // 2
  n_checks += 1
  if tests_pass[middle]:
    low = ...
  else:
    high = ...
print(f"By bisection: version {low}, {n_checks} checks")
```

### Solution

````python
n_checks = 0
for version, ok in enumerate(tests_pass):
  n_checks += 1
  if not ok:
    first_buggy = version
    break
print(f"In order: version {first_buggy}, {n_checks} checks")

low = 0
high = len(tests_pass) - 1
n_checks = 0
while low < high:
  middle = (low + high) // 2
  n_checks += 1
  if tests_pass[middle]:
    low = middle + 1
  else:
    high = middle
print(f"By bisection: version {low}, {n_checks} checks")

# Each check halves the range: 200 versions take at most 8 checks
# (2 ** 8 = 256), against 138 here in order.
````
