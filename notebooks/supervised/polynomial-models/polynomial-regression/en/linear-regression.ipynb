### Parameters Estimation

In order to estimate $\theta_0$ et $\theta_1$, we need 2 elements :

- a cost function, which will tell us for given parameters $\theta_0$ and $\theta_1$ telling us whether we are going good or bad
- a method for estimating the best parameters given this cost function

### Cost function

The most commonly used cost function is the average of the differences between the actual points and those obtained by the hypothesis with the current parameters, squared.

Denoting $L$ the cost function, $n$ the number of training examples, we can write :

$$
  L(\theta_0, \theta_1) = \sum\frac{(\mathbf{X}\theta_1 + \theta_0 - \mathbf{Y})^2}{2n}
$$

#### Solution

```python
def residuals(
  X: numpy.ndarray,
  Y: numpy.ndarray,
  theta_0: float,
  theta_1: float,
) -> numpy.ndarray:
  return predict(X, theta_0, theta_1) - Y


print("Residuals:", residuals(X_example, Y_example, theta_0_example, theta_1_example))


def cost(
  X: numpy.ndarray,
  Y: numpy.ndarray,
  theta_0: float,
  theta_1: float,
) -> numpy.float32:
  return numpy.mean(residuals(X=X, Y=Y, theta_0=theta_0, theta_1=theta_1) ** 2) / 2


print("Costs:", cost(X_example, Y_example, theta_0_example, theta_1_example))
```

### Parameters Optimization

Optimising the parameters means minimising the cost function:

$$\min_{\theta} \sum\frac{(\theta_0 + \mathbf{X}\theta_1 - \mathbf{Y})^2}{2n}$$

To do this, there is a direct formula that we will not use in this tutorial for two reasons: it is not applicable to very large datasets and the method we will use can be reused for logistic regression and other techniques (general neural networks, gradient boosted trees, etc).

We will use gradient descent. This method is iterative and takes successive steps by following the derivative (to maximise) or its opposite (to minimise) the loss function. In the case of linear regression, it converges to the global optimum (the best solution).

In pseudo-code, the algorithm is (with the correct gradients calculated):

$$
\begin{aligned}
& \text{until convergence :} \\
& \quad \theta_0 \leftarrow \theta_0 - \alpha \sum\frac{\theta_0 + \mathbf{x_i}\theta_1 - \mathbf{y_i}}{n} \\
& \quad \theta_1 \leftarrow \theta_1 - \alpha \frac{\mathbf{X}^T(\theta_0 + \mathbf{X}\theta_1 - \mathbf{Y})}{n} \\
\end{aligned}
$$

where $\alpha$ is the learning rate.


*Implement the `gradient` function with the help of the `residuals` function.*

```python
def gradient(
  X: numpy.ndarray, Y: numpy.ndarray, theta_0: float, theta_1: float
) -> tuple[float, float]:
  pass  # your code here
```

#### Solution

```python
def gradient(
  X: numpy.ndarray, Y: numpy.ndarray, theta_0: float, theta_1: float
) -> tuple[float, float]:
  r = residuals(X=X, Y=Y, theta_1=theta_1, theta_0=theta_0)
  g_0 = r.mean()
  g_1 = X.T.dot(r) / Y.size
  return g_0, g_1[0, 0]
```

### Gradient descent

*Implement gradient descent in a function `gradient_descent`.*

```python
def gradient_descent(
  X: numpy.ndarray, Y: numpy.ndarray, alpha: float, nb_iter: int, epsilon: float
) -> tuple[float, float, list[float]]:
  pass  # your code here
```

#### Solution

```python
def gradient_descent(
  X: numpy.ndarray,
  Y: numpy.ndarray,
  alpha: float = 0.1,
  nb_iter: int = 1000,
  epsilon: float = 10e-6,
) -> tuple[float, float, list[float]]:
  rng = numpy.random.default_rng()
  theta_0, theta_1 = rng.normal(size=(2,))
  print(f"Initial (random) paramters : θ₀ = {theta_0:.2f}, θ₁ = {theta_1:.2f}")
  initial_cost = cost(X, Y, theta_0, theta_1)
  print(f"Initial Cost : {initial_cost:.2f}")
  costs = [initial_cost]
  for i in range(nb_iter):
    g_0, g_1 = gradient(X, Y, theta_0, theta_1)
    theta_0 -= alpha * g_0
    theta_1 -= alpha * g_1
    costs.append(cost(X, Y, theta_0, theta_1))
    if costs[-2] - costs[-1] < epsilon:
      print(f"Stopping the procedure: not enough progress in the iteration {i}")
      break
  print(f"Final parameters : θ₀ = {theta_0:.2f}, θ₁ = {theta_1:.2f}")
  print(f"Final cost : {costs[-1]:.2f}")
  return theta_0, theta_1, costs
```

To finish, the call to the gradient_descent function

```python
theta_0, theta_1, costs = gradient_descent(X, Y, 0.1, 1000)
print(f"Coûts successifs : {', '.join(map(str, costs))}")
```

## Multivariate Linear Regression

To continue this practical work, we will repeat a full pre-processing run on the entire AMES data (not just the `GrLiveArea` column).

```python
def preprocess(train_file: str, test_file: str) -> numpy.ndarray:
  train_X = pandas.read_csv(train_file, index_col="Id")
  test_X = pandas.read_csv(test_file, index_col="Id")

  train_Y = train_X[["SalePrice"]]
  train_X = train_X.drop(columns=["SalePrice"])

  all_X = pandas.concat([train_X, test_X])

  # Fill with median
  cols_1 = ["LotFrontage"]
  all_X[cols_1] = all_X[cols_1].fillna(train_X[cols_1].median())

  # Fill with mode
  cols_2 = [
    "MSZoning",
    "Electrical",
    "KitchenQual",
    "Exterior1st",
    "Exterior2nd",
    "SaleType",
    "Utilities",
  ]
  all_X[cols_2] = all_X[cols_2].fillna(train_X[cols_2].mode().iloc[0, :])

  # Fill with 0
  cols_4 = [
    "GarageYrBlt",
    "GarageArea",
    "GarageCars",
    "BsmtFinSF1",
    "BsmtFinSF2",
    "BsmtFullBath",
    "BsmtHalfBath",
    "BsmtUnfSF",
    "MasVnrArea",
    "TotalBsmtSF",
  ]
  all_X[cols_4] = all_X[cols_4].fillna(0)

  # Other fills
  cols_5 = ["Functional"]
  all_X[cols_5] = all_X[cols_5].fillna("Typ")

  # Anything else is set to the String "NA"
  all_X = all_X.fillna("NA")

  # Numeric coded categories need to be cast as String to be interpreted as
  # categories and not numerical values.
  cols_numerical2label = ["MSSubClass"]
  all_X[cols_numerical2label] = all_X[cols_numerical2label].astype(str)

  quality_mapping = dict(NA=0, Po=1, Fa=2, TA=3, Gd=4, Ex=5)
  quality_columns = [
    "BsmtCond",
    "BsmtQual",
    "ExterCond",
    "ExterQual",
    "FireplaceQu",
    "GarageCond",
    "GarageQual",
    "HeatingQC",
    "KitchenQual",
    "PoolQC",
  ]
  street_mapping = dict(NA=0, Grvl=1, Pave=2)
  bsmt_fin_mapping = dict(NA=0, Unf=1, LwQ=2, Rec=3, BLQ=4, ALQ=5, GLQ=6)

  replace_mapping = dict(
    Alley=street_mapping,
    BsmtExposure=dict(NA=0, No=1, Mn=2, Av=3, Gd=4),
    BsmtFinType1=bsmt_fin_mapping,
    BsmtFinType2=bsmt_fin_mapping,
    Functional=dict(Sal=1, Sev=2, Maj2=3, Maj1=4, Mod=5, Min2=6, Min1=7, Typ=8),
    LandSlope=dict(Sev=1, Mod=2, Gtl=3),
    LotShape=dict(IR3=1, IR2=2, IR1=3, Reg=4),
    PavedDrive=dict(NA=0, N=1, P=2, Y=3),
    Street=dict(Grvl=1, Pave=2),
    Utilities=dict(ELO=1, NoSeWa=2, NoSewr=3, AllPub=4),
  )

  for quality_column in quality_columns:
    replace_mapping[quality_column] = quality_mapping

  all_X.replace(replace_mapping, inplace=True)

  print(f"# of NAs : {all_X.isnull().sum().sum()}")

  dummies = pandas.get_dummies(all_X)
  return (
    dummies.iloc[: train_X.shape[0], :].values,
    train_Y.values,
    dummies.iloc[train_X.shape[0] :, :].values,
    dummies.columns,
  )
```

```python
X_train, Y_train, X_test, columns = preprocess(
  "dataset-ames/train.csv", "dataset-ames/test.csv"
)
```
