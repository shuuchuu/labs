## Processing of a document

```
import en_core_web_md
import en_core_web_sm
import fr_core_news_sm


# Creation of the main spacy object
nlp = en_core_web_md.load()
# nlp = en_core_web_sm.load()
# nlp = fr_core_news_sm.load()
```

```
# Test text
text = ("Barack Obama was the 44th president of the United States. "
        "John ate a bad apple. "
        "a sentence with no uppercase first letter and no punctuation")

# Run spacy on our text
doc = nlp(text)
```

## Tokenization

```
for token in doc:
    print(token.text)
```

## Retrieving sentences

```
for sent in doc.sents:
  print("==================")
  print(sent)
```

## Token details and POS tags

```
for sent in doc.sents:
  print("=============")
  for token in sent:
    print(f"{token.text:15} {token.idx:<3} {token.pos_:5} {token.tag_}")
```

## Lemmatization

```
for sent in doc.sents:
  print("=============")
  for token in sent:
    print(f"{token.text:15} {token.lemma_}")
```

## Syntactic parsing

- `text` : original text
- `dep_` : syntactical relation between the token and its parent in the graph
- `head.text` : original text of the parent
- `head.pos_` : POS of the parent
- `children` : list of children

```
for token in doc:
  print(f"{token.text:15} {token.dep_:10} {token.head.text:15} "
        f"{token.head.pos_:5} {' '.join(map(str, token.children))}")
```

## Rendering of the syntactic parse

```
spacy.displacy.render(doc,
                      style="dep",
                      jupyter=True,
                      options=dict(distance=100))
```
