## Traitement d'un document

```
import fr_core_news_md
import fr_core_news_sm


# Création de l'objet qui va traiter des documents
nlp_fr = fr_core_news_md.load()
# nlp_fr = fr_core_news_sm.load()
```

```
# On crée un texte composé de plusieurs phrases
text_fr = ("Barack Obama fut le 44ème président des États-Unis. "
           "Jean mange une pomme acidulée. "
           "une phrase sans majuscule ni verbe ni point")

# On passe le texte par le pipeline
doc_fr = nlp_fr(text_fr)
```

## Tokenisation

Affichage des tokens

```
for token in doc_fr:
    print(token.text)
```

## Affichage des phrases

```
# On affiche les phrases
for sent in doc_fr.sents:
  print("==================")
  print(sent)
```

## Référence au texte d'origine et étiquettes morpho-syntaxiques

```
# On affiche les token de chaque phrase, l'index de départ dans le texte
# d'origine, ainsi que les POS-tags
for sent in doc_fr.sents:
  print("=============")
  for token in sent:
    print(f"{token.text:15} {token.idx:<3} {token.pos_:5} {token.tag_}")  
```

## Lemmatisation


```
for sent in doc_fr.sents:
  print("=============")
  for token in sent:
    print(f"{token.text:15} {token.lemma_}")
```

## Dépendances syntaxiques

- `text` : texte original.
- `dep_` : relation syntaxique entre le token et son prédécesseur dans le graphe.
- `head.text` : texte original du prédécesseur.
- `head.pos_` : étiquette morpho-syntaxique du prédécesseur.
- `children` : liste des fils du token.

```
for token in doc_fr:
  print(f"{token.text:15} {token.dep_:10} {token.head.text:15} "
        f"{token.head.pos_:5} {' '.join(map(str, token.children))}")
```

## Un résumé lisible

```
spacy.displacy.render(doc_fr,
                      style="dep",
                      jupyter=True,
                      options=dict(distance=100))
```
