## Conversion de types

Il est possible de passer d'un type à l'autre facilement. Vous pouvez utiliser les classes que vous venez de découvrir pour convertir un type en un autre type.

- *Utilisez les fonctions [`int`](https://docs.python.org/fr/3/library/functions.html#int), [`float`](https://docs.python.org/fr/3/library/functions.html#float), etc pour convertir des variables en le type souhaité.*

- *Que se passe-t-il quand la conversion échoue ?*

````
# Votre code ici
````

### Solution

````
# Vers entier
print("Entier   depuis '42'  ", int("42"))
print("Entier   depuis 42.6  ", int(42.6))
print("Entier   depuis True  ", int(True))

# Vers flottant
print("Flottant depuis '42'  ", float("42"))
print("Flottant depuis 42    ", float(42))
print("Flottant depuis True  ", float(True))

# Vers booléen
print("Booléen  depuis ''    ", bool(""))
print("Booléen  depuis 'a'   ", bool("a"))
print("Booléen  depuis 0     ", bool(0))
print("Booléen  depuis 1     ", bool(1))

# Vers chaîne de caractère
print("Chaîne   depuis 42    ", str(42))
print("Chaîne   depuis True  ", str(True))

# Quant la conversion n'est pas possible, une ValueError est lancée
int("a")
````

## Prédiction du résultat d'opérations et de conversions de types

Sans exécuter de code, essayez de prédire le résultat de ces opérations :

1. `str(4) * int("3")`
2. `int("3") * str(4)`
3. `int("3") + float("3.2")`
4. `str(4) * float(3)`
5. `str(3 / 4) * 2`
6. `str(42) * bool(42)`
7. `bool(0) * str(42)`

Votre réponse ici.

1.
2.
3.
4.
5.
6.
7.

### Solution

````
print(str(4) * int("3"))
print(int("3") * str(4))
print(int("3") + float("3.2"))
# print(str(4) * float(3)) produirait :
# TypeError: can't multiply sequence by non-int of type 'float'
print(str(3 / 4) * 2)
print(str(42) * bool(42))
print(bool(0) * str(42))
````
