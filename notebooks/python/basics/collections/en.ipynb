## Seasons

Predict what those expressions evaluate to:

- `seasons[2]`
- `seasons[1][0]`
- `seasons[1:2]`
- `seasons[:][1]`. How do you explain this last result?

```python
winter = ["December", "January", "February"]
spring = ["March", "April", "May"]
summer = ["June", "July", "August"]
autumn = ["September", "October", "November"]

seasons = [winter, spring, summer, autumn]
```

### Solution

```python
winter = ["December", "January", "February"]
spring = ["March", "April", "May"]
summer = ["June", "July", "August"]
autumn = ["September", "October", "November"]

seasons = [winter, spring, summer, autumn]

print(seasons[2])
print(seasons[1][0])
print(seasons[1:2])
print(seasons[:][1])
# Explanation of the last expression: a full slice returns a copy of the list,
# so it returns the same element as seasons[1] (the list of the second season,
# ["March", "April", "May"])
```

```python
js = {"dynamic typing", "interpreted", "web native", "prototype oriented"}
python = {"dynamic typing", "interpreted", "scientific computing", "object oriented"}
java = {"compiled", "object oriented", "static typing"}

langages = [js, python, java]
```

```python
# Your code here
```

What happens if you try to modify one of these tuples?

```python
# Your code here
```

*Your answer here*

### Solution

```python
week = ("monday", "tuesday", "wednesday", "thursday", "friday", "saturday", "sunday")
print(week[-1], week[:5], week[::-1])

print(tuple(range(9, 91, 9)))

# A tuple is immutable: any attempt to modify it raises a TypeError
try:
  week[0] = "monday morning"
except TypeError as error:
  print(f"TypeError: {error}")
```

Read operations (indexing, slicing, `len`, `in`, …) work as with lists, but a tuple cannot be modified once created: no item assignment, no `append`, no `remove`. A tuple is therefore used for a fixed-size sequence that must not change (days of the week, coordinates…), and unlike a list it can be used as a dictionary key or a set element.

## Dictionaries manipulation

Re-express the seasons data structure using a dictionary. It should have 4 entries, each one with the season name as a key and the list of months as a value.

- How to access the second month of winter?
- How to test if `summer` is in the dictionary `seasons`?

```python
# Your code here
```

### Solution

```python
winter = ["December", "January", "February"]
spring = ["March", "April", "May"]
summer = ["June", "July", "August"]
autumn = ["September", "October", "November"]

seasons = {"winter": winter, "spring": spring, "summer": summer, "autumn": autumn}
print(seasons["winter"][1])
print("summer" in seasons)
```
