## Admis ou recalé

Le fichier [`notes.txt`](https://raw.githubusercontent.com/mlambda/langage-python/main/data/notes.txt) contient les notes obtenues par des étudiants pour le cours de Python. Chaque ligne du fichier ne contient qu'une note.

Toutes les notes seront écrites avec une décimale. À titre d'exemple, voici les 3 premières lignes attendues pour le fichier `notes2.txt`&nbsp;:

```default

17.0 admis

```

````
# Votre code ici
````

### Solution

````
import pathlib


# Voir les solutions à l'exercice précédent pour d'autres versions
notes = [float(n)
         for n in pathlib.Path("notes.txt").read_text(encoding="utf8").split()]

with pathlib.Path("notes2.txt").open(mode="w", encoding="utf8") as fichier:
    for note in notes:
        verdict = "admis" if note >= 10 else "recalé"
        fichier.write(f"{note} {verdict}\n")

# En itérant sur les deux en même temps
with open("notes.txt", encoding="utf8") as fic_read, open(
    "notes2.txt", mode="w", encoding="utf8") as fic_write:
  for line in fic_read:
    note = float(line)
    # mention = "admis" if note >= 10 else "recalé"
    if note < 10:
      mention = "recalé"
    elif note >= 10:
      mention = "admis"
    fic_write.write(f"{note:04.1f} {mention}\n")
with open("notes2.txt", encoding="utf8") as fic_read2:
  print(fic_read2.read())
````

## Spirale

Dans cet exercice, vous allez calculer les coordonnées cartésiennes d'une spirale à deux dimensions.

Les coordonnées cartésiennes $x_A$ et $y_A$ d'un point $A$ sur un cercle de rayon $r$ s'expriment en fonction de l'angle $\theta$ représenté sur le schéma ci-dessous : $x_A = \cos(\theta) \times r$ et $y_A = \sin(\theta) \times r$.

![Point A sur le cercle de rayon r.](https://raw.githubusercontent.com/mlambda/langage-python/main/img/spirale-coord.png "Point A sur le cercle de rayon r.")

Pour calculer les coordonnées cartésiennes qui décrivent la spirale, vous allez faire varier deux variables en même temps&nbsp;:

- l'angle $\theta$, qui va prendre des valeurs de $0$ à $4\pi$ radians par pas de $0.1$, ce qui correspond à deux tours complets ;
- le rayon du cercle $r$, qui va prendre comme valeur initiale $0.5$ puis que vous allez incrémenter (c'est-à-dire augmenter) par pas de $0.1$.

Les fonctions trigonométriques sinus et cosinus sont disponibles dans le module `math` que vous découvrirez plus en détails dans la suite de la formation. Pour les utiliser, vous ajouterez au début de votre script l'instruction&nbsp;:

```python
import math
```

La fonction sinus sera [`math.sin`](https://docs.python.org/fr/3/library/math.html#math.sin) et la fonction cosinus [`math.cos`](https://docs.python.org/fr/3/library/math.html#math.cos). Ces deux fonctions prennent comme argument une valeur d'angle en radian. La constante mathématique $\pi$ sera également accessible grâce à ce module via [`math.pi`](https://docs.python.org/fr/3/library/math.html#math.pi). Par exemple&nbsp;:

```python
>>> math.sin(0)
0.0
>>> math.sin(math.pi / 2)
1.0
>>> math.cos(math.pi)
-1.0
```

Sauvegardez ensuite les coordonnées cartésiennes dans le fichier `spirale.dat` en respectant le format suivant&nbsp;:

- un couple de coordonnées ($x_A$ et $y_A$) par ligne ;
- au moins un espace entre les deux coordonnées $x_A$ et $y_A$ ;
- les coordonnées affichées sur 10 caractères avec 5 chiffres après la virgule.

Les premières lignes de `spirale.dat` devraient ressembler à&nbsp;:

```default
   0.50000    0.00000
   0.59700    0.05990
   0.68605    0.13907
   0.76427    0.23642
   0.82895    0.35048
   0.87758    0.47943
[...]
```

Vous pouvez ensuite essayer de jouer sur les paramètres $\theta$ et $r$, et leur pas d'incrémentation, pour construire de nouvelles spirales.

````
# À exécuter après avoir rempli spirale.dat ci-dessus
import matplotlib.pyplot as plt


xs = []
ys = []
with open("spirale.dat", "r") as f_in:
    for line in f_in:
        x, y = map(float, line.split())
        xs.append(x)
        ys.append(y)

plt.figure(figsize=(8, 8))
plt.axis("off")
plt.plot(xs, ys)
plt.show()
````

### Solution

````
import math

theta = 0
r = 0.5
with pathlib.Path("spirale.dat").open(mode="w", encoding="utf8") as fichier:
    while theta < math.pi * 4:
        fichier.write(f"{math.cos(theta) * r} {math.sin(theta) * r}\n")
        theta += 0.1
        r += 0.1
````

## Conversion de format

Reprenez le fichier `.csv` de l'exercice précédent et écrivez-le sous forme de fichier `.json` et `.jsonl`.

Pour le fichier `.json` :

1. Créer une liste qui contiendra autant d'éléments qu'il y a de lignes dans le csv
2. Chaque élément sera un dictionnaire qui contiendra comme clefs les colonnes du csv (headers) et comme valeurs les valeurs de la ligne csv
3. Appeler [`json.dump`](https://docs.python.org/fr/3/library/json.html#json.dump) pour écrire l'objet dans un fichier

Pour le fichier `.jsonl`, reprendre le même principe mais écrire le dictionnaire créé à chaque étape `2.` directement dans le fichier de sortie grâce à la méthode [`json.dumps`](https://docs.python.org/fr/3/library/json.html#json.dumps)

````
# Votre code ici
````

### Solution

Fichier `.json`

````
import csv
import json


data = []

with open("wiki_movie_plots_deduped.csv", newline='', encoding="utf8") as f:
    reader = csv.reader(f)
    headers = next(reader)
    for i, row in enumerate(reader):
      # Méthode 1, avec les indices des listes headers et row
      line_dict = {}
      for j in range(len(headers)):
        key = headers[j]
        value = row[j]
        line_dict[key] = value
      # Méthode 2, avec enumerate pour faire le lien entre headers et row
      line_dict = {}
      for j, value in enumerate(headers):
        key = headers[j]
        line_dict[key] = value
      # Méthode 3, avec zip
      line_dict = {}
      for key, value in zip(headers, row):
        line_dict[key] = value
      # Méthode 4, avec zip et une compréhension de dictionnaire
      line_dict = {key: value for key, value in zip(headers, row)}
      # Méthode 5, avec zip et un appel à dict
      line_dict = dict(zip(headers, row))
      data.append(line_dict)

with open("wiki_movie_plots_deduped.json", "w", encoding="utf8") as fh:
  json.dump(data, fh)
````

Fichier `.jsonl`

````
import csv
import json


with open("wiki_movie_plots_deduped.csv", newline='', encoding="utf8") as fhr, \
    open("wiki_movie_plots_deduped.jsonl", "w", encoding="utf8") as fhw:
    reader = csv.reader(fhr)
    headers = next(reader)
    for i, row in enumerate(reader):
      line_dict = dict(zip(headers, row))
      fhw.write(f"{json.dumps(line_dict)}\n")
````
