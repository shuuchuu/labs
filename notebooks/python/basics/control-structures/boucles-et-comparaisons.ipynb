## Boucles de base

Affichez l'ensemble des éléments de la liste `bibliotheques` (un élément par ligne) de trois manières différentes (deux avec `for` et une avec `while`).

````
bibliotheques = ["numpy", "pandas", "tensorflow", "flask"]
````

````
# Votre code ici
````

### Solution

````
print("Avec une boucle for sur les éléments")
for bibliotheque in bibliotheques:
    print(bibliotheque)

print()
print("Avec une boucle for sur les indices")
for i in range(len(bibliotheques)):
    print(bibliotheques[i])

print()
print("Avec une boucle while")
i = 0
while i < len(bibliotheques):
    print(bibliotheques[i])
    i += 1
````

## Produit de nombres consécutifs

Avec les fonctions [`list`](https://docs.python.org/fr/3/library/functions.html#list) et [`range`](https://docs.python.org/fr/3/library/functions.html#range), créez la liste `entiers` contenant les nombres entiers pairs de 2 à 20 inclus.

Calculez ensuite le produit des nombres consécutifs deux à deux de `entiers` en utilisant une boucle. Exemple pour les premières itérations&nbsp;:

```
8
24
48
[...]
```

````
# Votre code ici
````

### Solution

````
entiers = list(range(2, 21, 2))

# Avec les indices
for i in range(1, len(entiers)):
    print(entiers[i - 1] * entiers[i])

# Avec la fonction zip
for courant, suivant in zip(entiers, entiers[1:]):
    print(courant * suivant)
````

````
# Votre code ici
````

````
print("Premier exercice")
n = 5
for i in range(1, n):
    for j in range(i + 1, n + 1):
        print(i, j)

print()
print("Deuxième exercice")
for n in range(2, 10):
    total = 0
    for i in range(1, n):
        for j in range(i + 1, n + 1):
            total += 1
    print(n, total)
````

On peut trouver une formule directe : $\frac{n(n - 1)}{2}$

## Suite de Fibonacci

La [suite de Fibonacci](https://fr.wikipedia.org/wiki/Suite_de_Fibonacci) est une suite mathématique qui porte le nom de Leonardo Fibonacci, un mathématicien italien du $XIII^e$ siècle. Initialement, cette suite a été conçue pour décrire la croissance d'une population de lapins, mais elle peut également être utilisée pour décrire certains motifs géométriques retrouvés dans la nature (coquillages, fleurs de tournesol, …).

Pour la suite de Fibonacci $(x_n)$, le terme au rang $n$ (avec $n > 1$) est la somme des nombres aux rangs $n − 1$ et $n − 2$&nbsp;:

$x_n = x_{n − 1} + x_{n − 2}$

Par définition, les deux premiers termes sont $x_0 = 0$ et $x_1 = 1$.

À titre d'exemple, les 10 premiers termes de la suite de Fibonacci sont donc 0, 1, 1, 2, 3, 5, 8, 13, 21 et 34.

Créez un script qui construit une liste `fibo` avec les 15 premiers termes de la suite de Fibonacci puis l'affiche.

Améliorez ce script en affichant, pour chaque élément de la liste `fibo` avec $n > 1$, le rapport entre l'élément de rang $n$ et l'élément de rang $n − 1$. Ce rapport tend-il vers une constante&nbsp;? Si oui, laquelle ?

````
# Votre code ici
````

### Solution

````
n = 15
fibo = [0, 1]

while len(fibo) < n:
    fibo.append(fibo[-1] + fibo[-2])
print(fibo)

for i in range(2, n):
    print(fibo[i] / fibo[i - 1])
````

Le ratio $\frac{\text{fibo}_n}{\text{fibo}_{n - 1}}$ tend vers [le nombre d'or](https://fr.wikipedia.org/wiki/Nombre_d%27or).
