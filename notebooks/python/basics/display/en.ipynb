## Controlling the end character

The `print` function has several useful arguments. The first one is `end`, which controls what is displayed after the argument passed to `print`. By default, it is a newline.

*Call the `print` function several times in a row and make the successive outputs appear on the same line.*

```
# Your code here
```

### Solution

```
print("Hello", end=" ")
print("World")
```

## Floats

In addition to the slides, we can use https://pyformat.info/ as a reference for the following exercises.

Using the `.format` method on strings, display pi with only two decimal places.

```
import math


print(math.pi)  # Your code here
```

Now, print `420.671` and pi so that they are aligned on their central `.`

```
# Your code here
```

### Solution

```
import math


# Pi with two decimal places
print("{:.2f}".format(math.pi))

# Aligned on the dot: same total width and same number of decimal places
print("{:7.3f}".format(420.671))
print("{:7.3f}".format(math.pi))
```

## `f-strings`

Since Python 3.6, [f-strings](https://docs.python.org/fr/3/tutorial/inputoutput.html#tut-f-strings) are a shortcut to the `.format` method.

Use f-strings to display pi and `420.671` as in the last exercise.

```
# Your code here
```

### Solution

```
print(f"{math.pi:.2f}")

print(f"{420.671:7.3f}")
print(f"{math.pi:7.3f}")
```
