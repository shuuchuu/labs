## Predicting the result of operations

Without executing the code, try to predict the result of those operations:

1. `(1 + 2) ** 3`
2. `"Da" * 4`
3. `"Da" + 3`
4. `("Pa" + "La") * 2`
5. `("Da" * 4) / 2`
6. `"Da" + "3"`
7. `5 / 2`
8. `5 // 2`
9. `5 % 2`

Your answer here:

1. …
2. …
3. …
4. …
5. …
6. …
7. …
8. …
9. …

### Solution

````
print((1 + 2) ** 3)
print("Da" * 4)
# print("Da" + 3) would produce:
# TypeError: must be str, not int
print(("Pa" + "La") * 2)
# print(("Da" * 4) / 2) would produce:
# TypeError: unsupported operand type(s) for /: 'str' and 'int'
print("Da" + "3")
print(5 / 2)
print(5 // 2)
print(5 % 2)
````

## Predicting the result of operations with type conversion

Without executing the code, try to predict the result of those operations:

1. `str(4) * int("3")`
2. `int("3") * str(4)`
3. `int("3") + float("3.2")`
4. `str(4) * float(3)`
5. `str(3 / 4) * 2`
6. `str(42) * bool(42)`
7. `bool(0) * str(42)`

Your answer here:

1. …
2. …
3. …
4. …
5. …
6. …
7. …

### Solution

````
print(str(4) * int("3"))
print(int("3") * str(4))
print(int("3") + float("3.2"))
# print(str(4) * float(3)) would produce :
# TypeError: can't multiply sequence by non-int of type 'float'
print(str(3 / 4) * 2)
print(str(42) * bool(42))
print(bool(0) * str(42))
````
