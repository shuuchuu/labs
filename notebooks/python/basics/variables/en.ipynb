## Type conversion

It is easy to go from one type to another: you can use the classes you have just discovered to convert a value into another type.

- *Use the [`int`](https://docs.python.org/3/library/functions.html#int), [`float`](https://docs.python.org/3/library/functions.html#float), etc. functions to convert variables to the desired type.*

- *What happens when the conversion fails?*

````
# Your code here
````

### Solution

````
# To integer
print("Integer from '42'  ", int("42"))
print("Integer from 42.6  ", int(42.6))
print("Integer from True  ", int(True))

# To float
print("Float   from '42'  ", float("42"))
print("Float   from 42    ", float(42))
print("Float   from True  ", float(True))

# To boolean
print("Boolean from ''    ", bool(""))
print("Boolean from 'a'   ", bool("a"))
print("Boolean from 0     ", bool(0))
print("Boolean from 1     ", bool(1))

# To string
print("String  from 42    ", str(42))
print("String  from True  ", str(True))

# When the conversion is not possible, a ValueError is raised
int("a")
````

### Solution

````
print(str(4) * int("3"))
print(int("3") * str(4))
print(int("3") + float("3.2"))
# print(str(4) * float(3)) would produce :
# TypeError: can't multiply sequence by non-int of type 'float'
print(str(3 / 4) * 2)
print(str(42) * bool(42))
print(bool(0) * str(42))
````
