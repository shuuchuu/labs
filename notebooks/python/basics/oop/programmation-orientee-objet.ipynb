Créez une classe `Rectangle`, qui accepte deux nombres dans sa méthode `__init__` : une largeur et une hauteur et les stocke comme attributs. Ajoutez une méthode `calcule_surface` à cette classe qui renvoie la surface du rectangle (sa largeur multipliée par sa hauteur).

Créez une instance de cette classe et vérifiez que vous pouvez calculer sa surface.

```
# Votre code ici
```

### Solution

```
class Rectangle:

    def __init__(self, longueur, largeur):
        """Initialise un rectangle."""
        self.longueur = longueur
        self.largeur = largeur

    def calcule_surface(self):
        """Calcule la surface."""
        return self.longueur * self.largeur


rectangle = Rectangle(2, 4)
rectangle.calcule_surface()
```

### Solution

```
class Rectangle:

    def __init__(self, longueur, largeur):
        """Initialise un rectangle."""
        self.longueur = longueur
        self.largeur = largeur

    def calcule_surface(self):
        """Calcule la surface."""
        return self.longueur * self.largeur

    def __eq__(self, autre):
        if isinstance(autre, Rectangle):
            return (autre.longueur == self.longueur
                    and autre.largeur == self.largeur)
        return NotImplemented

    def __repr__(self):
        return (f"{self.__class__.__name__}(longueur={self.longueur}, "
                f"largeur={self.largeur})")
```

```
rectangle = Rectangle(10, 20)
print(rectangle)
print(repr(rectangle))
print(f"Largeur : {rectangle.largeur}, longueur : {rectangle.longueur}")
print(f"Surface : {rectangle.calcule_surface()}")
rectangle2 = Rectangle(10, 20)
# rectangle2 = eval(repr(rectangle))
print(rectangle is rectangle2)
print(rectangle == rectangle2)
```

## Classe `Losange`

Créez une classe `Losange`, similaire à `Rectangle`. On utilisera la longueur des diagonales pour créer un losange.

```
# Votre code ici
```

### Solution

```
class Losange:

    def __init__(self, diagonale1, diagonale2):
        """Initialise un losange."""
        self.diagonale1 = diagonale1
        self.diagonale2 = diagonale2

    def calcule_surface(self):
        """Calcule la surface."""
        return self.diagonale1 * self.diagonale2 / 2

    def __eq__(self, autre):
        if isinstance(autre, Losange):
            return (autre.diagonale1 == self.diagonale1
                    and autre.diagonale2 == self.diagonale2)
        return NotImplemented

    def __repr__(self):
        return (f"{self.__class__.__name__}(diagonale1={self.diagonale1}, "
                f"diagonale2={self.diagonale2})")


losange1 = Losange(3, 6)
losange2 = Losange(3, 6)
losange3 = Losange(2, 6)
print(losange1)
print(repr(losange1))
print(losange1.calcule_surface())
print(losange1 == losange2)
print(losange1 == losange3)
```

## Classe `Carre`

Héritez des classes `Rectangle` et `Losange` pour créer la classe `Carre`.

Définissez une méthode supplémentaire : `calcule_surface_cercle_inscrit`, qui renvoie l'aire du cercle inscrit dans le carré, en utilisant `calcule_surface` (l'aire du cercle inscrit est l'aire du carré multipliée par $\frac{\pi}{4}$).

Où est définie la méthode `calcule_surface` que `Carre` utilise ?

```
# Votre code ici
```

### Solution

```
import math


class Carre(Rectangle, Losange):
    def __init__(self, cote):
        self.cote = cote
        Rectangle.__init__(self, cote, cote)
        Losange.__init__(self, cote * math.sqrt(2), cote * math.sqrt(2))

    def calcule_surface_cercle_inscrit(self):
        return super().calcule_surface() * math.pi / 4

    def __eq__(self, autre):
        if isinstance(autre, Carre):
            return autre.cote == self.cote
        return NotImplemented


carre = Carre(1)
print(carre.calcule_surface_cercle_inscrit())
Carre.__mro__
```

## Comparaison de classes avec héritage

Essayez de faire une comparaison entre un carré et un rectangle puis déterminez les méthodes appellées et dans quel ordre ?

```
carre = Carre(2)
rectangle = Rectangle(2, 2)

carre == rectangle

rectangle == rectangle
```

### Solution

Quand l'égalité est testée `a == b`, les méthodes sont appellées dans cette ordre :
 1. Si `a` et `b` sont de même type, alors cela revient à `a.__eq__(b)`
 2. Si `a` et `b` sont de types différents et b est une classe héritée de a, alors cela revient à `b.__eq__(a)`
 3. Si `a` et `b` sont de types différentes et b n'est pas une classe héritée de `a`, alors cela revient à `a.__eq__(b)`. Si elle renvoie `NotImplemented`, la méthode de `b` est alors essayée `b.__eq__(a)`.
