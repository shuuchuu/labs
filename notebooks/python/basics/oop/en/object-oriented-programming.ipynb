### Solution

```
class Rectangle:

    def __init__(self, length, width):
        """Initialize a rectangle."""
        self.length = length
        self.width = width

    def compute_area(self):
        """Compute the area."""
        return self.length * self.width

    def __eq__(self, other):
        if isinstance(other, Rectangle):
            return (other.length == self.length
                    and other.width == self.width)
        return NotImplemented

    def __repr__(self):
        return (f"{self.__class__.__name__}(length={self.length}, "
                f"width={self.width})")
```

## `Rhombus` class

Create a `Rhombus` class, similar to `Rectangle`. A rhombus is created from the lengths of its diagonals.

```
# Your code here
```

### Solution

```
class Rhombus:

    def __init__(self, diagonal1, diagonal2):
        """Initialize a rhombus."""
        self.diagonal1 = diagonal1
        self.diagonal2 = diagonal2

    def compute_area(self):
        """Compute the area."""
        return self.diagonal1 * self.diagonal2 / 2

    def __eq__(self, other):
        if isinstance(other, Rhombus):
            return (other.diagonal1 == self.diagonal1
                    and other.diagonal2 == self.diagonal2)
        return NotImplemented

    def __repr__(self):
        return (f"{self.__class__.__name__}(diagonal1={self.diagonal1}, "
                f"diagonal2={self.diagonal2})")


rhombus1 = Rhombus(3, 6)
rhombus2 = Rhombus(3, 6)
rhombus3 = Rhombus(2, 6)
print(rhombus1)
print(repr(rhombus1))
print(rhombus1.compute_area())
print(rhombus1 == rhombus2)
print(rhombus1 == rhombus3)
```

Where is the `compute_area` method used by `Square` defined?

### Solution

```
import math


class Square(Rectangle, Rhombus):
    def __init__(self, side):
        self.side = side
        Rectangle.__init__(self, side, side)
        Rhombus.__init__(self, side * math.sqrt(2), side * math.sqrt(2))

    def compute_inscribed_circle_area(self):
        return super().compute_area() * math.pi / 4

    def __eq__(self, other):
        if isinstance(other, Square):
            return other.side == self.side
        return NotImplemented


square = Square(1)
print(square.compute_inscribed_circle_area())
Square.__mro__
```

## Comparing classes with inheritance

Compare a square and a rectangle, then determine which methods are called, and in which order.

```
square = Square(2)
rectangle = Rectangle(2, 2)

square == rectangle

rectangle == rectangle
```

### Solution

When the equality `a == b` is tested, the methods are called in this order:
 1. If `b` is an instance of a strict subclass of the type of `a` that overrides `__eq__`, then `b.__eq__(a)` is tried first
 2. Otherwise, `a.__eq__(b)` is called
 3. If the called method returns `NotImplemented`, the other operand is tried in turn (`a.__eq__(b)` or `b.__eq__(a)`)
 4. If both return `NotImplemented`, Python compares identities (`a is b`)

Here, `square == rectangle` first calls `Square.__eq__`, which returns `NotImplemented` since `rectangle` is not a `Square`, then `Rectangle.__eq__(rectangle, square)`, which compares length and width and returns `True`.
