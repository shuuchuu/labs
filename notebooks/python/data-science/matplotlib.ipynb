## Un premier plot simple

Créez un premier plot qui trace la fonction sinus et cosinus.

### Solution

```python
# Création des données de plot
x = np.linspace(0, np.pi * 2)
y_cos = np.cos(x)
y_sin = np.sin(x)

# Création figure et axe
fig, ax = plt.subplots(dpi=100)

# Création du plot
# Cosinus
ax.plot(x, y_cos, label="cos")
# Sinus
ax.plot(x, y_sin, label="sin")
# Coloration entre les deux courbes Couleur #B8DDF9
ax.fill_between(x, y_cos, y_sin, color="#b8ddf9")


# Ajouter un titre et une légende
ax.set_title("Fonction cosinus et sinus")
ax.legend()

fig.show()
```

## `imshow` et `contour`

Nous allons visualiser une fonction $f$ à deux paramètres $x$ et $y$ définie comme suit :

$$
  f(x,y) = z = \frac{\sin(x^2 + y^2)}{x^2 + y^2}
$$

Pour pouvoir l'afficher en deux dimensions, nous allons utiliser [`imshow`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.imshow.html) et [`contour`](https://matplotlib.org/stable/api/contour_api.html) :
* Créez les variables `x` et `y` nécessaires à l'affichage. Pour cela vous aurez besoin de [`np.linspace`](https://numpy.org/doc/stable/reference/generated/numpy.linspace.html) et [`np.meshgrid`](https://numpy.org/doc/stable/reference/generated/numpy.meshgrid.html) pour créer une valeur sur chaque pixel du résultat à afficher.
* Calculez `z`
* Affichez `z` avec [`imshow`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.imshow.html) en utilisant la transparence à 0.5 (`alpha=0.5`)
* Affichez `z` avec [`contour`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.contour.html)

```python
# # Création des variables x et y
# # dans l'intervalle [-5,5]
# x = np.linspace(-5, 5, 1000)
# y = np.linspace(-5, 5, 1000)

# # Création des données en meshgrid
# xx, yy = ???

# # Calcul de z à partir de xx et yy
# z = ???

# # Creation de fig et ax
# fig, ax = ???

# # Création du plot avec imshow et contour
# ax.???
# ax.???

# fig.show()
```

### Solution

```python
# Création des variables x et y
# dans l'intervalle [-5,5]
x = np.linspace(-5, 5, 1000)
y = np.linspace(-5, 5, 1000)

# Création des données en meshgrid
xx, yy = np.meshgrid(x, y)

# Calcul de z à partir de xx et yy
z = np.sin(xx**2 + yy**2) / (xx**2 + yy**2)

# Creation de fig et ax
fig, ax = plt.subplots(dpi=100)

# Création du plot avec imshow et contour
ax.imshow(z, alpha=0.5, extent=[xx.min(), xx.max(), yy.min(), yy.max()])
ax.contour(xx, yy, z)

fig.show()
```

Une autre solution consiste évidemment à faire directement un affichage en 3D

```python
# Création des variables x et y
# dans l'intervalle [-5,5]
x = np.linspace(-5, 5, 1000)
y = np.linspace(-5, 5, 1000)

# Création des données en meshgrid
xx, yy = np.meshgrid(x, y)

# Calcul de z à partir de xx et yy
z = np.sin(xx**2 + yy**2) / (xx**2 + yy**2)

# Creation de fig et ax
fig, ax = plt.subplots(dpi=150, subplot_kw=dict(projection="3d"))

# Création du plot avec imshow et contour
g = ax.plot_surface(xx, yy, z, cmap="coolwarm")
fig.colorbar(g)
plt.show()
```

Vous allez créer 4 plots différents des trois fonctions suivantes :
* $f(x) = 10^x$
* $f(x) = x$
* $f(x) = \log(x)$

Vous allez afficher ces trois fonctions sur 4 plots avec différentes échelles sur $x$ et $y$ :
1.   Axe $x$ linéaire et axe $y$ linéaire
2.   Axe $x$ logarithmique et axe $y$ linéaire
3.   Axe $x$ linéaire et axe $y$ logarithmique

en suivant la disposition suivante :

![](https://raw.githubusercontent.com/rougier/matplotlib-tutorial/master/figures/subplot-grid.png)

Afin de faire la visualisation :
* Utilisez [`subplots`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html?highlight=subplots#matplotlib.pyplot.subplots) pour créer un figure contenant 4 plots sur 2 lignes et 2 colonnes
* Utilisez [`set_xscale`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_xscale.html?highlight=set_xscale#matplotlib.axes.Axes.set_xscale) et [`set_yscale`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_yscale.html?highlight=set_yscale#matplotlib.axes.Axes.set_yscale) pour changer l'échelle des axes
* Utilisez [`set_xlim`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_xlim.html?highlight=set_xlim#matplotlib.axes.Axes.set_xlim) et [`set_ylim`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_ylim.html?highlight=set%20ylim#matplotlib.axes.Axes.set_ylim) pour définir les bornes d'affichages



```python
# X = np.linspace(0.001, 90, 5000)
# fig, ax = plt.subplots(2, 2, figsize=(8, 8), dpi=100)
```

### Solution

```python
x = np.linspace(0.001, 90, 5000)
fig, ax = plt.subplots(2, 2, figsize=(8, 8), dpi=100)


def subplot(ax: mpl.axes.Axes, xlog: bool = False, ylog: bool = False) -> None:
    if xlog:
        ax.set_xscale("log")
        ax.set_xlim(0.01, 100)
    else:
        ax.set_xlim(0, 10)
    if ylog:
        ax.set_yscale("log")
        ax.set_ylim(0.01, 100)
    else:
        ax.set_ylim(0, 10)
    ax.plot(x, 10**x, label="$f(x) = 10^x$")
    ax.plot(x, x, label="$f(x) = x$")
    ax.plot(x, np.log10(x), label=r"$f(x) = \log(x)$")
    ax.grid(visible=True, which="both", axis="both")
    x_title = f"$x$ {'logarithmique' if xlog else 'linéaire'}"
    y_title = f"$y$ {'logarithmique' if ylog else 'linéaire'}"
    ax.set_title(f"{x_title} - {y_title}")
    ax.legend()


# x linéaire, y linéaire
subplot(ax[0, 0])
ax[0, 0].set_ylabel("Linéaire")

# x log, y linéaire
subplot(ax[0, 1], xlog=True)

# x linéaire, y log
subplot(ax[1, 0], ylog=True)
ax[1, 0].set_xlabel("Linéaire")
ax[1, 0].set_ylabel("Log")

# x log, y log
subplot(ax[1, 1], xlog=True, ylog=True)
ax[1, 1].set_xlabel("Log")

fig.show()
```

## Projection polaire

Il est possible de faire des visualisations polaires en spécifiant une projection polaire.


```python
rng = np.random.default_rng()
x = rng.integers(0, 10, size=15)

fig = plt.figure(dpi=100)
ax = fig.add_subplot(1, 2, 1)
ax.hist(x)
ax = fig.add_subplot(1, 2, 2, projection="polar")
ax.hist(x)

fig.show()
```
