## Broadcast

Given the following tables :

```python
a = np.array([1, 2, 3])

b = np.array([[1, 2, 3]])

c = np.array([[1],
              [2],
              [3]])
```

Predict the shape of the following operations:
1. `a + a`
2. `a + b`
3. `a + c`
4. `b + c`
5. `a + a[np.newaxis, :]`
6. `a + a[:, np.newaxis]`

### Solution

#### Reminder
The broadcasting steps are :

1. Right alignment of the shapes
2. Add dimensions to `1` in the smallest array if necessary
3. The `1` dimensions of one array are expanded to equal the other array

#### Solutions

1. Two arrays of the same size are added: output of the same size. Broadcasting process:

| Arrays   | Shapes | Step 1  | Step 2  | Step 3  |
|----------|--------|--------:|--------:|--------:|
| `a`      | `3`    | `3`     | `3`     |   `3`   |
| `a`      | `3`    | `3`     | `3`     |   `3`   |

2. Broadcasting process:

| Arrays   | Shapes | Step 1  | Step 2  | Step 3  |
|----------|--------|--------:|--------:|--------:|
| `a`      | `3`    | `3`     | `1 3`   | `1 3`   |
| `b`      | `1 3`  | `1 3`   | `1 3`   | `1 3`   |

3. Broadcasting process:

| Arrays   | Shapes | Step 1  | Step 2  | Step 3  |
|----------|--------|--------:|--------:|--------:|
| `a`      | `3`    | `3`     | `1 3`   | `3 3`   |
| `c`      | `3 1`  | `3 1`   | `3 1`   | `3 3`   |

4. Same results as `3.` from step 2. Broadcasting process:

| Tableaux | Formes | Étape 1 | Étape 2 | Étape 3 |
|----------|--------|--------:|--------:|--------:|
| `b`      | `1 3`  | `1 3`   | `1 3`   | `3 3`   |
| `c`      | `3 1`  | `3 1`   | `3 1`   | `3 3`   |

5. `a[:, np.newaxis]` equivalent to `c`, same solution that `3.`
6. `a[np.newaxis, :]` equivalent to `b`, same solution that `2.`
