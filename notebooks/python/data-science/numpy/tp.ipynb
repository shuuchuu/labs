```python
import matplotlib.pyplot as plt
import numpy as np

def display(arr: np.ndarray) -> None:
  channels = arr.shape[-1] in {1, 3}
  batched = arr.ndim == (4 if channels else 3)
  if not batched:
    _, ax = plt.subplots(figsize=(arr.shape[1] // 28 * 2, arr.shape[0] // 28 * 2))
    ax.imshow(arr, cmap="gray_r")
    ax.axis("off")
    plt.show()
  if batched:
    _, ax = plt.subplots(2, 5, figsize=(10, 4))
    for index in range(10):
      axi = ax[index // 5, index % 5]
      axi.set_title(f"Élément {index}")
      axi.imshow(arr[index], cmap="gray_r" if index % 2 else "gray")
      axi.axis("off")
```

### Solution

```python
def print_array(array: np.ndarray, name: str) -> None:
  print(f"Tableau {name}")
  print(array)
  print()

a_1 = np.zeros(5)
print_array(a_1, "a_1")

a_2 = np.ones((2, 3))
print_array(a_2, "a_2")

a_3 = np.empty((2, 3))
a_3.fill(8)
print_array(a_3, "a_3")

a_4 = np.ones((2, 3), dtype="bool")
print_array(a_4, "a_4")

a_5 = np.random.normal(0, 1, (2, 3, 5))
print_array(a_5, "a_5")

a_6 = np.empty_like(a_5)
a_6.fill(42)
print_array(a_6, "a_6")

a_7_shape = list(a_5.shape)
a_7_shape[1] = 5
a_7 = np.zeros(a_7_shape, dtype="uint8")
print_array(a_7, "a_7")

a_8 = np.arange(10)
print_array(a_8, "a_8")

a_9 = np.linspace(0, 1, 101)
print_array(a_9, "a_9")

a_10 = np.array(data)
print_array(a_10, "a_10")
```

## Broadcast

Étant donné les tableaux suivants :

```python
a = np.array([1, 2, 3])

b = np.array([[1, 2, 3]])

c = np.array([[1],
              [2],
              [3]])
```

Prédisez la forme de sortie de ces opérations :
1. `a + a`
2. `a + b`
3. `a + c`
4. `b + c`
5. `a + a[np.newaxis, :]`
6. `a + a[:, np.newaxis]`

### Solution

#### Rappel
Les étapes d'un broadcasting sont :

1. Alignement à droite des formes
2. Ajout de dimensions à `1` dans le plus petit tableau si nécessaire
3. Les dimensions à `1` d'un tableau sont étendues pour égaler celle de l'autre tableau

#### Solutions

1. On additionne deux tableaux de même taille : sortie de même taille. Processus de broadcasting :

| Tableaux | Formes | Étape 1 | Étape 2 | Étape 3 |
|----------|--------|--------:|--------:|--------:|
| `a`      | `3`    | `3`     | `3`     |   `3`   |
| `a`      | `3`    | `3`     | `3`     |   `3`   |

2. Processus de broadcasting :

| Tableaux | Formes | Étape 1 | Étape 2 | Étape 3 |
|----------|--------|--------:|--------:|--------:|
| `a`      | `3`    | `3`     | `1 3`   | `1 3`   |
| `b`      | `1 3`  | `1 3`   | `1 3`   | `1 3`   |

3. Processus de broadcasting :

| Tableaux | Formes | Étape 1 | Étape 2 | Étape 3 |
|----------|--------|--------:|--------:|--------:|
| `a`      | `3`    | `3`     | `1 3`   | `3 3`   |
| `c`      | `3 1`  | `1 3`   | `3 1`   | `3 3`   |

4. Même résultat que `3.` à partir de l'étape 2. Processus de broadcasting :

| Tableaux | Formes | Étape 1 | Étape 2 | Étape 3 |
|----------|--------|--------:|--------:|--------:|
| `b`      | `1 3`  | `1 3`   | `1 3`   | `3 3`   |
| `c`      | `3 1`  | `1 3`   | `3 1`   | `3 3`   |

5. `a[:, np.newaxis]` est équivalent à `c`, donc même solution que `3.`
6. `a[np.newaxis, :]` est équivalent à `b`, donc même solution que `2.`
