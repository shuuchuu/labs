## Création des fichiers python

Créez un fichier `cli.py` dans le dossier source de votre paquet (`src/<nom_du_paquet>/`, où `<nom_du_paquet>` est le nom de votre paquet avec des `_` à la place des `-`) avec une fonction `main` qui fait ce que vous voulez.

Éditez le fichier `pyproject.toml` et ajoutez-y ces champs en remplaçant `<nom_du_paquet>` :

```toml
[project.scripts]
mycommand = "<nom_du_paquet>.cli:main"
```

C'est ce qui va permettre à votre fonction `main` d'être utilisable avec la ligne de commande `mycommand`

### Solution

Fichier `src/tp_2026_09_23_monpseudo/cli.py` :

```python
import sys


def main() -> None:
  print(sys.argv)
```

Ajout dans `pyproject.toml` :

```toml
[project.scripts]
mycommand = "tp_2026_09_23_monpseudo.cli:main"
```

## Ajout de l'index Test PyPI à la configuration

Pour que Test PyPI soit disponible pendant la publication, il faut ajouter les entrées suivantes au fichier `pyproject.toml` de votre projet&nbsp;:

```toml
[[tool.uv.index]]
name = "pypi"
url = "https://pypi.org/simple/"
publish-url = "https://pypi.org/legacy/"

[[tool.uv.index]]
name = "testpypi"
url = "https://test.pypi.org/simple/"
publish-url = "https://test.pypi.org/legacy/"
```

## Publication de paquet

Publiez maintenant votre paquet sur Test PyPI. Pour cela, par rapport à une publication normale, vous devrez préciser l'argument `--index testpypi` pour que le serveur choisi ne soit pas PyPI mais bien Test PyPI.

Authentifiez-vous avec l'argument `--token` et le token créé au début du TP. Ne l'écrivez pas en clair dans le notebook : récupérez-le avec [`getpass.getpass`](https://docs.python.org/fr/3/library/getpass.html#getpass.getpass), puis passez la variable à la commande avec `{token}`.
