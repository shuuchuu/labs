## Creating the python files

Create a `cli.py` file in your folder with a `main()` function that prints `Hello World`.

Edit the `pyproject.toml` file and add these fields, making sure to change the `labs_m09` part to point to your package source folder name (**not your project name**):

```toml
[project.scripts]
demo-packaging = 'labs_m09.cli:main'
```

This will make your `main` function usable with the `demo-packaging` command line

## Adding TestPyPI to `uv` configuration

Add these lines to your `pyproject.toml` file:

```toml
[[tool.uv.index]]
name = "pypi"
url = "https://pypi.org/simple/"
publish-url = "https://pypi.org/legacy/"

[[tool.uv.index]]
name = "testpypi"
url = "https://test.pypi.org/simple/"
publish-url = "https://test.pypi.org/legacy/"
```
