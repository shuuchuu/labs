```python
%%cython
# cython: language_level=3, boundscheck=False, wraparound=False

def cumsum(double[:] a, double[:] b):
    """
    Compute cumulative sum of 1D array `a` into `b`.
    """
    cdef Py_ssize_t i, n = a.shape[0]
    b[0] = a[0]
    for i in range(1, n):
        b[i] = b[i-1] + a[i]
    return b
```

```python
%%timeit
a = np.linspace(-1, 1, 1_000_000)
cumsum(a, np.empty_like(a))
```
