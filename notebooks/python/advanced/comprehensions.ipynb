## Compréhension simple

```python
d = {"a": [1, 2, 3], "b": [1, 2, 3, 4, 5]}

lengths = {}
for k, v in d.items():
    lengths[k] = len(v)
print(lengths)
```

```python
# Votre code ici
```

### Solution

```python
lengths = {k: len(v) for k, v in d.items()}
print(lengths)
```

## Compréhensions à itérations multiples

```python
a_list = [1, 2, 3]
b_list = ["a", "b", "c"]
cartesian_product = []
for a in a_list:
    for b in b_list:
        cartesian_product.append((a, b))
print(cartesian_product)
```

```python
# Votre code ici
```

### Solution

```python
cartesian_product = [(a, b) for a in a_list for b in b_list]
print(cartesian_product)
```

## Compréhensions à itérations multiples avec conditions

```python
a_list = [1, 2, 3]
b_list = ["a", "b", "c"]
cartesian_product = []
for i, a in enumerate(a_list):
    if i % 2 == 0:
        for j, b in enumerate(b_list):
            if j % 2 == 0:
                cartesian_product.append((a, b))
print(cartesian_product)
```

```python
# Votre code ici
```

### Solution

```python
cartesian_product = [
    (a, b)
    for i, a in enumerate(a_list)
    if i % 2 == 0
    for j, b in enumerate(b_list)
    if j % 2 == 0
]
print(cartesian_product)
```
