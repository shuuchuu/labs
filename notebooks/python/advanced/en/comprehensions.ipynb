# Comprehensions

In these practical exercises, you will have to reimplement the various iterations provided using comprehensions.


## Simple Comprehension

```python
d = {"a": [1, 2, 3], "b": [1, 2, 3, 4, 5]}

lengths = {}
for k, v in d.items():
  lengths[k] = len(v)
print(lengths)
```

```python
# Your code here
```

### Solution

```python
lengths = {k: len(v) for k, v in d.items()}
print(lengths)
```

## Nested Iteration Comprehensions

```python
a_list = [1, 2, 3]
b_list = ["a", "b", "c"]
cartesian_product = []
for a in a_list:
  for b in b_list:
    cartesian_product.append((a, b))
print(cartesian_product)
```

```python
# Your code here
```

### Solution

```python
cartesian_product = [(a, b) for a in a_list for b in b_list]
print(cartesian_product)
```

## Multiple Iteration Comprehensions with Conditions



```python
a_list = [1, 2, 3]
b_list = ["a", "b", "c"]
cartesian_product = []
for i, a in enumerate(a_list):
  if i % 2 == 0:
    for j, b in enumerate(b_list):
      if j % 2 == 0:
        cartesian_product.append((a, b))
print(cartesian_product)
```

```python
# Your code here
```

### Solution

```python
cartesian_product = [
    (a, b)
    for i, a in enumerate(a_list)
    if i % 2 == 0
    for j, b in enumerate(b_list)
    if j % 2 == 0
]
print(cartesian_product)
```
