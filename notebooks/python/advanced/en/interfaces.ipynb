## Informal Interfaces

Define the following informal interface and implementations:

- The interface is called `Reducer` and should have a method `reduce(items)`
- The first implementation is called `Sum` and implements `reduce` by summing the items
- The second implementation is called `Min` and implements `reduce` by finding the minimum. This implementation should accept a `key` parameter in its `__init__` method.

Test this code with a list you create.

```python
# Your code here
```

### Solution

```python
class Aggregator:
    def reduce(self, items):
        raise NotImplementedError()


class Sum(Aggregator):
    def reduce(self, items) -> float:
        return sum(items)


class Min(Aggregator):
    def __init__(self, key=None):
        self.key = key

    def reduce(self, items):
        return min(items, key=self.key)


data = [1, 2, 3]
print(Min().reduce(data))
print(Sum().reduce(data))
print(Min(key=lambda x: -x).reduce(data))
```

Consider the following class:

```python
class LegacyMaxMapReducer:
    def __init__(self, key):
        self.key = key

    def map(self, items):
        return [self.key(item) for item in items]

    def reduce(self, items):
        return max(items, key=self.key)
```

How can it be considered a subclass of `Aggregator`?

Does the `isinstance` issue persist with formal interfaces?

### Solution

```python
import abc


class Aggregator(abc.ABC):
    @abc.abstractmethod
    def reduce(self, items):
        pass


class Sum(Aggregator):
    def reduce(self, numbers) -> float:
        return sum(numbers)


class Min(Aggregator):
    def __init__(self, key=None):
        self.key = key

    def reduce(self, items):
        return min(items, key=self.key)


data = [1, 2, 3]
print(Min().reduce(data))
print(Sum().reduce(data))
print(Min(key=lambda x: -x).reduce(data))
print(isinstance(Sum(), Aggregator))
```
