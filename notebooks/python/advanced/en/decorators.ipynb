# Decorators

A decorator is a function that manipulates another function or class. It is a very useful feature in the Python language.

In these practical exercises, we will learn how to use decorators.


## Function decorators: functions that manipulate functions

Write a function `squared` that takes a function as input and returns the same function, but with the result squared:

- The original function always takes two arguments.
- The original function always returns a number.
- The `squared` function takes a function as input and returns a function as output.


```
def add(a, b):
  return a + b


def sub(a, b):
  return a - b


def mul(a, b):
  return a * b


def squared(function):
  return function  # Your code here


print(squared(add)(2, 2))  # Should be 16
print(squared(sub)(10, 5))  # Should be 25
print(squared(mul)(3, 4))  # Should be 144
```

### Solution

```
def squared(function):
  def new_function(a, b):
    return function(a, b) ** 2
  return new_function


print(squared(add)(2, 2))
print(squared(sub)(10, 5))
print(squared(mul)(3, 4))
```

## Decorator Syntax

To apply a decorator to a function, you can use the following syntax:


```
@squared
def div(a, b):
  return a / b


print(div(8, 4))
```

Without using the `@` syntax, how can we achieve the same result?


```
# Your code here
```

### Solution

```
def div(a, b):
    return a / b


div = squared(div)


print(div(8, 4))
```

## Measuring the Execution Time of a Function

Create a decorator `timeit` that displays the execution time of a function. You can use the [`time.time`](https://docs.python.org/3/library/time.html#time.time) function to help you.


```
# Your code here
```

### Solution

```
import time


def timeit(f):
  def new_f(*args, **kwargs):
    start = time.time()
    result = f(*args, **kwargs)
    print(f"Time elapsed during '{f.__name__}' call: {time.time() - start}s")
    return result
  return new_f


@timeit
def add(a, b):
  return a + b


add(1, 2)
```

## Definition of a decorator with an argument

Take the previous exercise but this time the decorator should take an argument `name` to use as the registry key instead of `cls.__name__`.


```
# Your code here
```

### Solution

```
from collections.abc import Callable
from typing import TypeVar


_plugins = {}


T = TypeVar("T")


def register(name: str) -> Callable[[type[T]], type[T]]:
  def decorator(cls: type[T]) -> type[T]:
    _plugins[name] = cls
    return cls
  return decorator


@register("B")
class A:
  pass


_plugins
```

## Using [`functools.wraps`](https://docs.python.org/3/library/functools.html#functools.wraps)

Try manipulating the modified functions in the rest of the practical exercises and find their name (attribute `__name__`) or their docstring (attribute `__doc__`). What do you notice?

Create a `timeit` decorator that takes a `unit` argument (`"s"` or `"ms"`), the unit in which to display the execution time, then use `functools.wraps` to solve this issue.

```
# Your code here
```

### Solution

```
import functools
import time


def timeit(unit="s"):
  def decorator(function):
    @functools.wraps(function)
    def new_f(*args, **kwargs):
      start = time.time()
      result = function(*args, **kwargs)
      duration = time.time() - start
      if unit == "ms":
        duration *= 1000
      elif unit != "s":
        raise ValueError("Can only use s or ms as unit argument")
      print(f"Time elapsed during '{function.__name__}' call: "
            f"{duration}{unit}")
      return result
    return new_f
  return decorator


@timeit(unit="ms")
def add(a, b):
  """Add two integers."""
  return a + b


print(add.__name__, add.__doc__)
```
