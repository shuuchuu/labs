# Design Patterns

## Builder

Implement a builder for the following class:


```python
import dataclasses
import typing


@dataclasses.dataclass
class User:
  name: str
  full_name: typing.Optional[str]
  addresses: typing.List[str]
```

### Solution

```python
class UserBuilder:
  def __init__(self):
    self._name = None
    self._full_name = None
    self._addresses = []

  def name(self, name: str) -> "UserBuilder":
    self._name = name
    return self

  def full_name(self, full_name: str) -> "UserBuilder":
    self._full_name = full_name
    return self

  def address(self, address: str) -> "UserBuilder":
    self._addresses.append(address)
    return self

  def build(self) -> User:
    if self._name is None:
      raise ValueError(
        "The name attribute must be set through a name() call before building."
      )
    return User(name=self._name, full_name=self._full_name, addresses=self._addresses)


user = (
  UserBuilder()
  .name("john")
  .full_name("John Doe")
  .address("johndoe@gmail.com")
  .address("john_doe@mit.edu")
  .build()
)
user
```

## Iterator

Implement the iterator pattern for the `Tree` class with depth-first traversal (DFS).


```python
import dataclasses
import typing


class Tree:
  def __init__(self, value, children):
    self.value = value
    self.children = children


tree = Tree("a", [Tree("b", [Tree("c", [])]), Tree("d", [])])
# print(f"DFS : {list(tree.dfs())}")
```

### Solution

```python
import dataclasses
import typing


class Tree:
  def __init__(self, value, children):
    self.value = value
    self.children = children

  def __iter__(self):
    yield self.value
    for child in self.children:
      yield from iter(child)


# Using extern class
class DepthFirstSearch:
  def __init__(self, tree):
    self._iterator = self._iterate(tree)

  @classmethod
  def _iterate(cls, tree):
    yield tree.value
    for child in tree.children:
      yield from cls._iterate(child)

  def __next__(self):
    return next(self._iterator)


class Tree:
  def __init__(self, value, children):
    self.value = value
    self.children = children

  def __iter__(self):
    return DepthFirstSearch(self)


tree = Tree("a", [Tree("b", [Tree("c", [])]), Tree("d", [])])
print(f"DFS : {list(tree)}")
```
