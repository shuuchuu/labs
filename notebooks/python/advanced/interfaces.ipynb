## Interface informelle

Définissez l'interface informelle et les implémentations suivantes :

- L'interface s'appelle `Reducer` et doit avoir une méthode `reduce(items)`
- La première implémentation s'appelle `Sum` et implémente `reduce` par la somme
- La deuxième implémentation s'appelle `Min` et implémente `reduce` par le minimum. Cette deuxième implémentation doit accepter un paramètre `key` dans sa méthode `__init__`

Testez ce code avec une liste que vous aurez créée.

```python
# Votre code ici
```

### Solution

```python
class Aggregator:
    def reduce(self, items):
        raise NotImplementedError()


class Sum(Aggregator):
    def reduce(self, items) -> float:
        return sum(items)


class Min(Aggregator):
    def __init__(self, key=None):
        self.key = key

    def reduce(self, items):
        return min(items, key=self.key)


data = [1, 2, 3]
print(Min().reduce(data))
print(Sum().reduce(data))
print(Min(key=lambda x: -x).reduce(data))
```

## Limitations de l'approche informelle

Que pouvez-vous dire du code suivant ?

```python
class Aggregator:
    def reduce(self, items):
        raise NotImplementedError()


class Sum(Aggregator):
    def sum(self, items) -> float:
        return sum(items)


class Min(Aggregator):
    def __init__(self, key=None):
        self.key = key

    def reduce(self, items):
        return min(items, key=self.key)


isinstance(Sum(), Aggregator)
```

Considérons la classe suivante :

```python
class LegacyMaxMapReducer:
    def __init__(self, key):
        self.key = key

    def map(self, items):
        return [self.key(item) for item in items]

    def reduce(self, items):
        return max(items, key=self.key)
```

Comment faire en sorte qu'elle soit considérée comme classe enfant de `Aggregator` ?

Le problème avec `isinstance` persiste-t-il avec les interfaces formelles ?

### Solution

```python
import abc


class Aggregator(abc.ABC):
    @abc.abstractmethod
    def reduce(self, items):
        pass


class Sum(Aggregator):
    def reduce(self, numbers) -> float:
        return sum(numbers)


class Min(Aggregator):
    def __init__(self, key=None):
        self.key = key

    def reduce(self, items):
        return min(items, key=self.key)


data = [1, 2, 3]
print(Min().reduce(data))
print(Sum().reduce(data))
print(Min(key=lambda x: -x).reduce(data))
print(isinstance(Sum(), Aggregator))
```
