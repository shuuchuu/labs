## Monteur

Implémentez un monteur pour la classe suivante :

```python
from dataclasses import dataclass


@dataclass
class User:
  name: str
  full_name: str | None
  addresses: list[str]
```

### Solution

```python
from typing import Self


class UserBuilder:
  def __init__(self):
    self._name = None
    self._full_name = None
    self._addresses = []

  def name(self, name: str) -> Self:
    self._name = name
    return self

  def full_name(self, full_name: str) -> Self:
    self._full_name = full_name
    return self

  def address(self, address: str) -> Self:
    self._addresses.append(address)
    return self

  def build(self) -> User:
    if self._name is None:
      raise ValueError("The name attribute must be set through a name() call "
                       "before building.")
    return User(name=self._name,
                full_name=self._full_name,
                addresses=self._addresses)


user = (UserBuilder().name("john")
                     .full_name("John Doe")
                     .address("johndoe@gmail.com")
                     .address("john_doe@mit.edu")
                     .build())
user
```

## Itérateur

Implémentez le patron itérateur pour la classe `Tree` avec le parcours en profondeur (DFS).

```python
from dataclasses import dataclass
from typing import Self


@dataclass
class Tree[T]:
    value: T
    children: list[Self]


tree = Tree("a",
            [Tree("b",
                  [Tree("c", [])]),
             Tree("d", [])])
# print(f"DFS : {list(tree.dfs())}")
```

### Solution

```python
from collections.abc import Iterator
from dataclasses import dataclass
from typing import Self


@dataclass
class Tree[T]:
    value: T
    children: list[Self]

    def __iter__(self) -> Iterator[T]:
        yield self.value
        for child in self.children:
            yield from iter(child)


# Option avec classe externe
class DepthFirstSearch[T](Iterator[T]):
    def __init__(self, tree: Tree[T]) -> None:
        self._iterator = self._iterate(tree)

    @classmethod
    def _iterate(cls, tree: Tree[T]) -> Iterator[T]:
        yield tree.value
        for child in tree.children:
            yield from cls._iterate(child)

    def __next__(self) -> T:
        return next(self._iterator)


@dataclass
class Tree[T]:
    value: T
    children: list[Self]

    def __iter__(self) -> Iterator[T]:
        return DepthFirstSearch(self)


tree = Tree("a", [Tree("b", [Tree("c", [])]), Tree("d", [])])
print(f"DFS : {list(tree)}")
```
