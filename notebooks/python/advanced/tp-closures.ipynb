# Fermetures

## Temps écoulé depuis la création d'une fermeture

Implémentez une fermeture qui renvoie le temps écoulé depuis la création de la fermeture quand on l'appelle.

````python
# Votre code ici
````

### Solution

````python
from time import sleep, time
from typing import Callable


def make_timer() -> Callable[[], float]:
  start = time()

  def timer() -> float:
    return time() - start

  return timer

timer = make_timer()
sleep(2)
print(timer())
sleep(1)
print(timer())
````

On peut récupérer les nom et valeurs des variables locales à la fermeture avec les attributs `__code__.co_freevars[index]` et les valeurs dans  `__closure__[index].cell_contents`

````python
print(timer.__code__.co_freevars, timer.__closure__[0].cell_contents)
````

## Fixer l'argument d'une fonction avec une fermeture

Par défaut, la fonction [`int`](https://docs.python.org/fr/3/library/functions.html#int) interprète les chaînes de caractères qui lui sont passées comme des nombres en base 10. Créez une fermeture qui renvoie une fonction qui crée des entiers dans la base qu'on lui a passée en argument.

````python
# Votre code ici
````

### Solution

````python
from typing import Any, Callable


def make_int_base(base: int) -> Callable[[str], int]:
  def int_base(x: str) -> int:
    return int(x, base=base)

  return int_base


int_base_2 = make_int_base(2)
print(int_base_2("10"))
print(int_base_2.__code__.co_freevars, int_base_2.__closure__[0].cell_contents)
````

## Création de plusieurs fermetures dans une boucle

Il faut faire attention lorsque l'on crée plusieurs fermetures dans une boucle. Pouvez-vous expliquer le résultat suivant ?

```python
multiplications_by_n = [lambda x: x * i for i in range(5)]


for multiplication_by_n in multiplications_by_n:
    print(multiplication_by_n(2))
```
```console
8
8
8
8
8
```

### Solution

Les fermetures se basent sur la dernière valeur de la variable `i`. C'est ce qu'on appelle le *late binding*. Il faut y faire très attention lorsque l'on crée des fermetures.

````python
multiplications_by_n = [lambda x: x * i for i in range(5)]

for multiplication_by_n in multiplications_by_n:
    print(multiplication_by_n(2))
````

````python
def binder(i):
    return lambda x: x * i

multiplications_by_n = [binder(i) for i in range(5)]

# print(*multiplications_by_n, sep="\n")

for multiplication_by_n in multiplications_by_n:
    print(multiplication_by_n(2))
````

````python
multiplications_by_n = []

for i in range(5):
    multiplications_by_n.append((lambda x, j=i: x * j))

for multiplication_by_n in multiplications_by_n:
    print(multiplication_by_n(2))
````
