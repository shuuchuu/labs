## Sequence Transformers

Generators are commonly used to transform and work with sequences without storing them entirely.

Write the following transformers that operate on sequences of integers:

- `double`: doubles the values passed to it
- `filter_even`: removes the odd numbers from the provided sequence

Test your transformers on a large sequence. Observe the memory usage of the instance.


```
# Your code here
```

### Solution

```
from typing import Iterable, Iterator


def double(iterator: Iterable[int]) -> Iterator[int]:
  for item in iterator:
    yield item * 2


def filter_even(iterator: Iterable[int]) -> Iterator[int]:
  for item in iterator:
    if item % 2:
      yield item

for i in filter_even(double(range(100_000_000))):
    pass
```

Also propose a method `take(iterator: Iterator[T], n: int) -> Iterator[T]` (which takes an iterator as input and modifies it) to use this infinite generator.

```
# Your code here
```

### Solution

```
import datetime
from typing import Iterable, Iterator, TypeVar


def gen_dates(date: datetime.date) -> Iterator[datetime.date]:
  week = datetime.timedelta(days=7)
  while True:
    date += week
    yield date


for i, d in enumerate(gen_dates(datetime.date(1900, 1, 1))):
  if i == 10:
    break
  print(d)


T = TypeVar("T")
def take(iterator: Iterable[T], n: int) -> Iterator[T]:
  for i, item in enumerate(iterator):
    if i == n:
      break
    yield item


for date in take(gen_dates(datetime.date(1900, 1, 1)), 10):
  print(date)
```

## Using the `send` method of a generator (very advanced)

Design a generator that, given an initial sum of money, will receive an interest rate from you and return the new value of the sum after applying the rate.


```
# Your code here
```

### Solution

```
from typing import Generator


def interests(amount: float) -> Generator[float, float, None]:
  while True:
    rate = (yield amount)
    amount *= rate


amounts = interests(100)
print(next(amounts))  # Generator initialization
for rate in [1.05, 1.08, 0.98]:
  print(amounts.send(rate))
```

## Using the `throw` method of a generator (very advanced)

Start from the code defined in the previous exercise and add a mechanism to add a sum of money (or withdraw with a negative sum). You can use `throw`, where the nature of the exception represents a message for the generator.


```
# Your code here
```

### Solution

```
from typing import Generator


class AddAmountException(Exception):
  def __init__(self, value):
    super().__init__(value)
    self.value = value


def interests(amount: float) -> Generator[float, float, None]:
  while True:
    try:
      rate = (yield amount)
      amount *= rate
    except AddAmountException as e:
      amount += e.value


amounts = interests(100)
amounts.send(None)  # Generator initialization
print(amounts.send(1.05))
print(amounts.send(1.08))
amounts.throw(AddAmountException(100))
print(amounts.send(0.98))
```
