# Création d'un outil de gestion de bibliothèque

## Objectif des travaux pratiques

On souhaite commencer à développer un outil de gestion d'emprunts dans une bibliothèque. Les points suivants en particulier nous intéresse :

- Modéliser la relation entre livres et membres de la bibliothèque en explicant les choix faits pour les tables créées
- Importer des données depuis des fichiers au format YAML dans la base de données. Exemple de fichier :
  ```yaml
  livres:
    - Vingt Mille Lieues sous les mers
    - Voyage au centre de la Terre
    - De la Terre à la Lune
    - Le Tour du monde en quatre-vingts jours
    - L'Île mystérieuse
  membres:
    - John Doe
    - Jane Doe
    - Sherlock Holmes
  emprunts:
    - livre: Voyage au centre de la Terre
      membre: Jane Doe
      date: "2024-01-03"
    - livre: Le Tour du monde en quatre-vingts jours
      membre: Sherlock Holmes
      date: "2024-01-05"
    - livre: L'Île mystérieuse
      membre: Sherlock Holmes
      date: "2024-01-07"
  ```
- Exporter les données en CSV
