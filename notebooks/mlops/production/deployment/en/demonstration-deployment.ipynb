# Deployment

This demonstration focuses the usage of [Docker](https://www.docker.com/) and [FastAPI](https://fastapi.tiangolo.com/).

The shown material can be found at the following address: https://dagshub.com/m09/landscape-classifier/src/solution

In particular:

- [`Dockerfile`](https://dagshub.com/m09/landscape-classifier/src/solution/Dockerfile)
- [Fast API file](https://dagshub.com/m09/landscape-classifier/src/solution/landscape_classifier/api.py)

The final command shown to start the API with docker was:

    docker run --rm -it -p 8000:80 fastapi
