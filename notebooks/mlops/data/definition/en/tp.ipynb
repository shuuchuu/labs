## Computing inter-rater agreement

Let's compute [Fleiss' kappa](https://en.wikipedia.org/wiki/Fleiss%27_kappa)

- The score is between -1 & 1
- 0 is a random agreement
- Below 0, raters agree less than random
- Conversely, above 0, raters agree more than random
- Direct interpretation is hard as it depends from the context, the score is preferably used to compare two ratings for the same problem
