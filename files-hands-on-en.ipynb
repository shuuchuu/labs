A `for` loop over the open file gives its lines one by one, each with its trailing `"\n"`. `line.strip()` removes the spaces and the line break at both ends, an empty string is false in an `if`, `startswith("#")` spots the comments and `split()` separates the date from the temperature. `readings.values()` gives the temperatures, whose minimum, maximum and sum `min`, `max` and `sum` compute. ### Solution

````python
readings = {}
with open("readings.txt", encoding="utf-8") as file:
  for line in file:
    line = line.strip()
    if not line or line.startswith("#"):
      continue
    date, temperature = line.split()
    readings[date] = float(temperature)

temperatures = readings.values()
print(f"{len(readings)} readings")
print(f"Minimum: {min(temperatures):.1f} °C, maximum: {max(temperatures):.1f} °C")
print(f"Average: {sum(temperatures) / len(temperatures):.1f} °C")
````

### Additional hints

`json.dump(obj, file)` writes to an open file, `json.dumps(obj)` returns a string. Their `indent=2` parameter indents the result and `ensure_ascii=False` keeps the accents instead of writing them as escape sequences (`\u00e9` for “é”). `json.load(file)` returns the object read back. `movies.jsonl` should have 6 lines.

```python

import json

movies = []
with open("movies.csv", encoding="utf-8", newline="") as file:
  for movie in csv.DictReader(file):
    movie["year"] = ...
    movie["duration"] = ...
    movies.append(movie)

with open("movies.json", mode="w", encoding="utf-8") as file:
  json.dump(movies, file, ...)

with open("movies.jsonl", mode="w", encoding="utf-8") as file:
  for movie in movies:
    file.write(...)

with open("movies.json", encoding="utf-8") as file:

```

### Solution

````python
import csv
import json

movies = []
with open("movies.csv", encoding="utf-8", newline="") as file:
  for movie in csv.DictReader(file):
    movie["year"] = int(movie["year"])
    movie["duration"] = int(movie["duration"])
    movies.append(movie)

with open("movies.json", mode="w", encoding="utf-8") as file:
  json.dump(movies, file, indent=2, ensure_ascii=False)

with open("movies.jsonl", mode="w", encoding="utf-8") as file:
  for movie in movies:
    file.write(json.dumps(movie, ensure_ascii=False) + "\n")

with open("movies.json", encoding="utf-8") as file:
  read_back = json.load(file)
print(read_back[1]["title"], type(read_back[1]["year"]))  # <class 'int'>: JSON keeps it

with open("movies.jsonl", encoding="utf-8") as file:
  print(file.read())
````
