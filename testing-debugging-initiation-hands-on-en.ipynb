In Colab, the debugger opens below the cell: type a command in the input field that appears, then Enter. The commands of the course:

- `n` runs the next line;
- `s` steps into the function the line calls;
- `c` continues until the next stop;
- `p expression` prints the value of an expression, for example `p i`;
- `q` quits the debugger and stops the cell.

While the debugger waits for a command, the cell is running, and no other cell runs: always end with `c` until the end, or with `q`. After `q`, a `BdbQuit` traceback shows below the cell: it's normal, not an error in your code; `c` finishes the cell without it.
````python
def count_vowels(word: str) -> int:
  """Return the number of vowels in word."""
  nb_vowels = 0
  for i in range(1, len(word)):
    letter = word[i]
    # breakpoint()
    if letter in "aeiou":
      nb_vowels = nb_vowels + 1
  return nb_vowels


print(count_vowels("apple"))  # expected: 2
print(count_vowels("cat"))  # expected: 1
print(count_vowels("orange"))  # expected: 3
````

## Stepping into a Called Function

A stationery shop computes the price of a basket with two functions: `line_price` gives the price of an item ordered in several copies, and `total_price` adds up the lines of the basket. The rule: from 10 copies of the same item on, each one costs $1 less. The basket of the next cell, 10 pens at $2 and 2 notebooks at $3, must therefore cost 10 × 1 + 2 × 3 = $16. Run the cell: it prints something else.

Quit with `q`, and write down the faulty line as a comment.
2. Put the `#` back in front of `breakpoint()`, then write a fixed version of `line_price` in the exercise cell, and check that `total_price` prints 16.

````python
prices = {"pen": 2, "notebook": 3, "binder": 5}


def line_price(item: str, quantity: int, prices: dict[str, int]) -> int:
  """Return the price of quantity copies of item."""
  unit_price = prices[item]
  if quantity > 10:
    unit_price = unit_price - 1
  return unit_price * quantity


def total_price(basket: dict[str, int], prices: dict[str, int]) -> int:
  """Return the price of the basket, which maps each item to its quantity."""
  # breakpoint()
  total = 0
  for item, quantity in basket.items():
    total = total + line_price(item, quantity, prices)
  return total


print(total_price({"pen": 10, "notebook": 2}, prices))  # expected: 16
````
