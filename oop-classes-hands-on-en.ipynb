### Additional hints

`Shape` inherits from `ABC` and marks `compute_area` with `@abstractmethod`, both imported from the `abc` module: Python then refuses to create an instance of a class that hasn't defined all its abstract methods. `__lt__` calls `compute_area`, which each subclass defines: it doesn't need to know which shape it is. from abc import ABC, abstractmethod


```python
class Shape(ABC):
  @abstractmethod
  def compute_area(self) -> float: ...

  def __lt__(self, other: object) -> bool:
    if not isinstance(other, Shape):
      return NotImplemented
    return ...


class Rhombus(Shape):
  def __init__(self, diagonal1: float, diagonal2: float) -> None:
    ...

  def compute_area(self) -> float:
    return ...

  def __repr__(self) -> str:
    return ...


try:
  Shape()
except TypeError as e:
  print(e)
```

### Solution

````python
from abc import ABC, abstractmethod


class Shape(ABC):
  @abstractmethod
  def compute_area(self) -> float: ...

  def __lt__(self, other: object) -> bool:
    if not isinstance(other, Shape):
      return NotImplemented
    return self.compute_area() < other.compute_area()


class Rectangle(Shape):
  def __init__(self, length: float, width: float) -> None:
    self.length = length
    self.width = width

  @staticmethod
  def _check_dimension(value: float) -> float:
    if value <= 0:
      raise ValueError(f"A dimension must be strictly positive, not {value}")
    return value

  @property
  def length(self) -> float:
    return self._length

  @length.setter
  def length(self, value: float) -> None:
    self._length = self._check_dimension(value)

  @property
  def width(self) -> float:
    return self._width

  @width.setter
  def width(self, value: float) -> None:
    self._width = self._check_dimension(value)

  def compute_area(self) -> float:
    return self.length * self.width

  def __repr__(self) -> str:
    return f"{type(self).__name__}(length={self.length}, width={self.width})"

  def __eq__(self, other: object) -> bool:
    if not isinstance(other, Rectangle):
      return NotImplemented
    return self.length == other.length and self.width == other.width


class Rhombus(Shape):
  def __init__(self, diagonal1: float, diagonal2: float) -> None:
    self.diagonal1 = diagonal1
    self.diagonal2 = diagonal2

  def compute_area(self) -> float:
    return self.diagonal1 * self.diagonal2 / 2

  def __repr__(self) -> str:
    return (
      f"{type(self).__name__}(diagonal1={self.diagonal1}, diagonal2={self.diagonal2})"
    )


print(Rectangle(3, 4) < Rhombus(4, 7))  # True: 12 < 14
try:
  Shape()
except TypeError as e:
  print(f"Refused: {e}")
````

### Additional hints

`super()` doesn't mean the parent class, but the class that comes next in the instance's method resolution order (MRO). For a `Square`, this order is `Square`, `Rectangle`, `Rhombus`, `Shape`, then `ABC` and `object`: `Rectangle`'s `super().__init__` then calls `Rhombus`'s. Each `__init__` therefore takes its own parameters, receives the others in `**kwargs` and passes them on to `super().__init__(**kwargs)`. `Square` passes every parameter by keyword.

```python
class Rectangle(Shape):
  def __init__(self, length: float, width: float, **kwargs: float) -> None:
    super().__init__(**kwargs)
    self.length = length
    self.width = width

  ...  # the rest unchanged


class Rhombus(Shape):
  def __init__(self, diagonal1: float, diagonal2: float, **kwargs: float) -> None:
    ...

  ...


class Square(Rectangle, Rhombus):
  def __init__(self, side: float) -> None:
    diagonal = side * math.sqrt(2)
    super().__init__(length=side, width=side, diagonal1=..., diagonal2=...)

  def __repr__(self) -> str:
    return ...


square = Square(2)
print(square, square.compute_area(), square.diagonal1)
print(Square.__mro__)
print(square == Rectangle(2, 2), Rectangle(2, 2) == square)
```

### Solution

````python
class Rectangle(Shape):
  def __init__(self, length: float, width: float, **kwargs: float) -> None:
    super().__init__(**kwargs)
    self.length = length
    self.width = width

  @staticmethod
  def _check_dimension(value: float) -> float:
    if value <= 0:
      raise ValueError(f"A dimension must be strictly positive, not {value}")
    return value

  @property
  def length(self) -> float:
    return self._length

  @length.setter
  def length(self, value: float) -> None:
    self._length = self._check_dimension(value)

  @property
  def width(self) -> float:
    return self._width

  @width.setter
  def width(self, value: float) -> None:
    self._width = self._check_dimension(value)

  def compute_area(self) -> float:
    return self.length * self.width

  def __repr__(self) -> str:
    return f"{type(self).__name__}(length={self.length}, width={self.width})"

  def __eq__(self, other: object) -> bool:
    if not isinstance(other, Rectangle):
      return NotImplemented
    return self.length == other.length and self.width == other.width


class Rhombus(Shape):
  def __init__(self, diagonal1: float, diagonal2: float, **kwargs: float) -> None:
    super().__init__(**kwargs)
    self.diagonal1 = diagonal1
    self.diagonal2 = diagonal2

  def compute_area(self) -> float:
    return self.diagonal1 * self.diagonal2 / 2

  def __repr__(self) -> str:
    return (
      f"{type(self).__name__}(diagonal1={self.diagonal1}, diagonal2={self.diagonal2})"
    )


class Square(Rectangle, Rhombus):
  def __init__(self, side: float) -> None:
    diagonal = side * math.sqrt(2)
    super().__init__(length=side, width=side, diagonal1=diagonal, diagonal2=diagonal)

  def __repr__(self) -> str:
    return f"Square(side={self.length})"


square = Square(2)
print(square, square.compute_area(), square.diagonal1)
print(Square.__mro__)
print(square == Rectangle(2, 2), Rectangle(2, 2) == square)

# 1. Rectangle.compute_area: Rectangle comes before Rhombus in Square's MRO, and
#    returns 4 (Rhombus.compute_area would return 4.000000000000001, because of the
#    square root of 2).
# 2. For a Square, the super() of Rectangle.__init__ leads to Rhombus, the next class
#    in the instance's MRO; for a Rectangle, it leads to Shape, then to object.
#    That's why each __init__ passes on the arguments it doesn't know.
# 3. True and True. Square doesn't define __eq__: it's Rectangle's, the first in the
#    MRO, that compares lengths and widths. In the second case, Square being a
#    subclass of Rectangle, Python first tries the right operand's method, which is
#    the same one.
#
# A limit of this model: Square inherits Rectangle's setters. After
# square.width = 3, the square is no longer one (area 6, diagonals unchanged), and its
# __repr__ still shows Square(side=2). It violates the Liskov substitution principle:
# an object of a subclass must be able to replace an object of the parent class
# anywhere, but code that changes a Rectangle's width without touching its length
# breaks a Square. An immutable square, or a Square class that doesn't inherit from
# Rectangle, avoids the problem.
````
