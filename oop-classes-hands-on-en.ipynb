````python
class Rectangle:
  def __init__(self, width: float, height: float) -> None:
    self.width = width
    self.height = height

  def _check_dimension(self, value: float) -> float:
    if value <= 0:
      raise ValueError(f"A dimension must be strictly positive, not {value}")
    return value

  @property
  def width(self) -> float:
    return self._width

  @width.setter
  def width(self, value: float) -> None:
    self._width = self._check_dimension(value)

  @property
  def height(self) -> float:
    return self._height

  @height.setter
  def height(self, value: float) -> None:
    self._height = self._check_dimension(value)

  def area(self) -> float:
    return self.width * self.height

  def can_contain(self, other: "Rectangle") -> bool:
    return other.width <= self.width and other.height <= self.height

  def scale(self, factor: float) -> None:
    self.width *= factor
    self.height *= factor

  def __repr__(self) -> str:
    return f"{type(self).__name__}(width={self.width}, height={self.height})"

  def __eq__(self, other: object) -> bool:
    if not isinstance(other, Rectangle):
      return NotImplemented
    return self.width == other.width and self.height == other.height


rectangle = Rectangle(2, 4)
rectangle.width = 5
print(rectangle, rectangle.area())

try:
  Rectangle(0, 3)
except ValueError as e:
  print(f"Refused: {e}")

try:
  rectangle.height = -1
except ValueError as e:
  print(f"Refused: {e}")
print(rectangle)  # Unchanged: the setter raised before assigning
````

class Shape(ABC):
  @abstractmethod
  def area(self) -> float: ...

  def __lt__(self, other: object) -> bool:
    if not isinstance(other, Shape):
      return NotImplemented
    return ...


class Rhombus(Shape):
  def __init__(self, diagonal1: float, diagonal2: float) -> None: ...

    return ...

### Solution

````python
import math
from abc import ABC, abstractmethod
from dataclasses import dataclass


class Shape(ABC):
  @abstractmethod
  def area(self) -> float: ...

  def __lt__(self, other: object) -> bool:
    if not isinstance(other, Shape):
      return NotImplemented
    return self.area() < other.area()


class Rectangle(Shape):
  def __init__(self, width: float, height: float) -> None:
    self.width = width
    self.height = height

  def _check_dimension(self, value: float) -> float:
    if value <= 0:
      raise ValueError(f"A dimension must be strictly positive, not {value}")
    return value

  @property
  def width(self) -> float:
    return self._width

  @width.setter
  def width(self, value: float) -> None:
    self._width = self._check_dimension(value)

  @property
  def height(self) -> float:
    return self._height

  @height.setter
  def height(self, value: float) -> None:
    self._height = self._check_dimension(value)

  def area(self) -> float:
    return self.width * self.height

  def can_contain(self, other: "Rectangle") -> bool:
    return other.width <= self.width and other.height <= self.height

  def scale(self, factor: float) -> None:
    self.width *= factor
    self.height *= factor

  def __repr__(self) -> str:
    return f"{type(self).__name__}(width={self.width}, height={self.height})"

  def __eq__(self, other: object) -> bool:
    if not isinstance(other, Rectangle):
      return NotImplemented
    return self.width == other.width and self.height == other.height


class Rhombus(Shape):
  def __init__(self, diagonal1: float, diagonal2: float) -> None:
    self.diagonal1 = diagonal1
    self.diagonal2 = diagonal2

  def area(self) -> float:
    return self.diagonal1 * self.diagonal2 / 2

  def __repr__(self) -> str:
    return (
      f"{type(self).__name__}(diagonal1={self.diagonal1}, diagonal2={self.diagonal2})"
    )


@dataclass
class Circle(Shape):
  radius: float

  def area(self) -> float:
    return math.pi * self.radius**2


print(Rectangle(3, 4) < Rhombus(4, 7))  # True: 12 < 14
try:
  Shape()
except TypeError as e:
  print(f"Refused: {e}")

shapes = [Rectangle(3, 4), Rhombus(4, 5), Circle(1.5), Rectangle(1, 2)]
for shape in sorted(shapes):
  print(shape, round(shape.area(), 2))
````

### Additional hints

`super()` doesn't mean the parent class, but the class that comes next in the instance's method resolution order (MRO). For a `Square`, this order is `Square`, `Rectangle`, `Rhombus`, `Shape`, then `ABC` and `object`: `Rectangle`'s `super().__init__` then calls `Rhombus`'s. Each `__init__` therefore takes its own parameters, receives the others in `**kwargs` and passes them on to `super().__init__(**kwargs)`. `Square` passes every parameter by keyword.

```python
class Rectangle(Shape):
  def __init__(self, width: float, height: float, **kwargs: float) -> None:
    super().__init__(**kwargs)
    self.width = width
    self.height = height

  ...  # the rest unchanged


class Rhombus(Shape):
  def __init__(self, diagonal1: float, diagonal2: float, **kwargs: float) -> None: ...

  ...


class Square(Rectangle, Rhombus):
  def __init__(self, side: float) -> None:
    diagonal = side * math.sqrt(2)
    super().__init__(width=side, height=side, diagonal1=..., diagonal2=...)

  def __repr__(self) -> str:
    return ...


square = Square(2)
print(square, square.area(), square.diagonal1)
print(Square.__mro__)
print(square == Rectangle(2, 2), Rectangle(2, 2) == square)
```

### Solution

````python
class Rectangle(Shape):
  def __init__(self, width: float, height: float, **kwargs: float) -> None:
    super().__init__(**kwargs)
    self.width = width
    self.height = height

  def _check_dimension(self, value: float) -> float:
    if value <= 0:
      raise ValueError(f"A dimension must be strictly positive, not {value}")
    return value

  @property
  def width(self) -> float:
    return self._width

  @width.setter
  def width(self, value: float) -> None:
    self._width = self._check_dimension(value)

  @property
  def height(self) -> float:
    return self._height

  @height.setter
  def height(self, value: float) -> None:
    self._height = self._check_dimension(value)

  def area(self) -> float:
    return self.width * self.height

  def can_contain(self, other: "Rectangle") -> bool:
    return other.width <= self.width and other.height <= self.height

  def scale(self, factor: float) -> None:
    self.width *= factor
    self.height *= factor

  def __repr__(self) -> str:
    return f"{type(self).__name__}(width={self.width}, height={self.height})"

  def __eq__(self, other: object) -> bool:
    if not isinstance(other, Rectangle):
      return NotImplemented
    return self.width == other.width and self.height == other.height


class Rhombus(Shape):
  def __init__(self, diagonal1: float, diagonal2: float, **kwargs: float) -> None:
    super().__init__(**kwargs)
    self.diagonal1 = diagonal1
    self.diagonal2 = diagonal2

  def area(self) -> float:
    return self.diagonal1 * self.diagonal2 / 2

  def __repr__(self) -> str:
    return (
      f"{type(self).__name__}(diagonal1={self.diagonal1}, diagonal2={self.diagonal2})"
    )


class Square(Rectangle, Rhombus):
  def __init__(self, side: float) -> None:
    diagonal = side * math.sqrt(2)
    super().__init__(width=side, height=side, diagonal1=diagonal, diagonal2=diagonal)

  def __repr__(self) -> str:
    return f"Square(side={self.width})"


square = Square(2)
print(square, square.area(), square.diagonal1)
print(Square.__mro__)
print(square == Rectangle(2, 2), Rectangle(2, 2) == square)

# 1. Rectangle.area: Rectangle comes before Rhombus in Square's MRO, and
#    returns 4 (Rhombus.area would return 4.000000000000001, because of the
#    square root of 2).
# 2. For a Square, the super() of Rectangle.__init__ leads to Rhombus, the next class
#    in the instance's MRO; for a Rectangle, it leads to Shape, then to object.
#    That's why each __init__ passes on the arguments it doesn't know.
# 3. True and True. Square doesn't define __eq__: it's Rectangle's, the first in the
#    MRO, that compares widths and heights. In the second case, Square being a
#    subclass of Rectangle, Python first tries the right operand's method, which is
#    the same one.
#
# A limit of this model: Square inherits Rectangle's setters. After
# square.height = 3, the square is no longer one (area 6, diagonals unchanged), and its
# __repr__ still shows Square(side=2). It violates the Liskov substitution principle:
# an object of a subclass must be able to replace an object of the parent class
# anywhere, but code that changes a Rectangle's height without touching its width
# breaks a Square. An immutable square, or a Square class that doesn't inherit from
# Rectangle, avoids the problem.
````
