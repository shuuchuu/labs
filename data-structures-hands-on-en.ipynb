### Additional hints

`hall[1]` is a list, which `hall[1][0]` indexes in turn, and a slice always returns a list, even of a single item. `copy()` creates a new list, but its items are the very objects of `hall`: it is a *shallow* copy. `is` tests whether two expressions are the same object.

```python
print(hall[2])  # Prediction: ...
print(hall[1][0])  # Prediction: ...
print(hall[1:2])  # Prediction: ...
print(hall[-1][-1])  # Prediction: ...
print(len(hall[2]))  # Prediction: ...

copy = hall.copy()
copy[1].append("B3")
copy.append(["D1", "D2"])
print(hall[1], len(hall))  # Prediction: ...
print(copy[1] is hall[1])  # Prediction: ...
```

### Solution

````python
print(hall[2])  # ['C1', 'C2', 'C3', 'C4']: the third row
print(hall[1][0])  # B1: the first seat of the second row
print(hall[1:2])  # [['B1', 'B2']]: a list holding one row
print(hall[-1][-1])  # C4
print(len(hall[2]))  # 4

copy = hall.copy()
copy[1].append("B3")
copy.append(["D1", "D2"])
print(hall[1], len(hall))  # ['B1', 'B2', 'B3'] 3
print(copy[1] is hall[1])  # True
# copy is a new list: adding row D to it leaves hall unchanged. But its items are
# the very lists of hall: row B of copy is row B of hall. For independent rows,
# each one has to be copied (copy.deepcopy does it for the whole structure).
````

### Additional hints

In a multiple assignment, the tuple on the right is built entirely before being unpacked into the names on the left. A dictionary key can be any *hashable* object: the error message names what a list lacks. You should print `green red`, then `Lyon`.

```python
a, b = ...
print(a, b)

```

### Solution

### Additional hints

`&` gives the intersection, `|` the union, `-` the difference, and `<` tests strict inclusion; each one chains like `+` (`a | b | c`). A set has no order: its items print in any order. You should find 8 features in all, a single one of which is Python's own.

```python
print(js & python)
print(...)  # Java's features that Python lacks
all_features = ...
print(all_features, len(all_features))
print(...)  # Python's own features
print({"interpreted", "object-oriented"} < python)
print(type({}))  # Prediction: ...
no_features = ...
```

### Solution

````python
print(js & python)  # {'dynamic typing', 'interpreted'}
print(java - python)  # {'compiled', 'static typing'}
all_features = js | python | java
print(all_features, len(all_features))  # 8 features
print(python - js - java)  # {'scientific computing'}
print({"interpreted", "object-oriented"} < python)  # True
print(type({}))  # <class 'dict'>: {} is an empty dictionary
no_features = set()
print(no_features, len(no_features))  # set() 0
````
