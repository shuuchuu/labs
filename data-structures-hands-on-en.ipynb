### Additional hints

`hall[1]` is a list, which `hall[1][0]` indexes in turn, and a slice always returns a list, even of a single item. `copy()` creates a new list, but its items are the very objects of `hall`: it is a *shallow* copy. `is` tests whether two expressions are the same object.

```python
print(hall[2])  # Prediction: ...
print(hall[1][0])  # Prediction: ...
print(hall[1:2])  # Prediction: ...

print(len(hall[2]))  # Prediction: ...

copy = hall.copy()
copy[1].append("B3")
copy.append(["D1", "D2"])

print(copy[1] is hall[1])  # Prediction: ...
```

### Solution

````python
print(hall[2])  # ['C1', 'C2', 'C3', 'C4']: the third row
print(hall[1][0])  # B1: the first seat of the second row
print(hall[1:2])  # [['B1', 'B2']]: a list holding one row
print(hall[-1][-1])  # C4
print(len(hall[2]))  # 4

copy = hall.copy()
copy[1].append("B3")
copy.append(["D1", "D2"])
print(hall[1], len(hall))  # ['B1', 'B2', 'B3'] 3
print(copy[1] is hall[1])  # True
# copy is a new list: adding row D to it leaves hall unchanged. But its items are
# the very lists of hall: row B of copy is row B of hall. For independent rows,
# each one has to be copied (copy.deepcopy does it for the whole structure).
````

### Additional hints

`contacts["Bob"]` is itself a dictionary, which `contacts["Bob"]["tel"]` indexes in turn, and `contacts["Alice"]["cities"]` is a list, with its own methods. `in` tests a dictionary's keys, not its values, and `get` returns `None` for a missing key. ```python

### Solution

### Additional hints

`divmod(a, b)` returns the tuple `(a // b, a % b)`: `hours, rest = divmod(duration_s, 3600)` unpacks its two items. A dictionary key can be any *hashable* object: the error message names what a list lacks. You should print `2 h 3 min 4 s`, then `Lyon`.


### Additional hints

`&` gives the intersection, `|` the union, `-` the difference, and `<=` tests inclusion (`<` strict inclusion); each one chains like `+` (`a | b | c`). A set has no order: its items print in any order. You should find 7 features in all, a single one of which is Python's own.

```python
print(js & python)
print(...)  # Java's features that Python lacks
all_features = ...
print(all_features, len(all_features))
print(...)  # Python's own features
print({"garbage collection", "class-based inheritance"} <= python)
print(type({}))  # Prediction: ...
no_features = ...
```

### Solution

````python
print(js & python)  # {'dynamic typing', 'garbage collection'}
print(java - python)  # {'static typing', 'curly braces'}
all_features = js | python | java
print(all_features, len(all_features))  # 7 features
print(python - js - java)  # {'significant indentation'}
print({"garbage collection", "class-based inheritance"} <= python)  # True
# < would test strict inclusion: True here too, but False for python <= python.
print(type({}))  # <class 'dict'>: {} is an empty dictionary
no_features = set()
print(no_features, len(no_features))  # set() 0
````
