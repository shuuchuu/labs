### Additional hints

`__name__` is `"__main__"` when the file is run as a script, and the module's name when it is imported: so the `if` block doesn't run on import. A module's docstring is a string on the first line of the file, available afterwards as `geometry.__doc__`. Python imports a module only once: if you modify `geometry.py` after importing it, reload it with `importlib.reload(geometry)`. The script should print 3.0, and `geometry.distance((3, 4, 0))` is 5.0.

```python

  a: tuple[float, float, float], b: tuple[float, float, float] = (0, 0, 0)

  ...


if __name__ == "__main__":
  print(...)
```

### Solution

````python
%%writefile geometry.py
"""Geometry computations in space."""

import math


def distance(
  a: tuple[float, float, float], b: tuple[float, float, float] = (0, 0, 0)
) -> float:
  """Return the Euclidean distance between points a and b (the origin by default)."""
  xa, ya, za = a
  xb, yb, zb = b
  return math.sqrt((xb - xa) ** 2 + (yb - ya) ** 2 + (zb - za) ** 2)


if __name__ == "__main__":
  print(distance((1, 2, 2)))
````

````python
import geometry  # Prints nothing: on import, __name__ is "geometry"

print(geometry.__doc__)  # Geometry computations in space.
print(geometry.distance((3, 4, 0)))  # 5.0
!python geometry.py
````
