### Conseils supplémentaires

La propriété se lit avec la méthode décorée par `@property`, et s'affecte avec celle décorée par `@longueur.setter`, qui porte le même nom. La valeur elle-même est rangée dans un attribut d'un autre nom, `_longueur`, sans quoi le setter s'appellerait lui-même. Dans `__init__`, `self.longueur = longueur` passe déjà par le setter. ### Solution

````python
class Rectangle:
  def __init__(self, longueur: float, largeur: float) -> None:
    self.longueur = longueur
    self.largeur = largeur

  @staticmethod
  def _verifie_dimension(valeur: float) -> float:
    if valeur <= 0:
      raise ValueError(f"Une dimension doit être strictement positive, pas {valeur}")
    return valeur

  @property
  def longueur(self) -> float:
    return self._longueur

  @longueur.setter
  def longueur(self, valeur: float) -> None:
    self._longueur = self._verifie_dimension(valeur)

  @property
  def largeur(self) -> float:
    return self._largeur

  @largeur.setter
  def largeur(self, valeur: float) -> None:
    self._largeur = self._verifie_dimension(valeur)

  def calcule_surface(self) -> float:
    return self.longueur * self.largeur

  def __repr__(self) -> str:
    return f"{type(self).__name__}(longueur={self.longueur}, largeur={self.largeur})"

  def __eq__(self, autre: object) -> bool:
    if not isinstance(autre, Rectangle):
      return NotImplemented
    return self.longueur == autre.longueur and self.largeur == autre.largeur


rectangle = Rectangle(2, 4)
rectangle.longueur = 5
print(rectangle, rectangle.calcule_surface())

try:
  Rectangle(0, 3)
except ValueError as e:
  print(f"Refusé : {e}")

try:
  rectangle.largeur = -1
except ValueError as e:
  print(f"Refusé : {e}")
print(rectangle)  # Inchangé : le setter a levé avant d'affecter
````

### Conseils supplémentaires

`Forme` hérite d'`ABC` et marque `calcule_surface` avec `@abstractmethod`, tous deux importés du module `abc` : Python refuse alors de créer une instance d'une classe qui n'a pas défini toutes ses méthodes abstraites. `__lt__` appelle `calcule_surface`, que chaque sous-classe définit : elle n'a pas à savoir de quelle forme il s'agit. Comme `__eq__`, elle renvoie `NotImplemented` quand l'autre objet n'est pas une forme. Dans `Rectangle`, seule la première ligne change : `class Rectangle(Forme):`. Attendu : 12 < 14, donc `True`.

```python
from abc import ABC, abstractmethod


class Forme(ABC):
  @abstractmethod
  def calcule_surface(self) -> float: ...

  def __lt__(self, autre: object) -> bool:
    if not isinstance(autre, Forme):
      return NotImplemented
    return ...


class Losange(Forme):
  def __init__(self, diagonale1: float, diagonale2: float) -> None:
    ...

  def calcule_surface(self) -> float:
    return ...

  def __repr__(self) -> str:
    return ...


try:
  Forme()
except TypeError as e:
  print(e)
```

### Solution

````python
from abc import ABC, abstractmethod


class Forme(ABC):
  @abstractmethod
  def calcule_surface(self) -> float: ...

  def __lt__(self, autre: object) -> bool:
    if not isinstance(autre, Forme):
      return NotImplemented
    return self.calcule_surface() < autre.calcule_surface()


class Rectangle(Forme):
  def __init__(self, longueur: float, largeur: float) -> None:
    self.longueur = longueur
    self.largeur = largeur

  @staticmethod
  def _verifie_dimension(valeur: float) -> float:
    if valeur <= 0:
      raise ValueError(f"Une dimension doit être strictement positive, pas {valeur}")
    return valeur

  @property
  def longueur(self) -> float:
    return self._longueur

  @longueur.setter
  def longueur(self, valeur: float) -> None:
    self._longueur = self._verifie_dimension(valeur)

  @property
  def largeur(self) -> float:
    return self._largeur

  @largeur.setter
  def largeur(self, valeur: float) -> None:
    self._largeur = self._verifie_dimension(valeur)

  def calcule_surface(self) -> float:
    return self.longueur * self.largeur

  def __repr__(self) -> str:
    return f"{type(self).__name__}(longueur={self.longueur}, largeur={self.largeur})"

  def __eq__(self, autre: object) -> bool:
    if not isinstance(autre, Rectangle):
      return NotImplemented
    return self.longueur == autre.longueur and self.largeur == autre.largeur


class Losange(Forme):
  def __init__(self, diagonale1: float, diagonale2: float) -> None:
    self.diagonale1 = diagonale1
    self.diagonale2 = diagonale2

  def calcule_surface(self) -> float:
    return self.diagonale1 * self.diagonale2 / 2

  def __repr__(self) -> str:
    return (
      f"{type(self).__name__}(diagonale1={self.diagonale1}, "
      f"diagonale2={self.diagonale2})"
    )


print(Rectangle(3, 4) < Losange(4, 7))  # True : 12 < 14
try:
  Forme()
except TypeError as e:
  print(f"Refusé : {e}")
````

### Conseils supplémentaires

`super()` ne désigne pas la classe parente, mais la classe qui suit dans l'ordre de résolution des méthodes (MRO) de l'instance. Pour un `Carre`, cet ordre est `Carre`, `Rectangle`, `Losange`, `Forme`, puis `ABC` et `object` : le `super().__init__` de `Rectangle` appelle alors celui de `Losange`. Chaque `__init__` prend donc ses propres paramètres, reçoit les autres dans `**kwargs` et les transmet à `super().__init__(**kwargs)`. `Carre` passe tous les paramètres par mot-clef.

```python
class Rectangle(Forme):
  def __init__(self, longueur: float, largeur: float, **kwargs: float) -> None:
    super().__init__(**kwargs)
    self.longueur = longueur
    self.largeur = largeur

  ...  # le reste sans changement


class Losange(Forme):
  def __init__(self, diagonale1: float, diagonale2: float, **kwargs: float) -> None:
    ...

  ...


class Carre(Rectangle, Losange):
  def __init__(self, cote: float) -> None:
    diagonale = cote * math.sqrt(2)
    super().__init__(longueur=cote, largeur=cote, diagonale1=..., diagonale2=...)

  def __repr__(self) -> str:
    return ...


carre = Carre(2)
print(carre, carre.calcule_surface(), carre.diagonale1)
print(Carre.__mro__)
print(carre == Rectangle(2, 2), Rectangle(2, 2) == carre)
```

### Solution

````python
class Rectangle(Forme):
  def __init__(self, longueur: float, largeur: float, **kwargs: float) -> None:
    super().__init__(**kwargs)
    self.longueur = longueur
    self.largeur = largeur

  @staticmethod
  def _verifie_dimension(valeur: float) -> float:
    if valeur <= 0:
      raise ValueError(f"Une dimension doit être strictement positive, pas {valeur}")
    return valeur

  @property
  def longueur(self) -> float:
    return self._longueur

  @longueur.setter
  def longueur(self, valeur: float) -> None:
    self._longueur = self._verifie_dimension(valeur)

  @property
  def largeur(self) -> float:
    return self._largeur

  @largeur.setter
  def largeur(self, valeur: float) -> None:
    self._largeur = self._verifie_dimension(valeur)

  def calcule_surface(self) -> float:
    return self.longueur * self.largeur

  def __repr__(self) -> str:
    return f"{type(self).__name__}(longueur={self.longueur}, largeur={self.largeur})"

  def __eq__(self, autre: object) -> bool:
    if not isinstance(autre, Rectangle):
      return NotImplemented
    return self.longueur == autre.longueur and self.largeur == autre.largeur


class Losange(Forme):
  def __init__(self, diagonale1: float, diagonale2: float, **kwargs: float) -> None:
    super().__init__(**kwargs)
    self.diagonale1 = diagonale1
    self.diagonale2 = diagonale2

  def calcule_surface(self) -> float:
    return self.diagonale1 * self.diagonale2 / 2

  def __repr__(self) -> str:
    return (
      f"{type(self).__name__}(diagonale1={self.diagonale1}, "
      f"diagonale2={self.diagonale2})"
    )


class Carre(Rectangle, Losange):
  def __init__(self, cote: float) -> None:
    diagonale = cote * math.sqrt(2)
    super().__init__(
      longueur=cote, largeur=cote, diagonale1=diagonale, diagonale2=diagonale
    )

  def __repr__(self) -> str:
    return f"Carre(cote={self.longueur})"


carre = Carre(2)
print(carre, carre.calcule_surface(), carre.diagonale1)
print(Carre.__mro__)
print(carre == Rectangle(2, 2), Rectangle(2, 2) == carre)

# 1. Rectangle.calcule_surface : Rectangle précède Losange dans le MRO de Carre, et
#    renvoie 4 (Losange.calcule_surface renverrait 4.000000000000001, à cause de la
#    racine de 2).
# 2. Pour un Carre, le super() de Rectangle.__init__ mène à Losange, la classe suivante
#    dans le MRO de l'instance ; pour un Rectangle, il mène à Forme, puis à object.
#    C'est pourquoi chaque __init__ transmet les arguments qu'il ne connaît pas.
# 3. True et True. Carre ne définit pas __eq__ : c'est celle de Rectangle, la première
#    du MRO, qui compare longueurs et largeurs. Dans le second cas, Carre étant une
#    sous-classe de Rectangle, Python essaie d'abord la méthode de l'opérande de droite,
#    qui est la même.
#
# Limite de ce modèle : Carre hérite des mutateurs de Rectangle. Après
# carre.largeur = 3, le carré n'en est plus un (surface 6, diagonales inchangées), et
# son __repr__ affiche encore Carre(cote=2). Il viole le principe de substitution de
# Liskov : un objet d'une sous-classe doit pouvoir remplacer un objet de la classe mère
# partout, or un code qui change la largeur d'un Rectangle sans toucher à sa longueur
# casse un Carre. Un carré immuable, ou une classe Carre qui n'hérite pas de Rectangle,
# évite le problème.
````
