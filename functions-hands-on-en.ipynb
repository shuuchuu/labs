### Solution

````python
import math


def distance(
  a: tuple[float, float, float], b: tuple[float, float, float] = (0, 0, 0)
) -> float:
  """Return the Euclidean distance between points a and b (the origin by default)."""
  xa, ya, za = a
  xb, yb, zb = b
  return math.sqrt((xb - xa) ** 2 + (yb - ya) ** 2 + (zb - za) ** 2)


print(distance((1, 1, 1)), math.sqrt(3))  # 1.7320508075688772 twice
print(distance((1, 2, 3), (4, 6, 3)))  # 5.0
print(distance(b=(1, 2, 2), a=(0, 0, 0)))  # 3.0
````

### Additional hints

A parameter and a variable assigned in a function are local: they shadow the global variable of the same name without modifying it. A default value is computed only once, when Python runs the `def`: all the calls that use it therefore share the same list. The usual fix takes `None` as the default and creates the list in the function's body. The `global` keyword lets a function assign a global variable.   return tasks

### Solution

````python
# 1. The predictions:
# display()                default: the parameter takes its default value
# display(message)         global: the global variable is passed as an argument
# modify()                 local: message is a variable local to modify
# print(message)           global: modify didn't touch the global variable
# add_task("groceries")    ['groceries']
# add_task("cleaning")     ['groceries', 'cleaning']: the default list is shared
# add_task("garden", …)    ['watering', 'garden']: a list is given
# add_task("dishes")       ['groceries', 'cleaning', 'dishes']: the shared list


# 2. None as the default value, and a new list on every call
def add_task_fixed(task: str, tasks: list[str] | None = None) -> list[str]:
  if tasks is None:
    tasks = []
  tasks.append(task)
  return tasks


print(add_task_fixed("groceries"))  # ['groceries']
print(add_task_fixed("cleaning"))  # ['cleaning']
print(add_task_fixed("garden", ["watering"]))  # ['watering', 'garden']


# 3. global: the assignment targets the module's variable
def modify_global() -> None:
  global message
  message = "modified by the function"


modify_global()
print(message)  # modified by the function
# With global, any function can change the variable: to understand its value, you
# have to read the whole program. A function that returns a value says what it
# produces in its signature, and the caller decides what to do with it.
````

### Additional hints

The middle of a list is at index `len(bins) // 2`. The slice `bins[:middle]` is the left half, without the middle, and `bins[middle + 1:]` the right half, without the middle either. Each recursive call must work on a shorter list: that's what guarantees reaching a base case. A function that calls itself endlessly raises `RecursionError` after about 1000 nested calls. You should get `True`, `True`, `False` and `False`, and 4 calls to find bin 41.

```python
def is_free(bins: list[int], number: int) -> bool:
  if ...:  # base case: empty list
    return False
  middle = len(bins) // 2
  if bins[middle] == number:  # base case: found
    return ...
  if number < bins[middle]:
    return is_free(..., number)  # recursive call on the left half
  return is_free(..., number)  # recursive call on the right half


print(is_free(free_bins, 41))
print(is_free(free_bins, 91))
print(is_free(free_bins, 40))
print(is_free(free_bins, 100))
```

### Solution

````python
# 1. Two base cases, then a recursive call on one half, without the middle
def is_free(bins: list[int], number: int) -> bool:
  """Return True if number is in the sorted list bins, by binary search."""
  if len(bins) == 0:  # base case: nothing to search
    return False
  middle = len(bins) // 2
  if bins[middle] == number:  # base case: found
    return True
  if number < bins[middle]:
    return is_free(bins[:middle], number)
  return is_free(bins[middle + 1 :], number)


print(is_free(free_bins, 41))  # True
print(is_free(free_bins, 91))  # True
print(is_free(free_bins, 40))  # False
print(is_free(free_bins, 100))  # False

# 2. For bin 41, print(bins) prints 4 lists, each half as long as the previous one:
# the 16 bins, [3, 8, 12, 15, 21, 27, 33, 41], [27, 33, 41], then [41].
# Going through the list from the start compares 41 with 8 bins. Binary search makes at
# most 6 calls for 16 bins, and one more each time the list doubles: about ten for
# 1000 bins, where going through the list can take 1000 comparisons.

# 3. With bins[middle:], the right half keeps the middle: for bin 95, the list ends up
# reduced to [91], and each call calls itself again with that same list. The base case
# is never reached, and Python raises
# RecursionError: maximum recursion depth exceeded
````

In the function, `details` is the tuple of the extra positional arguments, and `context` the dictionary of the keyword arguments that match no parameter. A parameter placed after `*details` can only be passed by keyword. `" | ".join(list)` joins strings with a separator.

### Solution

````python
def log(message: str, *details: str, level: str = "INFO", **context) -> str:
  """Return a log line: level, message, details and context."""
  line = f"[{level}] {message}"
  for detail in details:
    line += f" | {detail}"
  if context:
    pairs = [f"{key}={value}" for key, value in context.items()]
    line += f" ({', '.join(pairs)})"
  return line


print(log("Startup"))
print(log("Item picked", "Keyboards", "bin B4", arm=2))
print(log("Low battery", level="ALERT", arm=1, charge="12%"))
print(log("Test", "ALERT"))  # [INFO] Test | ALERT
# "ALERT" is passed by position: it is a detail. level, placed after *details, can
# only get a value by keyword.
print(log(message="Stop"))  # [INFO] Stop
# message is an ordinary parameter: it can also be passed by keyword. Declared before
# /, it could only be passed by position: message="Stop" would go into context, and
# the call would raise TypeError: log() missing 1 required positional argument:
# 'message'.
````

### Solution

````python
from collections.abc import Iterator


def useful_lines(path: str) -> Iterator[str]:
  """Yield the lines of path that are neither empty nor comments, stripped."""
  with open(path, encoding="utf-8") as file:
    for line in file:
      line = line.strip()
      if line and not line.startswith("#"):
        yield line


measures = [float(line) for line in useful_lines("humidity.txt")]
average = sum(measures) / len(measures)
print(average)  # 61.8
print([measure for measure in measures if measure > average])  # [62.5, 64.0, 63.0]

lines = useful_lines("humidity.txt")
print(lines)  # <generator object useful_lines at 0x...>: nothing read yet
print(next(lines))  # 62.5: the body runs up to the first yield
print(list(lines))  # ['64.0', '58.5', '61.0', '63.0']: the remaining lines
print(list(lines))  # []: the generator is exhausted
````

````python
def closest(
  positions: dict[str, tuple[float, float, float]],
  n: int = 3,
  start: tuple[float, float, float] = (0, 0, 0),
) -> list[str]:
  """Return the n items closest to start, from the closest to the farthest."""
  ordered = sorted(positions, key=lambda item: distance(positions[item], start))
  return ordered[:n]


print(closest(positions))  # ['Cables', 'Keyboards', 'Headsets']
print(closest(positions, n=2, start=positions["Keyboards"]))
# ['Keyboards', 'Headsets']
distances = {item: round(distance(position), 1) for item, position in positions.items()}
print(distances)
````
