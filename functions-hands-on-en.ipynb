### Additional hints

A parameter and a variable assigned in a function are local: they shadow the global variable of the same name without modifying it. A default value is computed only once, when Python runs the `def`: all the calls that use it therefore share the same list. The usual fix takes `None` as the default and creates the list in the function's body. The `global` keyword lets a function assign a global variable. You should get `['groceries']`, `['cleaning']` and `['watering', 'garden']` with the fixed version.

```python
def add_task_fixed(task: str, tasks: list[str] | None = None) -> list[str]:
  if ...:
    tasks = []
  tasks.append(task)
  return tasks


print(add_task_fixed("groceries"))
print(add_task_fixed("cleaning"))
print(add_task_fixed("garden", ["watering"]))

  ...
  message = "modified by the function"


modify_global()
print(message)
```

### Solution

````python
# 1. The predictions:
# display()                default: the parameter takes its default value
# display(message)         global: the global variable is passed as an argument
# modify()                 local: message is a variable local to modify
# print(message)           global: modify didn't touch the global variable
# add_task("groceries")    ['groceries']
# add_task("cleaning")     ['groceries', 'cleaning']: the default list is shared
# add_task("garden", …)    ['watering', 'garden']: a list is given
# add_task("dishes")       ['groceries', 'cleaning', 'dishes']: the shared list


# 2. None as the default value, and a new list on every call
def add_task_fixed(task: str, tasks: list[str] | None = None) -> list[str]:
  if tasks is None:
    tasks = []
  tasks.append(task)
  return tasks


print(add_task_fixed("groceries"))  # ['groceries']
print(add_task_fixed("cleaning"))  # ['cleaning']
print(add_task_fixed("garden", ["watering"]))  # ['watering', 'garden']


# 3. global: the assignment targets the module's variable
def modify_global() -> None:
  global message
  message = "modified by the function"


modify_global()
print(message)  # modified by the function
# With global, any function can change the variable: to understand its value, you
# have to read the whole program. A function that returns a value says what it
# produces in its signature, and the caller decides what to do with it.
````

A docstring is a triple-quoted string placed as the first statement of the function's body: a first line that sums up what the function does, then, after an empty line, the details. An argument passed by keyword (`exponent=3`) can come in any order. You should get 9, 1024 and 8.

````python
def power(x: float, exponent: float = 2) -> float:
  """Return x raised to the power exponent.

  x is the number to raise, exponent the power: 2 (the square) by default.
  """
  return x**exponent


print(power(3))  # 9
print(power(2, 10))  # 1024
print(power(exponent=3, x=2))  # 8
print(power(9, 0.5))  # 3.0: the square root, a float
help(power)
````

A tuple unpacks into as many variables as it has items (`xa, ya, za = a`). A default value can be a tuple: it is immutable, so without the trap of the first exercise's list. `math.sqrt(3)` is about 1.732.

````python
import math


def distance(
  a: tuple[float, float, float], b: tuple[float, float, float] = (0, 0, 0)
) -> float:
  """Return the Euclidean distance between points a and b (the origin by default)."""
  xa, ya, za = a
  xb, yb, zb = b
  return math.sqrt((xb - xa) ** 2 + (yb - ya) ** 2 + (zb - za) ** 2)


print(distance((1, 1, 1)), math.sqrt(3))  # 1.7320508075688772 twice
print(distance((1, 2, 3), (4, 6, 3)))  # 5.0
print(distance(b=(1, 2, 2), a=(0, 0, 0)))  # 3.0
````

### Additional hints

`next(file)` reads and skips the header line, and `split(";")` splits a line. The `key` parameter of `sorted` takes a function called on each item: the items are sorted by what it returns. Sorting the dictionary sorts its keys, the item names. You should get `['Cables', 'Keyboards', 'Headsets']` from the base, then `['Keyboards', 'Headsets']`, and 3.0 for the cables in `distances`.

```python
def read_positions(path: str) -> dict[str, tuple[float, float, float]]:

positions = read_positions("bins.csv")
print(closest(positions))
print(closest(positions, n=..., start=positions[...]))

```

### Solution

````python
def read_positions(path: str) -> dict[str, tuple[float, float, float]]:
  """Read a positions file and return the dictionary item -> (x, y, z)."""
  positions = {}
  with open(path, encoding="utf8") as file:
    next(file)  # The header line
    for line in file:
      item, x, y, z = line.strip().split(";")
      positions[item] = (float(x), float(y), float(z))
  return positions


def closest(
  positions: dict[str, tuple[float, float, float]],
  n: int = 3,
  start: tuple[float, float, float] = (0, 0, 0),
) -> list[str]:
  """Return the n items closest to start, from the closest to the farthest."""
  ordered = sorted(positions, key=lambda item: distance(positions[item], start))
  return ordered[:n]


positions = read_positions("bins.csv")
print(closest(positions))  # ['Cables', 'Keyboards', 'Headsets']
print(closest(positions, n=2, start=positions["Keyboards"]))
# ['Keyboards', 'Headsets']
distances = {item: round(distance(position), 1) for item, position in positions.items()}
print(distances)
````

### Additional hints

In the function, `details` is the tuple of the extra positional arguments, and `context` the dictionary of the keyword arguments that match no parameter. A parameter placed after `*details` can only be passed by keyword. `" | ".join(list)` joins strings with a separator.

```python
def log(message: str, *details: str, level: str = "INFO", **context) -> str:
  line = f"[{level}] {message}"
  for detail in details:
    line += ...
  if context:
    pairs = [f"{key}={value}" for key, value in ...]
    line += ...

```

### Solution

````python
def log(message: str, *details: str, level: str = "INFO", **context) -> str:
  """Return a log line: level, message, details and context."""
  line = f"[{level}] {message}"
  for detail in details:
    line += f" | {detail}"
  if context:
    pairs = [f"{key}={value}" for key, value in context.items()]
    line += f" ({', '.join(pairs)})"
  return line


print(log("Startup"))
print(log("Item picked", "Keyboards", "bin B4", arm=2))
print(log("Low battery", level="ALERT", arm=1, charge="12%"))
print(log("Test", "ALERT"))  # [INFO] Test | ALERT
# "ALERT" is passed by position: it is a detail. level, placed after *details, can
# only get a value by keyword.
````

### Additional hints

A function that contains `yield` is a generator: calling it doesn't run its body, but returns an object that you go through with a `for` loop, a comprehension or `next`. Each `yield` produces a value and suspends the function until the next request. You should find an average of 61.8.

```python
      line = line.strip()

```

### Solution

````python
from collections.abc import Iterator


def useful_lines(path: str) -> Iterator[str]:
  """Yield the lines of path that are neither empty nor comments, stripped."""
  with open(path, encoding="utf8") as file:
    for line in file:
      line = line.strip()
      if line and not line.startswith("#"):
        yield line


measures = [float(line) for line in useful_lines("humidity.txt")]
average = sum(measures) / len(measures)
print(average)  # 61.8
print([measure for measure in measures if measure > average])  # [62.5, 64.0, 63.0]

lines = useful_lines("humidity.txt")
print(lines)  # <generator object useful_lines at 0x...>: nothing read yet
print(next(lines))  # 62.5: the body runs up to the first yield
````
