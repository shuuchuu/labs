### Conseils supplémentaires

`salle[1]` est une liste, que `salle[1][0]` indice à son tour, et une tranche renvoie toujours une liste, même d'un seul élément. `copy()` crée une nouvelle liste, mais ses éléments sont les mêmes objets que ceux de `salle` : c'est une copie *superficielle*. `is` teste si deux expressions désignent le même objet.

```python
print(salle[2])  # Prédiction : ...
print(salle[1][0])  # Prédiction : ...
print(salle[1:2])  # Prédiction : ...
print(salle[-1][-1])  # Prédiction : ...
print(len(salle[2]))  # Prédiction : ...

copie = salle.copy()
copie[1].append("B3")
copie.append(["D1", "D2"])
print(salle[1], len(salle))  # Prédiction : ...
print(copie[1] is salle[1])  # Prédiction : ...
```

### Solution

````python
print(salle[2])  # ['C1', 'C2', 'C3', 'C4'] : la troisième rangée
print(salle[1][0])  # B1 : le premier siège de la deuxième rangée
print(salle[1:2])  # [['B1', 'B2']] : une liste qui contient une rangée
print(salle[-1][-1])  # C4
print(len(salle[2]))  # 4

copie = salle.copy()
copie[1].append("B3")
copie.append(["D1", "D2"])
print(salle[1], len(salle))  # ['B1', 'B2', 'B3'] 3
print(copie[1] is salle[1])  # True
# copie est une nouvelle liste : lui ajouter la rangée D ne change pas salle. Mais
# ses éléments sont les mêmes listes que ceux de salle : la rangée B de copie est
# celle de salle. Pour des rangées indépendantes, il faut copier chacune d'elles
# (copy.deepcopy le fait pour toute la structure).
````

### Conseils supplémentaires

`divmod(a, b)` renvoie le n-uplet `(a // b, a % b)` : `heures, reste = divmod(duree_s, 3600)` en déballe les deux éléments. Une clef de dictionnaire peut être tout objet *hashable* : le message d'erreur nomme ce qui manque à une liste. Vous devez afficher `2 h 3 min 4 s`, puis `Lyon`.

```python
heures, reste = divmod(duree_s, ...)
minutes, secondes = divmod(...)
print(heures, "h", minutes, "min", secondes, "s")

villes = {(48.857, 2.352): "Paris", ...: "Lyon"}

x, y = (1, 2, 3)  # Prédiction : ...
```

### Solution

````python
heures, reste = divmod(duree_s, 3600)
minutes, secondes = divmod(reste, 60)
print(heures, "h", minutes, "min", secondes, "s")  # 2 h 3 min 4 s

villes = {(48.857, 2.352): "Paris", (45.764, 4.836): "Lyon"}
villes[(43.296, 5.370)] = "Marseille"
print(villes[(45.764, 4.836)])  # Lyon

# {[48.857, 2.352]: "Paris"} lève TypeError: unhashable type: 'list'. Une clef doit
# être hashable : sa valeur ne doit pas pouvoir changer une fois rangée dans le
# dictionnaire. Une liste est modifiable, un n-uplet de nombres ne l'est pas :
# coordonnees = (48.857, 2.352)
# coordonnees[0] = 48.86 lève TypeError: 'tuple' object does not support item
# assignment.

# x, y = (1, 2, 3) lève ValueError: too many values to unpack : trois éléments
# pour deux noms. Un déballage demande autant de noms que le n-uplet a d'éléments.
````

### Conseils supplémentaires

`&` donne l'intersection, `|` l'union, `-` la différence, `<=` teste l'inclusion (`<` l'inclusion stricte) ; chacun s'enchaîne comme `+` (`a | b | c`). Un ensemble n'a pas d'ordre : ses éléments s'affichent dans un ordre quelconque. Vous devez trouver 7 caractéristiques en tout, dont une seule propre au Python.

```python
print(js & python)
print(...)  # Celles du Java absentes du Python
toutes = ...
print(toutes, len(toutes))
print(...)  # Celles du Python seul
print({"ramasse-miettes", "héritage par classes"} <= python)
print(type({}))  # Prédiction : ...
aucune = ...
```

### Solution

````python
print(js & python)  # {'typage dynamique', 'ramasse-miettes'}
print(java - python)  # {'typage statique', 'accolades'}
toutes = js | python | java
print(toutes, len(toutes))  # 7 caractéristiques
print(python - js - java)  # {'indentation significative'}
print({"ramasse-miettes", "héritage par classes"} <= python)  # True
# < testerait l'inclusion stricte : True ici aussi, mais False pour python <= python.
print(type({}))  # <class 'dict'> : {} est un dictionnaire vide
aucune = set()
print(aucune, len(aucune))  # set() 0
````
