## 3. Formatting numbers

In an f-string, a format placed after `:` controls how a value is displayed: `f"{value:format}"`. With f-strings:

1. display `math.pi` with two decimal places (`3.14`);
2. display `420.671` and `math.pi` one below the other, with three decimal places, aligned on their point:

3. display a success rate of `0.875` as a percentage, with one decimal place (`87.5%`);
4. display the order number `42` on 5 digits, padded with zeros (`00042`).

The strings' `format` method accepts the same formats: `"{:.2f}".format(math.pi)`.

**Useful documentation**

- [Format Specification Mini-Language](https://docs.python.org/3/library/string.html#formatspec)

````python
import math

# Your code here
````

### Additional hints

The formats to use:

- `.2f`: a `float` with two decimal places
- a number before the point sets the minimum width, in characters: `7.3f` takes 7 characters, point and decimals included, right-aligned. Same width and same number of decimals: the points line up
- `%` multiplies by 100 and adds the `%` sign: `.1%` keeps one decimal place
- `05d`: an integer on 5 characters, padded on the left with zeros

A skeleton to copy into the exercise cell, replacing each `...`:

```python
import math

print(f"{math.pi:...}")
print(f"{420.671:...}")
print(f"{math.pi:...}")
print(f"{0.875:...}")
print(f"{42:...}")
```

### Solution

````python
import math

print(f"{math.pi:.2f}")
# Same width (7) and same number of decimals (3): the points line up
print(f"{420.671:7.3f}")
print(f"{math.pi:7.3f}")
print(f"{0.875:.1%}")
print(f"{42:05d}")
````

## 4. A receipt

From the given variables, display this receipt with f-strings:

```
Receipt #007
Keyboard      2 x   49.90    99.80
Mouse         1 x   19.90    19.90
USB-C cable   3 x    7.50    22.50
----------------------------------
Total                       142.20
incl. VAT 20.0%              23.70
```

An item line gives the name, left-aligned on 12 characters, then the quantity, the unit price and the line total, right-aligned. The total and the VAT (included in the prices) are computed, not written by hand: change a price, and the whole receipt follows.

````python
number = 7
item1 = "Keyboard"
quantity1 = 2
price1 = 49.90
item2 = "Mouse"
quantity2 = 1
price2 = 19.90
item3 = "USB-C cable"
quantity3 = 3
price3 = 7.50
vat_rate = 0.2  # VAT included in the prices

# Your code here
````

### Additional hints

The formats and calculations to use:

- `<12` left-aligns on 12 characters, `>9.2f` right-aligns on 9 characters with two decimal places
- an expression can go inside the braces: `f"{quantity1 * price1:>9.2f}"`
- `03d` pads an integer with zeros, `.1%` displays a percentage
- the prices include VAT: the amount before tax is `total / (1 + vat_rate)`, and the VAT the difference with `total`
- an item line is 12 + 3 + 3 + 7 + 9 = 34 characters long; the last line is built in two steps: its label in a variable, then that label aligned on 25 characters

A skeleton to copy into the exercise cell, replacing each `...`:

```python
total = quantity1 * price1 + quantity2 * price2 + quantity3 * price3
vat = total - total / (1 + vat_rate)
print(f"Receipt #{number:...}")
print(f"{item1:<12}{quantity1:>3} x {price1:>7.2f}{quantity1 * price1:>9.2f}")
print(...)
print(...)
print("-" * 34)
print(f"{'Total':<25}{total:...}")
vat_label = f"incl. VAT {vat_rate:...}"
print(f"{vat_label:...}{vat:...}")
```

### Solution

````python
total = quantity1 * price1 + quantity2 * price2 + quantity3 * price3
vat = total - total / (1 + vat_rate)
print(f"Receipt #{number:03d}")
print(f"{item1:<12}{quantity1:>3} x {price1:>7.2f}{quantity1 * price1:>9.2f}")
print(f"{item2:<12}{quantity2:>3} x {price2:>7.2f}{quantity2 * price2:>9.2f}")
print(f"{item3:<12}{quantity3:>3} x {price3:>7.2f}{quantity3 * price3:>9.2f}")
print("-" * 34)
print(f"{'Total':<25}{total:>9.2f}")
vat_label = f"incl. VAT {vat_rate:.1%}"
print(f"{vat_label:<25}{vat:>9.2f}")
````
