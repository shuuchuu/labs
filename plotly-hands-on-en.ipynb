### 2.2 A surface: the saddle point

The function $f(x, y) = x^2 - y^2$ has a *saddle point* at $(0, 0)$: its slope there is zero in every direction, as at a minimum, and yet it isn't one. Optimization algorithms, such as gradient descent, often run into them.

#### Additional hints

`np.meshgrid(x, y)` returns two arrays of 81 × 81 values, the x and the y of each point of the grid: $f$ is computed on them in a single expression, with no loop. `go.Surface` takes one-dimensional `x` and `y`, and a two-dimensional `z`. `update_traces(contours_z={"show": True, "project_z": True})` draws the contour lines of `z` and projects them onto the bottom plane.

```python
x = np.linspace(-2, 2, 81)
y = ...
X, Y = np.meshgrid(x, y)
Z = ...
fig_saddle = go.Figure(go.Surface(x=x, y=y, z=Z))
fig_saddle.update_traces(contours_z=...)
fig_saddle.show()
```

#### Solution

````python
x = np.linspace(-2, 2, 81)
y = np.linspace(-2, 2, 81)
X, Y = np.meshgrid(x, y)
Z = X**2 - Y**2
fig_saddle = go.Figure(go.Surface(x=x, y=y, z=Z))
fig_saddle.update_traces(contours_z={"show": True, "project_z": True})
fig_saddle.update_layout(title="f(x, y) = x² − y²")
fig_saddle.show()
# Along the x axis (y = 0), f is x²: (0, 0) is a minimum there.
# Along the y axis (x = 0), f is −y²: (0, 0) is a maximum there.
````

### Additional hints

`fig.write_html(path)` writes the page. Its `include_plotlyjs` argument says how the page gets plotly.js, the JavaScript library that draws the chart: copied into the page (`True`, the default) or downloaded when it opens from a CDN, a public server (`"cdn"`). `Path(path).stat().st_size` gives a file's size in bytes. The default page weighs several megabytes.

```python

  print(path, ..., "kB")
```

### Solution

````python
fig_gdp.write_html("gdp.html")
fig_gdp.write_html("gdp_cdn.html", include_plotlyjs="cdn")
for path in ["gdp.html", "gdp_cdn.html"]:
  print(path, Path(path).stat().st_size // 1000, "kB")
# gdp.html holds plotly.js (more than 4 MB): it opens with no connection.
# gdp_cdn.html (about fifteen kB) holds only the data and downloads plotly.js
# when it opens: much lighter, but it needs the Internet.
````
