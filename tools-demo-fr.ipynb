## Notebooks

Les notebooks sont des environnements interactifs de travail, vous pouvez y exécuter du code et y afficher du texte. Ils sont très pratiques pour :

1. Itérer rapidement sur du code et des données en phase de prototypage d'une étude ou d'un projet
1. Présenter les résultats de travaux avec tout l'avantage de code exécutable au service de graphiques / représentations.

Un notebook est une suite de cellules. Vous pouvez à tout moment supprimer ou ajouter des cellules, puis les exécuter (si vous avez les droits d'édition).

### Cellules textuelles

La syntaxe est celle du Markdown, un langage très simple permettant de décrire du texte formaté ; le détail suit.

Double-cliquez sur une cellule textuelle pour voir son Markdown. Le [guide officiel](https://colab.research.google.com/notebooks/markdown_guide.ipynb) donne le reste, tableaux compris.

#### Portée

La portée des variables (là où elles sont visibles) est particulière en Python : une variable n'est pas visible seulement dans son bloc (un `if`, une boucle), mais dans toute la fonction, ou tout le module, où elle est définie. En revanche, une affectation dans une fonction crée une variable locale à cette fonction, qui ne modifie pas la variable du même nom hors d'elle.

````python
booleen = True

if True:
  booleen = False


def fonction():
  booleen = True


# fonction() affecte sa propre variable locale booleen
fonction()

# Affiche False : le if a modifié booleen, fonction() n'y a pas touché
print(booleen)

if True:
  nouvelle_variable = "Hello World!"

# Hors du bloc de définition, ne compilerait pas en Java par exemple
print(nouvelle_variable)
````

### Compréhensions de listes et générateurs

On peut définir des listes très rapidement grâce aux *compréhensions de listes* en Python. Elles utilisent la syntaxe

```python
[resultat for variable in liste if condition]
```
où `resultat` et `condition` utilisent `variable`.

````python
carres = [k**2 for k in range(10)]
print(carres)

carres_pairs = [k for k in carres if k % 2 == 0]
print(carres_pairs)

for i, j in [(k, k**2) for k in range(10)]:
  print(i, j)
````

Une expression génératrice s'écrit comme une compréhension de liste, mais avec des parenthèses au lieu des crochets.

La différence est que la liste définie par le générateur est calculée seulement quand elle est utilisée, et ne peut être utilisée qu'une fois (elle n'est pas stockée en mémoire mais calculée à la volée).

````python
def generateur_carres_pairs():
  return (k**2 for k in range(10) if k % 2 == 0)


carres_pairs = generateur_carres_pairs()

for valeur in carres_pairs:
  print(valeur)

print("***")

# Ne fait rien : le générateur a déjà été utilisé
for valeur in carres_pairs:
  print(valeur)

print("***")

# Il faut recréer le générateur pour ré-itérer
for valeur in generateur_carres_pairs():
  print(valeur)
````

### Manipulation de fichier

Pour manipuler des fichiers, on utilise un bloc `with`, qui s'assure que la ressource est bien refermée même en présence d'exceptions.

La cellule suivante crée un petit fichier texte à lire : la commande magique `%%writefile` écrit le reste de la cellule dans un fichier.

````python
%%writefile lorem.txt
Lorem ipsum dolor sit amet, consectetur adipiscing elit.
Sed do eiusmod tempor incididunt ut labore et dolore magna aliqua.
Ut enim ad minim veniam, quis nostrud exercitation ullamco laboris.
Duis aute irure dolor in reprehenderit in voluptate velit esse cillum.
Excepteur sint occaecat cupidatat non proident, sunt in culpa qui officia.
````

````python
from pathlib import Path

with Path("lorem.txt").open(encoding="utf8") as fichier:
  contenu = fichier.read()

lignes = contenu.splitlines()
print(lignes[3])
````

Pour écrire le mécanisme est le même mais on précise le mode *writing* `w` :

````python
from pathlib import Path

with Path("sortie.txt").open(encoding="utf8", mode="w") as fichier:
  fichier.write("Hello World!")

print(Path("sortie.txt").read_text(encoding="utf8"))
````

### Annotations de type

Complètement en option : ces annotations ne changent pas le sens du code exécuté. Elles peuvent être ajoutées pour documenter le sens d'une fonction et pour améliorer l'efficacité des outils d'aide au développement.

````python
def carre(x: int) -> int:
  return x**2


for i in range(10):
  print(i, carre(i))

# carre("Hello World") : un éditeur signale que l'argument n'est pas un int,
# mais Python ne vérifie pas les annotations : l'appel lèverait TypeError sur **
````

### Dépaquetage de liste et de dictionnaire

L'opérateur `*` permet d'empaqueter des éléments épars en tuple, ou de dépaqueter des listes en éléments épars, l'opérateur `**` est l'équivalent pour les dictionnaires.

````python
def somme_des_carres(x, y):
  return x**2 + y**2


print(somme_des_carres(*[2, 4]))


# *args empaquette "Greetings:" et "Hello" en ("Greetings:", "Hello")
# **kwargs empaquette qui="World!" en {"qui": "World!"}
def salue(*args, **kwargs):
  # *args dépaquette ("Greetings:", "Hello") en "Greetings:", "Hello"
  print(*args, kwargs["qui"])


salue("Greetings:", "Hello", qui="World!")
# **{...} dépaquette le dictionnaire en arguments nommés : qui="World!"
salue("Greetings:", "Hello", **{"qui": "World!"})
````
