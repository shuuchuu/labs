## Contrôle du caractère de fin

La fonction `print` a plusieurs arguments utiles. Le premier est l'argument `end`, qui permet de contrôler le caractère affiché après l'argument passé à `print`. Par défaut, c'est un retour à la ligne.

*Appelez la fonction `print` plusieurs fois à la suite et faites en sorte que les affichages successifs se fassent sur la même ligne.*

```
# Votre code ici
```

### Solution

```
print("Hello", end=" ")
print("World")
```

## Flottants

En plus du cours, vous pouvez utiliser https://pyformat.info/ comme référence pour la suite de ces travaux pratiques.

En utilisant la méthode `.format` des chaînes de caractères, affichez pi avec seulement ses deux premières décimales.

```
import math


print(math.pi)  # Votre modification ici
```

Affichez maintenant `420.671` et pi de manière à ce que les deux nombres soient alignés sur leur `.` central.

```
# Votre code ici
```

### Solution

```
import math


# Pi avec deux décimales
print("{:.2f}".format(math.pi))

# Alignement sur le point : même largeur totale et même nombre de décimales
print("{:7.3f}".format(420.671))
print("{:7.3f}".format(math.pi))
```

## `f-strings`

Depuis Python 3.6, les [`f-strings`](https://docs.python.org/fr/3/tutorial/inputoutput.html#tut-f-strings) permettent de réaliser des affichages complexes sans utiliser la fonction `.format` qui est parfois un peu lourde syntaxiquement.

Réalisez les affichages de l'exercice précédent avec des `f-string`s.

```
# Votre code ici
```

### Solution

```
print(f"{math.pi:.2f}")

print(f"{420.671:7.3f}")
print(f"{math.pi:7.3f}")
```
