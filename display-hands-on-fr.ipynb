### Conseils supplémentaires

Les formats à utiliser :

- `.2f` : un `float` avec deux décimales
- un nombre devant le point fixe la largeur minimale, en caractères : `7.3f` occupe 7 caractères, point et décimales compris, aligné à droite. Même largeur et même nombre de décimales : les points s'alignent
- `%` multiplie par 100 et ajoute le signe `%` : `.1%` garde une décimale
- `05d` : un entier sur 5 caractères, complété à gauche par des zéros

Un squelette à copier dans la cellule de l'exercice, en remplaçant chaque `...` :

```python

print(f"{math.pi:...}")
print(f"{420.671:...}")
print(f"{math.pi:...}")
print(f"{0.875:...}")
print(f"{42:...}")
```

### Solution

````python
import math

print(f"{math.pi:.2f}")
# Même largeur (7) et même nombre de décimales (3) : les points s'alignent
print(f"{420.671:7.3f}")
print(f"{math.pi:7.3f}")
print(f"{0.875:.1%}")
print(f"{42:05d}")
````

À partir des variables données, affichez ce ticket de caisse avec des f-strings :

```
Ticket n°007
Clavier       2 x   49.90    99.80

Total                       142.20
```

Une ligne d'article donne le nom, aligné à gauche sur 12 caractères, puis la quantité, le prix unitaire et le total de la ligne, alignés à droite. Le total et la TVA (comprise dans les prix) sont calculés, pas écrits à la main : changez un prix, tout le ticket suit.

````python
numero = 7
article1 = "Clavier"
quantite1 = 2
prix1 = 49.90
article2 = "Souris"
quantite2 = 1
prix2 = 19.90
article3 = "Câble USB-C"
quantite3 = 3
prix3 = 7.50
taux_tva = 0.2  # TVA comprise dans les prix

# Votre code ici
````

### Conseils supplémentaires

Les formats et les calculs à utiliser :

- `<12` aligne à gauche sur 12 caractères, `>9.2f` aligne à droite sur 9 caractères avec deux décimales
- une expression peut se placer dans les accolades : `f"{quantite1 * prix1:>9.2f}"`
- `03d` complète un entier par des zéros, `.1%` affiche un pourcentage
- les prix sont TTC : le montant hors taxes vaut `total / (1 + taux_tva)`, et la TVA la différence avec `total`
- une ligne d'article fait 12 + 3 + 3 + 7 + 9 = 34 caractères ; la dernière ligne se construit en deux temps : son libellé dans une variable, puis ce libellé aligné sur 25 caractères

Un squelette à copier dans la cellule de l'exercice, en remplaçant chaque `...` :

```python
total = quantite1 * prix1 + quantite2 * prix2 + quantite3 * prix3
tva = total - total / (1 + taux_tva)
print(f"Ticket n°{numero:...}")
print(f"{article1:<12}{quantite1:>3} x {prix1:>7.2f}{quantite1 * prix1:>9.2f}")
print(...)
print(...)
print("-" * 34)

libelle_tva = f"dont TVA {taux_tva:...}"

```

### Solution

````python
total = quantite1 * prix1 + quantite2 * prix2 + quantite3 * prix3
tva = total - total / (1 + taux_tva)
print(f"Ticket n°{numero:03d}")
print(f"{article1:<12}{quantite1:>3} x {prix1:>7.2f}{quantite1 * prix1:>9.2f}")
print(f"{article2:<12}{quantite2:>3} x {prix2:>7.2f}{quantite2 * prix2:>9.2f}")
print(f"{article3:<12}{quantite3:>3} x {prix3:>7.2f}{quantite3 * prix3:>9.2f}")
print("-" * 34)
print(f"{'Total':<25}{total:>9.2f}")
libelle_tva = f"dont TVA {taux_tva:.1%}"
print(f"{libelle_tva:<25}{tva:>9.2f}")
````
