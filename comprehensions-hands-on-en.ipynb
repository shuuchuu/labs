## Simple Comprehension

The loop below stores in `lengths` the number of items of each list of `d`. Build the same dictionary, `{'a': 3, 'b': 5}`, with a dictionary comprehension.

````python
d = {"a": [1, 2, 3], "b": [1, 2, 3, 4, 5]}

lengths = {}
for k, v in d.items():
  lengths[k] = len(v)
print(lengths)
````

````python
# Your code here
````

### Solution

````python
lengths = {k: len(v) for k, v in d.items()}
print(lengths)
````

## File

The loop below counts, line by line, the characters of the source code of the standard library's `string` module. Compute this total with `sum`, without building a list of the lengths: on a file of several million lines, it would fill the memory for nothing. You should get the number the loop prints.

````python
import pathlib
import string

# The source code of the standard library's string module
source = pathlib.Path(string.__file__)
with source.open(encoding="utf8") as fh:
  total = 0
  for line in fh:
    total += len(line)
print(total)
````

````python
# Your code here
````

### Solution

````python
with source.open(encoding="utf8") as fh:
  total = sum(len(line) for line in fh)
print(total)
````

## Nested Iteration Comprehensions

The loop below builds the Cartesian product of `numbers` and `letters`: every `(number, letter)` pair. Build the same list of 9 pairs, in the same order, with a single list comprehension.

````python
numbers = [1, 2, 3]
letters = ["a", "b", "c"]
cartesian_product = []
for a in numbers:
  for b in letters:
    cartesian_product.append((a, b))
print(cartesian_product)
````

````python
# Your code here
````

### Solution

````python
cartesian_product = [(a, b) for a in numbers for b in letters]
print(cartesian_product)
````

`matrix` has two rows of three values. With one comprehension each time:

1. flatten it into `flat`, the list of its values, row after row: `[1, 2, 3, 4, 5, 6]`;
2. transpose it into `transposed`, the list of its columns: `[[1, 4], [2, 5], [3, 6]]`.

Both comprehensions have two `for`: what makes one give a flat list, and the other a list of lists?

````python
matrix = [[1, 2, 3], [4, 5, 6]]
````

### Additional hints

To flatten, write both `for` in the same comprehension, in the order of the nested loops: first each `row` of `matrix`, then each `value` of `row`. To transpose, column `j` is `[row[j] for row in matrix]`: a second comprehension, around it, builds it for each `j` of `range(len(matrix[0]))`, the column indices. Two `for` in the same comprehension give a flat list; a comprehension placed in the result of another gives a list of lists.

```python
flat = [value for row in matrix for value in ...]
transposed = [[... for row in matrix] for j in range(len(matrix[0]))]
```

### Solution

````python
flat = [value for row in matrix for value in row]
print(flat)
# A comprehension in the result of another: one list per column
transposed = [[row[j] for row in matrix] for j in range(len(matrix[0]))]
print(transposed)
````

## Distinct Words

`words` holds the words of a sentence, some of them twice, and "The" with a capital letter. Build `distinct`, these words without duplicates and in lowercase, with the right comprehension. Print `sorted(distinct)`: you should get `['and', 'cat', 'dog', 'runs', 'sees', 'the']`.

````python
text = "The cat sees the dog and the cat runs"
words = text.split()
print(words)
````

````python
# Your code here
````

### Additional hints

A comprehension between `{}`, with no `:`, builds a set, which keeps a single copy of each value: no need to test whether a word is already in it. `word.lower()` puts a word in lowercase. `sorted` returns a sorted list, since the display order of a set isn't guaranteed.

```python
distinct = {... for word in words}
print(sorted(distinct))
```

### Solution

````python
distinct = {word.lower() for word in words}
print(sorted(distinct))
````

## Inverting a Dictionary

`capitals` gives the capital of each country. Build `country_by_capital`, the inverse dictionary, which gives the country of each capital: `country_by_capital["Rome"]` should be `'Italy'`. What would the result become if two countries had the same capital?

````python
capitals = {"France": "Paris", "Italy": "Rome", "Spain": "Madrid"}
````

````python
# Your code here
````

### Additional hints

`capitals.items()` gives the `(country, capital)` pairs; a dictionary comprehension puts them the other way round, the capital as the key. If two countries had the same capital, the same key would be written twice, and only the last country would remain.

```python
country_by_capital = {...: ... for country, capital in capitals.items()}
```

### Solution

````python
country_by_capital = {capital: country for country, capital in capitals.items()}
print(country_by_capital)
print(country_by_capital["Rome"])
````

## Replacing Negative Values

`measurements` holds a sensor's readings, whose negative values are errors. Build `non_negative`, the same list where each negative value is replaced by 0: `[3, 0, 4, 0, 5, 0, 2]`. The list keeps its length: this isn't a filter.

### Additional hints

An `if` at the end of a comprehension filters: it removes the items that don't meet the condition. To keep every item while changing its value, the conditional expression `a if condition else b` takes the place of the result, at the start of the comprehension. An `if … else` at the end is a `SyntaxError`.

```python
non_negative = [x if ... else ... for x in measurements]
```

### Solution

````python
non_negative = [x if x >= 0 else 0 for x in measurements]
print(non_negative)
````

## Measuring with `timeit`

Does a comprehension make code faster? Write three functions with no argument that return the list of the squares of `values`, the integers from 0 to 999:

1. `with_loop`, with a `for` loop and `append`;

Check that they return the same list, then measure each one with `timeit.timeit` over 1000 calls, and print the duration of one call in microseconds. The durations vary from one run to the next: run the cell again before concluding. What do you conclude?

````python
import timeit

values = list(range(1000))
````

### Additional hints

`timeit.timeit(function, number=1000)` calls `function` 1000 times and returns the total duration, in seconds: divided by 1000 then multiplied by `1e6`, it gives microseconds per call. A loop over the list `[with_loop, with_comprehension, with_map]` saves writing the measurement three times, and `function.__name__` gives the name to print.

```python
def with_loop() -> list[int]:
  squares = []
  for x in values:
    squares.append(...)
  return squares


def with_comprehension() -> list[int]:
  return [...]


def with_map() -> list[int]:
  return list(map(lambda x: ..., values))


print(with_loop() == with_comprehension() == with_map())
for function in [with_loop, with_comprehension, with_map]:
  duration = timeit.timeit(function, number=...)
  print(f"{function.__name__}: {duration / 1000 * 1e6:.0f} µs per call")
```

### Solution

````python
def with_loop() -> list[int]:
  squares = []
  for x in values:
    squares.append(x * x)
  return squares


def with_comprehension() -> list[int]:
  return [x * x for x in values]


def with_map() -> list[int]:
  return list(map(lambda x: x * x, values))


print(with_loop() == with_comprehension() == with_map())
for function in [with_loop, with_comprehension, with_map]:
  duration = timeit.timeit(function, number=1000)
  print(f"{function.__name__}: {duration / 1000 * 1e6:.0f} µs per call")

# Loop and comprehension: same order of magnitude, the comprehension slightly ahead
# or even, depending on the machine and the Python version. `map` + `lambda` is the
# slowest: one function call per value. A comprehension is chosen for its
# readability, not for a speed gain.
````
