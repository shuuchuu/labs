### Additional hints

In the loop, first test for the end of the transmission (`break`), then for a missing reading (count it, then `continue` moves on to the next one); the rest of the loop only sees valid readings. You should get 6 valid readings, 2 missing ones and a mean of `3.33` mm. For the question, try `not 0.0`.

```python
total = 0

for reading in rain_mm:
  if reading == -1:
    ...
  if reading is None:
    n_missing += 1
    ...

```

### Solution

````python
total = 0
n_valid = 0
n_missing = 0
for reading in rain_mm:
  if reading == -1:
    break
  if reading is None:
    n_missing += 1
    continue
  total += reading
  n_valid += 1
print(f"{n_valid} valid readings, {n_missing} missing")
print(f"Mean: {total / n_valid:.2f} mm")

# `not reading` is also true for 0.0, a day without rain: a valid reading.
# Testing `not reading` would drop the two dry days, and the mean would become
# 20.0 / 4 = 5.00 mm. `is None` only targets the missing reading.
````

### Additional hints

`zip(altitudes, altitudes[1:])` pairs each reading with the next one: `(1020, 1085)`, `(1085, 1160)`… For each pair, the difference `after - before` is a climb if it is positive, a descent otherwise. ```python

```python
  difference = after - before
  if difference > 0:
    gain += difference
    if ...:
      steepest_climb = difference

```

### Solution

````python
gain = 0
loss = 0
steepest_climb = 0
for before, after in zip(altitudes, altitudes[1:]):
  difference = after - before
  if difference > 0:
    gain += difference
    if difference > steepest_climb:
      steepest_climb = difference
  else:
    loss += -difference
print(f"Gain: {gain} m, loss: {loss} m")
print(f"Steepest climb: {steepest_climb} m")

# itertools.pairwise(altitudes) gives the same pairs as this zip.
# The gain minus the loss is the end altitude minus the start altitude:
print(gain - loss, altitudes[-1] - altitudes[0])  # 260 260
````

### Additional hints

Two loops: the first one, over `visits.values()`, finds the record, the second one, over `visits.items()`, prints. A string multiplied by an integer is repeated: `"#" * 3` is `"###"`, and `95 // 10` is `9`. The `:<10` format aligns a string to the left on 10 characters.

```python
record = 0
for count in visits.values():
  if ...:
    record = count

for day, count in visits.items():
  line = f"{day:<10}{...} {count}"
  if ...:
    line += " <- record"
  print(line)
```

### Solution

````python
record = 0
for count in visits.values():
  if count > record:
    record = count

for day, count in visits.items():
  line = f"{day:<10}{'#' * (count // 10)} {count}"
  if count == record:
    line += " <- record"
  print(line)
````

### Additional hints

To see each pair only once, pair each team with the teams that follow it in the list only: with `enumerate`, `teams[i + 1:]` gives the teams after the one at index `i`. Home and away, each team plays all the others, and you only have to leave out a team playing itself. With 5 teams, you should get 10 matches, then 20.

```python
number = 0

  for team_2 in ...:
    number += 1
    print(f"Match {number}: {team_1} – {team_2}")

n_home_and_away = 0
```

### Solution

````python
number = 0
for i, team_1 in enumerate(teams):
  for team_2 in teams[i + 1 :]:
    number += 1
    print(f"Match {number}: {team_1} – {team_2}")

n_home_and_away = 0
for home in teams:
  for away in teams:
    if home != away:
      n_home_and_away += 1
print(n_home_and_away, "home-and-away matches")

# Each of the n teams hosts the n - 1 others: n(n - 1) home-and-away matches,
# and half as many, n(n - 1) / 2, single matches.
````

### Solution

````python
if payment <= principal * monthly_rate:
  print("Payment too low: the loan would never be paid back")
else:
  balance = principal
  n_months = 0
  total_interest = 0
  while balance > 0:
    interest = balance * monthly_rate
    balance += interest
    total_interest += interest
    if balance < payment:
      balance = 0
    else:
      balance -= payment
    n_months += 1
  print(f"Paid back in {n_months} months, interest: {total_interest:.2f} €")

# With 40 €, the payment only just covers the month's interest (10,000 × 0.004):
# the balance never goes down, and the loop would never stop.
````

### Additional hints

Iterate over the possible hours with `range`; `hour not in busy` tests whether an hour is free. A loop's `else` clause only runs if the loop went all the way through without a `break`: that is where we know that no hour fitted. A two-hour meeting starts at 15 at the latest. You should find 12 for one hour, and no slot for two hours.

```python
for hour in range(...):
  if hour not in busy:
    print(f"First free hour: {hour}")
    ...
else:
  print("No free slot")

for hour in range(...):
  if ...:
    print(f"First two-hour slot: {hour}")
    break
else:
  print("No two-hour slot")
```

### Solution

````python
for hour in range(9, 17):
  if hour not in busy:
    print(f"First free hour: {hour}")
    break
else:
  print("No free slot")

for hour in range(9, 16):
  if hour not in busy and hour + 1 not in busy:
    print(f"First two-hour slot: {hour}")
    break
else:
  print("No two-hour slot")
````
