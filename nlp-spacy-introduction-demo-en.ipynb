# Introduction to spaCy

## Model download

To use spaCy, we first need to download a pre-trained model: here `en_core_web_sm`, the small English model.

```python
!python -m spacy download en_core_web_sm  # small English model
import spacy
```

## Processing of a document

```python
# Creation of the object that processes documents: the model's pipeline
nlp = spacy.load("en_core_web_sm")
```

```python
# Test text
text = (
  "Barack Obama was the 44th president of the United States. "
  "John ate a bad apple. "
  "a sentence with no uppercase first letter and no punctuation"
)

# Run the pipeline on our text
doc = nlp(text)
```

## Tokenization

```python
for token in doc:
  print(token.text)
```

## Retrieving sentences

```python
for sent in doc.sents:
  print("==================")
  print(sent)
```

Sentences come from the syntactic parse, not from punctuation alone: the third one, with no uppercase letter and no period, is properly isolated.

## Token details and POS tags

```python
# Print the tokens of each sentence, their start position in the original
# text, their part-of-speech tag (POS) and their morphological features
for sent in doc.sents:
  print("=============")
  for token in sent:
    print(f"{token.text:15} {token.idx:<3} {token.pos_:5} {token.morph}")
```

`token.morph` gives the token's morphological features: mood, tense and person of `was` (`Mood=Ind|Number=Sing|Person=3|Tense=Past|VerbForm=Fin`). English inflects little: beyond verbs, the features are mostly the number of nouns (`Number=Sing`) and the definiteness of determiners. The fine-grained tag (`NNP`, `VBD`) is in `token.tag_`.

## Lemmatization

```python
for sent in doc.sents:
  print("=============")
  for token in sent:
    print(f"{token.text:15} {token.lemma_}")
```

Irregular forms get back to their lemma (`was` → `be`, `ate` → `eat`); `44th` stays as it is.

## Named entities

```python
for entity in doc.ents:
  print(f"{entity.text:20} {entity.label_}")
```

```python
spacy.displacy.render(doc, style="ent", jupyter=True)
```

The English label set is rich (18 labels): `44th` and `first` are `ORDINAL`, `the United States` is a `GPE` (a country, city or state).

## Syntactic parsing

- `text`: original text.
- `dep_`: syntactic relation between the token and its head (the word it depends on) in the tree.
- `head.text`: original text of the head.
- `head.pos_`: POS of the head.
- `children`: list of the words that depend on the token.

The root of each sentence (`ROOT`) is its own head: `was` in the first one; `Barack` has `Obama` as its head, and `Obama` has `was`.

```python
for token in doc:
  print(
    f"{token.text:15} {token.dep_:10} {token.head.text:15} "
    f"{token.head.pos_:5} {' '.join(map(str, token.children))}"
  )
```

## Rendering of the syntactic parse

```python
spacy.displacy.render(doc, style="dep", jupyter=True, options=dict(distance=100))
```
