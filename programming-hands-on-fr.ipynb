### Solution

````python
# 1. Pseudo-code de convertir
# Entrées : montant, un nombre ; depuis et vers, deux codes de devise
# Sortie : le montant converti dans la devise vers, arrondi au centime
# Instructions :
#   en_euros ← montant / taux[depuis]
#   resultat ← en_euros × taux[vers]
#   RENVOYER arrondi(resultat, 2)


# 2. Chaque ligne du pseudo-code devient une ligne de Python
def convertir(montant: float, depuis: str, vers: str) -> float:
  """Renvoie montant converti de la devise depuis à la devise vers, au centime."""
  en_euros = montant / taux[depuis]
  resultat = en_euros * taux[vers]
  return round(resultat, 2)


print(convertir(100, "USD", "EUR"))  # 90.91
print(convertir(100, "EUR", "JPY"))  # 16000.0
print(convertir(50, "GBP", "CHF"))  # 55.88

# 3. Pseudo-code de total_depenses
# Entrées : depenses, une liste de n-uplets (montant, devise) ; vers, une devise
# Sortie : le total des dépenses dans la devise vers, arrondi au centime
# Instructions :
#   total ← 0
#   POUR CHAQUE (montant, devise) DANS depenses FAIRE
#     total ← total + convertir(montant, devise, vers)
#   FIN POUR
#   RENVOYER arrondi(total, 2)


def total_depenses(depenses: list[tuple[float, str]], vers: str) -> float:
  """Renvoie le total de depenses, converti dans la devise vers, au centime."""
  total = 0
  for montant, devise in depenses:
    total = total + convertir(montant, devise, vers)
  return round(total, 2)


print(total_depenses(depenses, "EUR"))  # 324.53
print(total_depenses(depenses, "USD"))  # 356.97
````

## Tri à bulles

La liste `references`, que crée la cellule suivante, donne les références des articles d'un magasin, dans l'ordre où ils ont été saisis. Le tri à bulles les range par ordre croissant : il parcourt la liste en comparant chaque élément à son voisin de droite, et les échange s'ils sont dans le désordre. À la fin d'un passage, le plus grand élément est arrivé au bout, comme une bulle qui remonte : le passage suivant peut s'arrêter un cran plus tôt. En pseudo-code, avec des indices qui commencent à 0, comme en Python (le pseudo-code du cours numérote à partir de 1), et des bornes de `POUR` incluses :

```text
Entrée : valeurs, une liste de nombres
Sortie : une nouvelle liste, les valeurs triées par ordre croissant
Instructions :
  triee ← une copie de valeurs
  n ← Taille(triee)
  POUR passage DE 0 À n - 2 FAIRE
    POUR i DE 0 À n - 2 - passage FAIRE
      SI triee[i] > triee[i + 1] ALORS

      FIN SI
    FIN POUR
  FIN POUR
  RENVOYER triee
```

### Solution

````python
def tri_a_bulles(valeurs: list[int]) -> list[int]:
  """Renvoie une nouvelle liste, les éléments de valeurs par ordre croissant."""
  triee = valeurs.copy()  # trier une copie laisse valeurs inchangée
  n = len(triee)
  for passage in range(n - 1):  # passage de 0 à n - 2
    for i in range(n - 1 - passage):  # i de 0 à n - 2 - passage
      if triee[i] > triee[i + 1]:
        triee[i], triee[i + 1] = triee[i + 1], triee[i]
  return triee


references_triees = tri_a_bulles(references)
print(references_triees)
# [1033, 1876, 2219, 2540, 3398, 4127, 5503, 6075, 6612, 7781, 8264, 9025]
print(references_triees == sorted(references))  # True
print(references)  # inchangée : [4127, 1033, 7781, ...]

# 3. Le premier passage compare 11 couples de voisins, le deuxième 10, ..., le
# dernier 1 : 11 + 10 + ... + 1 = 12 * 11 / 2 = 66 comparaisons. Pour N éléments,
# (N - 1) + (N - 2) + ... + 1 = N(N - 1) / 2 : une complexité en O(N²), comme le tri
# par sélection du cours.
````

### Solution

````python
def recherche_lineaire(valeurs: list[int], cible: int) -> int | None:
  """Renvoie l'indice du premier élément de valeurs égal à cible, ou None."""
  for i in range(len(valeurs)):
    if valeurs[i] == cible:
      return i
  return None


def recherche_dichotomique(valeurs: list[int], cible: int) -> int | None:
  """Renvoie un indice de cible dans la liste triée valeurs, ou None."""
  bas = 0
  haut = len(valeurs) - 1
  while bas <= haut:  # cible ne peut être qu'entre les indices bas et haut
    milieu = (bas + haut) // 2
    if valeurs[milieu] == cible:
      return milieu
    if valeurs[milieu] < cible:
      bas = milieu + 1  # cible ne peut être qu'à droite du milieu
    else:
      haut = milieu - 1  # cible ne peut être qu'à gauche du milieu
  return None


for reference in [6075, 9025, 5000]:
  print(
    reference,
    recherche_lineaire(inventaire, reference),
    recherche_dichotomique(inventaire, reference),
  )
# 6075 7 7
# 9025 11 11
# 5000 None None

# 4. Chaque tour de recherche_dichotomique divise à peu près par deux le nombre
# d'éléments qui restent à examiner : 12, puis au plus 6, 3 et 1, soit au plus 4
# tours pour 12 références. Pour un million, au plus 20 tours (2 ** 20 = 1 048 576) :
# une complexité en O(log N). recherche_lineaire fait un tour par élément pour une
# référence absente : 12 tours, et un million pour un million de références, en O(N).
````

## Jeu d'essai et cas limites

````python
def verifier_tri(tri) -> None:
  """Vérifie la fonction de tri tri sur un jeu d'essai, cas limites compris."""
  assert tri([]) == []  # liste vide
  assert tri([7]) == [7]  # un seul élément
  assert tri([3, 1, 3, 2, 1]) == [1, 1, 2, 3, 3]  # doublons
  assert tri([1, 2, 3]) == [1, 2, 3]  # déjà triée
  assert tri([3, 2, 1]) == [1, 2, 3]  # triée à l'envers
  valeurs = [3, 1, 2]
  tri(valeurs)
  assert valeurs == [3, 1, 2]  # la liste reçue n'a pas changé
  print("Tous les essais du tri passent")


def verifier_recherche(recherche) -> None:
  """Vérifie la fonction de recherche recherche sur un jeu d'essai."""
  assert recherche([], 5) is None  # liste vide
  assert recherche([7], 7) == 0  # un seul élément, présent
  assert recherche([7], 5) is None  # un seul élément, absent
  assert recherche([2, 4, 6, 8], 2) == 0  # au début
  assert recherche([2, 4, 6, 8], 8) == 3  # à la fin
  assert recherche([2, 4, 6, 8], 5) is None  # absente, entre deux valeurs
  assert recherche([2, 4, 6, 8], 9) is None  # absente, après la dernière
  doublons = [1, 3, 3, 3, 5]
  assert doublons[recherche(doublons, 3)] == 3  # doublons : un indice de la cible
  print("Tous les essais de la recherche passent")


verifier_tri(tri_a_bulles)
verifier_tri(sorted)
verifier_recherche(recherche_lineaire)
verifier_recherche(recherche_dichotomique)

# 3. verifier_recherche(recherche_boguee) lève une AssertionError à l'essai
# recherche([7], 7) == 0 : sur une liste d'un seul élément, bas et haut valent 0, la
# condition bas < haut est fausse dès le départ, et la boucle ne regarde jamais cet
# élément. Il faut bas <= haut : quand bas == haut, il reste un élément à examiner.
# verifier_recherche(recherche_boguee)
````

## Mesurer les temps d'exécution

La complexité d'un algorithme dit comment son temps d'exécution grandit avec la taille de ses données : pour le voir, on mesure ce temps pour des tailles croissantes. `time.perf_counter()` renvoie l'heure d'une horloge précise, en secondes : la durée d'un calcul est la différence entre l'heure après et l'heure avant. ### Solution

````python
def tri_rapide(valeurs: list[int]) -> list[int]:
  """Renvoie une nouvelle liste, les éléments de valeurs triés, par tri rapide."""
  if len(valeurs) <= 1:  # cas de base : déjà triée
    return valeurs.copy()
  pivot = valeurs[len(valeurs) // 2]
  plus_petits = []
  egaux = []
  plus_grands = []
  for valeur in valeurs:
    if valeur < pivot:
      plus_petits.append(valeur)
    elif valeur == pivot:
      egaux.append(valeur)
    else:
      plus_grands.append(valeur)
  return tri_rapide(plus_petits) + egaux + tri_rapide(plus_grands)


verifier_tri(tri_rapide)

durees_rapide = []
for taille in tailles_tri:
  nombres = valeurs_au_hasard(taille)
  debut = time.perf_counter()
  tri_rapide(nombres)
  durees_rapide.append(time.perf_counter() - debut)
tracer(
  tailles_tri,
  {"tri à bulles": durees_bulles, "sorted": durees_sorted, "tri rapide": durees_rapide},
  "Durée des tris",
)

# 4. Le tri à bulles prend environ 4 fois plus de temps quand la taille double :
# 100 000 éléments, 50 fois plus que 2000, lui demandent 50² = 2500 fois plus de
# temps : plusieurs minutes, quand 2000 éléments prennent quelques dixièmes de
# seconde.
nombres = valeurs_au_hasard(100_000)
debut = time.perf_counter()
tri_rapide(nombres)
print(time.perf_counter() - debut)  # quelques dixièmes de seconde
# Le tri rapide est en O(N log N) en moyenne. Son pire cas est en O(N²) : quand le
# pivot est chaque fois le plus petit ou le plus grand élément, l'une des deux listes
# reçoit tous les autres. Le pivot du milieu l'évite sur une liste déjà triée, où
# prendre le premier élément comme pivot y mènerait.
````
