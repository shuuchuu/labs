### Conseils supplémentaires

`__name__` vaut `"__main__"` quand le fichier est exécuté comme un script, et le nom du module quand il est importé : le bloc `if` ne s'exécute donc pas à l'import. La docstring d'un module est une chaîne en première ligne du fichier, accessible ensuite par `geometrie.__doc__`. Python n'importe un module qu'une fois : si vous modifiez `geometrie.py` après l'avoir importé, rechargez-le avec `importlib.reload(geometrie)`. Le script doit afficher 3.0, et `geometrie.distance((3, 4, 0))` vaut 5.0.

```python
%%writefile geometrie.py
"""..."""

import math


def distance(
  a: tuple[float, float, float], b: tuple[float, float, float] = (0, 0, 0)
) -> float:
  ...


if __name__ == "__main__":
  print(...)
```

Puis, dans une autre cellule :

```python
import geometrie

print(geometrie.__doc__)
print(geometrie.distance(...))  # Prédiction : ...
!python geometrie.py
```

### Solution

````python
%%writefile geometrie.py
"""Calculs de géométrie dans l'espace."""

import math


def distance(
  a: tuple[float, float, float], b: tuple[float, float, float] = (0, 0, 0)
) -> float:
  """Renvoie la distance euclidienne entre les points a et b (l'origine par défaut)."""
  xa, ya, za = a
  xb, yb, zb = b
  return math.sqrt((xb - xa) ** 2 + (yb - ya) ** 2 + (zb - za) ** 2)


if __name__ == "__main__":
  print(distance((1, 2, 2)))
````

````python
import geometrie  # N'affiche rien : à l'import, __name__ vaut "geometrie"

print(geometrie.__doc__)  # Calculs de géométrie dans l'espace.
print(geometrie.distance((3, 4, 0)))  # 5.0
!python geometrie.py
````
