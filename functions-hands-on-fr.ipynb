Une fonction sans `return` renvoie `None`, même quand elle affiche quelque chose : afficher n'est pas renvoyer. Calculer avec ce `None` lève une `TypeError`. Une fonction qui renvoie sa valeur laisse l'appelant l'afficher ou calculer avec. L'absence de valeur se teste avec `is None`. Vous devez obtenir 13.5, puis « Pas de notes ».

### Solution

````python
# 1. afficher_moyenne([12, 15]) affiche 13.5, mais ne renvoie rien (pas de return) :
# resultat vaut None, print(resultat) affiche None, et resultat * 2 lève
# TypeError: unsupported operand type(s) for *: 'NoneType' and 'int'


# 2. La fonction renvoie la moyenne : l'appelant décide de l'afficher
def moyenne(notes: list[float]) -> float | None:
  """Renvoie la moyenne de notes, ou None si notes est vide."""
  if len(notes) == 0:
    return None
  return sum(notes) / len(notes)


print(moyenne([12, 15]))  # 13.5
resultat = moyenne([])
if resultat is None:
  print("Pas de notes")

# 3. La docstring
help(moyenne)
````

### Conseils supplémentaires

Un n-uplet se déballe en autant de variables qu'il a d'éléments (`xa, ya, za = a`). Une valeur par défaut peut être un n-uplet : il est immuable, donc sans le piège des valeurs par défaut mutables vu en cours. `math.sqrt(3)` vaut environ 1.732.

```python
import math


def distance(
  a: tuple[float, float, float], b: tuple[float, float, float] = ...
) -> float:
  """..."""
  xa, ya, za = a
  xb, yb, zb = ...
  return math.sqrt(...)


print(distance((1, 1, 1)), math.sqrt(3))
print(distance((1, 2, 3), (4, 6, 3)))
print(distance(b=(1, 2, 2), a=(0, 0, 0)))
```

### Solution

````python
import math


def distance(
  a: tuple[float, float, float], b: tuple[float, float, float] = (0, 0, 0)
) -> float:
  """Renvoie la distance euclidienne entre les points a et b (l'origine par défaut)."""
  xa, ya, za = a
  xb, yb, zb = b
  return math.sqrt((xb - xa) ** 2 + (yb - ya) ** 2 + (zb - za) ** 2)


print(distance((1, 1, 1)), math.sqrt(3))  # 1.7320508075688772 deux fois
print(distance((1, 2, 3), (4, 6, 3)))  # 5.0
print(distance(b=(1, 2, 2), a=(0, 0, 0)))  # 3.0
````

### Conseils supplémentaires

Un paramètre et une variable affectée dans une fonction sont locaux : ils masquent la variable globale du même nom sans la modifier. Une valeur par défaut est calculée une seule fois, quand Python exécute le `def` : tous les appels qui l'utilisent partagent donc la même liste. La correction habituelle prend `None` par défaut et crée la liste dans le corps de la fonction. Le mot-clef `global` permet d'affecter une variable globale depuis une fonction. Vous devez obtenir `['courses']`, `['ménage']` et `['arrosage', 'jardin']` avec la version corrigée.

```python
def ajoute_tache_corrigee(tache: str, taches: list[str] | None = None) -> list[str]:
  if ...:
    taches = []
  taches.append(tache)
  return taches


print(ajoute_tache_corrigee("courses"))
print(ajoute_tache_corrigee("ménage"))
print(ajoute_tache_corrigee("jardin", ["arrosage"]))


def modifie_globale() -> None:
  ...
  message = "modifié par la fonction"


modifie_globale()
print(message)
```

### Solution

````python
# 1. Les prédictions :
# affiche()                 défaut : le paramètre prend sa valeur par défaut
# affiche(message)          global : on passe la variable globale en argument
# modifie()                 local : message est une variable locale à modifie
# print(message)            global : modifie n'a pas touché la variable globale
# ajoute_tache("courses")   ['courses']
# ajoute_tache("ménage")    ['courses', 'ménage'] : la liste par défaut est partagée
# ajoute_tache("jardin", …) ['arrosage', 'jardin'] : une liste est fournie
# ajoute_tache("vaisselle") ['courses', 'ménage', 'vaisselle'] : la liste partagée


# 2. Valeur par défaut None, et une nouvelle liste à chaque appel
def ajoute_tache_corrigee(tache: str, taches: list[str] | None = None) -> list[str]:
  if taches is None:
    taches = []
  taches.append(tache)
  return taches


print(ajoute_tache_corrigee("courses"))  # ['courses']
print(ajoute_tache_corrigee("ménage"))  # ['ménage']
print(ajoute_tache_corrigee("jardin", ["arrosage"]))  # ['arrosage', 'jardin']


# 3. global : l'affectation vise la variable du module
def modifie_globale() -> None:
  global message
  message = "modifié par la fonction"


modifie_globale()
print(message)  # modifié par la fonction
# Avec global, n'importe quelle fonction peut changer la variable : pour comprendre
# sa valeur, il faut lire tout le programme. Une fonction qui renvoie une valeur
# dit ce qu'elle produit dans sa signature, et l'appelant décide quoi en faire.
````

### Conseils supplémentaires

Le milieu d'une liste est à l'indice `len(casiers) // 2`. La tranche `casiers[:milieu]` est la moitié gauche, sans le milieu, et `casiers[milieu + 1:]` la moitié droite, sans le milieu non plus. Chaque appel récursif doit porter sur une liste plus courte : c'est ce qui garantit d'atteindre un cas de base. Une fonction qui s'appelle sans fin lève `RecursionError` après environ 1000 appels imbriqués. Vous devez obtenir `True`, `True`, `False` et `False`, et 4 appels pour trouver le casier 41.

```python
def est_libre(casiers: list[int], numero: int) -> bool:
  if ...:  # cas de base : liste vide
    return False
  milieu = len(casiers) // 2
  if casiers[milieu] == numero:  # cas de base : trouvé
    return ...
  if numero < casiers[milieu]:
    return est_libre(..., numero)  # appel récursif sur la moitié gauche
  return est_libre(..., numero)  # appel récursif sur la moitié droite


print(est_libre(casiers_libres, 41))
print(est_libre(casiers_libres, 91))
print(est_libre(casiers_libres, 40))
print(est_libre(casiers_libres, 100))
```

### Solution

````python
# 1. Deux cas de base, puis un appel récursif sur une moitié, sans le milieu
def est_libre(casiers: list[int], numero: int) -> bool:
  """Renvoie True si numero est dans la liste triée casiers, par dichotomie."""
  if len(casiers) == 0:  # cas de base : rien à chercher
    return False
  milieu = len(casiers) // 2
  if casiers[milieu] == numero:  # cas de base : trouvé
    return True
  if numero < casiers[milieu]:
    return est_libre(casiers[:milieu], numero)
  return est_libre(casiers[milieu + 1 :], numero)


print(est_libre(casiers_libres, 41))  # True
print(est_libre(casiers_libres, 91))  # True
print(est_libre(casiers_libres, 40))  # False
print(est_libre(casiers_libres, 100))  # False

# 2. Pour le casier 41, print(casiers) affiche 4 listes, chacune deux fois plus courte :
# les 16 casiers, [3, 8, 12, 15, 21, 27, 33, 41], [27, 33, 41], puis [41].
# Un parcours depuis le début compare 41 à 8 casiers. La recherche dichotomique fait au
# plus 6 appels pour 16 casiers, et un de plus chaque fois que la liste double : une
# dizaine pour 1000 casiers, là où un parcours peut demander 1000 comparaisons.

# 3. Avec casiers[milieu:], la moitié droite garde le milieu : pour le casier 95, la
# liste finit par se réduire à [91], et chaque appel se rappelle avec cette même liste.
# Le cas de base n'est jamais atteint, et Python lève
# RecursionError: maximum recursion depth exceeded
````

### Solution

````python
def journal(message: str, *details: str, niveau: str = "INFO", **contexte) -> str:
  """Renvoie une ligne de journal : niveau, message, détails et contexte."""
  ligne = f"[{niveau}] {message}"
  for detail in details:
    ligne += f" | {detail}"
  if contexte:
    paires = [f"{cle}={valeur}" for cle, valeur in contexte.items()]
    ligne += f" ({', '.join(paires)})"
  return ligne


print(journal("Démarrage"))
print(journal("Article saisi", "Claviers", "casier B4", bras=2))
print(journal("Batterie faible", niveau="ALERTE", bras=1, charge="12 %"))
print(journal("Test", "ALERTE"))  # [INFO] Test | ALERTE
# "ALERTE" est passé par position : c'est un détail. niveau, placé après *details,
# ne peut recevoir une valeur que par mot-clef.
print(journal(message="Arrêt"))  # [INFO] Arrêt
# message est un paramètre ordinaire : il se passe aussi par mot-clef. Déclaré avant
# /, il ne se passerait que par position : message="Arrêt" irait dans contexte, et
# l'appel lèverait TypeError: journal() missing 1 required positional argument:
# 'message'.
````

### Conseils supplémentaires

Une fonction qui contient `yield` est un générateur : l'appeler n'exécute pas son corps, mais renvoie un objet qu'on parcourt avec une boucle `for`, une compréhension ou `next`. Chaque `yield` produit une valeur et suspend la fonction jusqu'à la demande suivante. Un générateur épuisé ne donne plus rien. Vous devez trouver une moyenne de 61.8.

```python
from collections.abc import Iterator


def lignes_utiles(chemin: str) -> Iterator[str]:
  with open(chemin, encoding="utf-8") as fichier:
    for ligne in fichier:
      ligne = ligne.strip()
      if ...:
        yield ...


mesures = [... for ligne in lignes_utiles("humidite.txt")]
moyenne = ...
print(moyenne)
print([... for mesure in mesures if ...])

lignes = lignes_utiles("humidite.txt")
print(lignes)  # Prédiction : ...
print(next(lignes))  # Prédiction : ...
print(list(lignes))  # Prédiction : ...
print(list(lignes))  # Prédiction : ...
```

### Solution

````python
from collections.abc import Iterator


def lignes_utiles(chemin: str) -> Iterator[str]:
  """Produit les lignes de chemin ni vides ni commentaires, sans espaces de bord."""
  with open(chemin, encoding="utf-8") as fichier:
    for ligne in fichier:
      ligne = ligne.strip()
      if ligne and not ligne.startswith("#"):
        yield ligne


mesures = [float(ligne) for ligne in lignes_utiles("humidite.txt")]
moyenne = sum(mesures) / len(mesures)
print(moyenne)  # 61.8
print([mesure for mesure in mesures if mesure > moyenne])  # [62.5, 64.0, 63.0]

lignes = lignes_utiles("humidite.txt")
print(lignes)  # <generator object lignes_utiles at 0x...> : rien n'est encore lu
print(next(lignes))  # 62.5 : le corps s'exécute jusqu'au premier yield
print(list(lignes))  # ['64.0', '58.5', '61.0', '63.0'] : les lignes restantes
print(list(lignes))  # [] : le générateur est épuisé
````

Le paramètre `key` de `sorted` reçoit une fonction appelée sur chaque élément : les éléments sont triés selon ce qu'elle renvoie. Trier le dictionnaire trie ses clefs, les noms des articles.   n: int = 3,

### Solution

````python
def plus_proches(
  positions: dict[str, tuple[float, float, float]],
  n: int = 3,
  depart: tuple[float, float, float] = (0, 0, 0),
) -> list[str]:
  """Renvoie les n articles les plus proches de depart, du plus proche au plus loin."""
  tries = sorted(positions, key=lambda article: distance(positions[article], depart))
  return tries[:n]


print(plus_proches(positions))  # ['Câbles', 'Claviers', 'Casques']
print(plus_proches(positions, n=2, depart=positions["Claviers"]))
# ['Claviers', 'Casques']
distances = {
  article: round(distance(position), 1) for article, position in positions.items()
}
print(distances)
````
