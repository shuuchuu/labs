````python
# 1. Pseudocode of convert
# Inputs: amount, a number; source and target, two currency codes
# Output: the amount converted into the currency target, rounded to the cent
# Instructions:
#   in_euros ← amount / rates[source]
#   result ← in_euros × rates[target]
#   RETURN round(result, 2)


# 2. Each line of pseudocode becomes a line of Python
def convert(amount: float, source: str, target: str) -> float:
  """Return amount converted from the currency source to target, to the cent."""
  in_euros = amount / rates[source]
  result = in_euros * rates[target]
  return round(result, 2)


print(convert(100, "USD", "EUR"))  # 90.91
print(convert(100, "EUR", "JPY"))  # 16000.0
print(convert(50, "GBP", "CHF"))  # 55.88

# 3. Pseudocode of total_expenses
# Inputs: expenses, a list of tuples (amount, currency); target, a currency
# Output: the total of the expenses in the currency target, rounded to the cent
# Instructions:
#   total ← 0
#   FOR EACH (amount, currency) IN expenses DO
#     total ← total + convert(amount, currency, target)
#   END FOR
#   RETURN round(total, 2)


def total_expenses(expenses: list[tuple[float, str]], target: str) -> float:
  """Return the total of expenses, converted into the currency target, to the cent."""
  total = 0
  for amount, currency in expenses:
    total = total + convert(amount, currency, target)
  return round(total, 2)


print(total_expenses(expenses, "EUR"))  # 324.53
print(total_expenses(expenses, "USD"))  # 356.97
````

## Bubble Sort

The list `references`, which the next cell creates, gives the references of a shop's items, in the order they were entered. Bubble sort puts them in increasing order: it goes through the list comparing each element with its right-hand neighbor, and swaps them if they are out of order. At the end of a pass, the largest element has reached the end, like a bubble rising: the next pass can stop one step earlier. In pseudocode, with indices starting at 0, as in Python (the course's pseudocode numbers from 1), and inclusive `FOR` bounds:


### Solution

````python
def bubble_sort(values: list[int]) -> list[int]:
  """Return a new list, the elements of values in increasing order."""
  ordered = values.copy()  # sorting a copy leaves values unchanged
  n = len(ordered)
  for sweep in range(n - 1):  # sweep from 0 to n - 2
    for i in range(n - 1 - sweep):  # i from 0 to n - 2 - sweep
      if ordered[i] > ordered[i + 1]:
        ordered[i], ordered[i + 1] = ordered[i + 1], ordered[i]
  return ordered


sorted_references = bubble_sort(references)
print(sorted_references)
# [1033, 1876, 2219, 2540, 3398, 4127, 5503, 6075, 6612, 7781, 8264, 9025]
print(sorted_references == sorted(references))  # True
print(references)  # unchanged: [4127, 1033, 7781, ...]

# 3. The first pass compares 11 pairs of neighbors, the second 10, ..., the last
# one 1: 11 + 10 + ... + 1 = 12 * 11 / 2 = 66 comparisons. For N elements,
# (N - 1) + (N - 2) + ... + 1 = N(N - 1) / 2: a complexity in O(N²), like the
# course's selection sort.
````

### Solution

````python
def linear_search(values: list[int], target: int) -> int | None:
  """Return the index of the first element of values equal to target, or None."""
  for i in range(len(values)):
    if values[i] == target:
      return i
  return None


def binary_search(values: list[int], target: int) -> int | None:
  """Return an index of target in the sorted list values, or None."""
  low = 0
  high = len(values) - 1
  while low <= high:  # target can only be between the indices low and high
    middle = (low + high) // 2
    if values[middle] == target:
      return middle
    if values[middle] < target:
      low = middle + 1  # target can only be on the right of the middle
    else:
      high = middle - 1  # target can only be on the left of the middle
  return None


for reference in [6075, 9025, 5000]:
  print(
    reference,
    linear_search(inventory, reference),
    binary_search(inventory, reference),
  )
# 6075 7 7
# 9025 11 11
# 5000 None None

# 4. Each turn of binary_search roughly halves the number of elements left to
# examine: 12, then at most 6, 3 and 1, so at most 4 turns for 12 references. For a
# million, at most 20 turns (2 ** 20 = 1,048,576): a complexity in O(log N).
# linear_search takes one turn per element for a missing reference: 12 turns, and a
# million for a million references, in O(N).
````

## Going Further: Quicksort

This exercise is optional. Quicksort sorts a list by splitting it:

- a list of zero or one element is already sorted: we return a copy of it;
- otherwise, we pick an element, the pivot, here the middle one, and we share the elements out into three lists: those smaller than the pivot, those equal to the pivot and those larger;
- we sort the smaller and the larger ones with the same algorithm, through a recursive call, and we return the smaller ones sorted, followed by the equal ones, then by the larger ones sorted.

The course takes the first value as the pivot, and puts the values equal to the pivot with the larger ones. On an already sorted list, that pivot is each time the smallest value: each call removes only the pivot, so a list of about 1,000 elements goes past Python's limit on nested calls and raises a `RecursionError`. A list whose values are all equal does the same, since they all go to the same side. Hence the two choices of this exercise: the middle pivot makes an already sorted list a favorable case, and the third list, that of the equal ones, settles the case of equal values.

### Solution

````python
def quick_sort(values: list[int]) -> list[int]:
  """Return a new list, the elements of values sorted, by quicksort."""
  if len(values) <= 1:  # base case: already sorted
    return values.copy()
  pivot = values[len(values) // 2]
  smaller = []
  equal = []
  larger = []
  for value in values:
    if value < pivot:
      smaller.append(value)
    elif value == pivot:
      equal.append(value)
    else:
      larger.append(value)
  return quick_sort(smaller) + equal + quick_sort(larger)


check_sort(quick_sort)

quick_durations = []
for size in sort_sizes:
  numbers = random_values(size)
  start = time.perf_counter()
  quick_sort(numbers)
  quick_durations.append(time.perf_counter() - start)
plot(
  sort_sizes,
  {
    "bubble sort": bubble_durations,
    "sorted": sorted_durations,
    "quicksort": quick_durations,
  },
  "Duration of the sorts",
)

# 4. Bubble sort takes about 4 times longer when the size doubles: 100,000
# elements, 50 times more than 2000, take it 50² = 2500 times longer: several
# minutes, when 2000 elements take a few tenths of a second.
numbers = random_values(100_000)
start = time.perf_counter()
quick_sort(numbers)
print(time.perf_counter() - start)  # a few tenths of a second
# Quicksort is in O(N log N) on average. Its worst case is in O(N²): when the pivot
# is each time the smallest or the largest element, one of the two lists gets all
# the others. The middle pivot avoids it on an already sorted list, where taking
# the first element as the pivot would lead to it.
````
