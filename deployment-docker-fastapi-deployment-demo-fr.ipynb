# Déploiement

Cette démonstration porte sur l'utilisation de [Docker](https://www.docker.com/) et de [FastAPI](https://fastapi.tiangolo.com/).

Le matériel présenté se trouve à l'adresse suivante : https://dagshub.com/m09/landscape-classifier/src/solution

En particulier :

- [`Dockerfile`](https://dagshub.com/m09/landscape-classifier/src/solution/Dockerfile)
- [Fichier FastAPI](https://dagshub.com/m09/landscape-classifier/src/solution/landscape_classifier/api.py)

La commande finale montrée pour lancer l'API avec Docker était :

    docker run --rm -it -p 8000:80 fastapi
