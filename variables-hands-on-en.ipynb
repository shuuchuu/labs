The calls and the rules to use:

### Additional hints

The rules to apply:

- precedence: `**` first, then `*`, `/`, `//` and `%`, then `+` and `-`; parentheses come before everything
- at equal precedence, evaluation goes left to right, except `**`, right to left: `a ** b ** c` is `a ** (b ** c)`
- `//` rounds down (towards minus infinity), not towards zero, and the remainder `%` has the sign of the divisor: `a == (a // b) * b + a % b`
- `+` between two strings joins them end to end; `*` between a string and an integer repeats it
- Python never converts a string to a number by itself, nor the reverse: `+` between a `str` and an `int` raises a `TypeError`; convert first with `int` or `str`

### Additional hints

The rules to apply:

- comparisons chain: `a < b < c` means `a < b and b < c`
- two strings are compared character by character, by their Unicode code: uppercase letters come before lowercase ones (`"Z"` is 90, `"a"` 97), and the digits of a string are compared as characters, not as numbers
- a `float` is stored in binary, where `0.1` or `1.1` have no exact representation: calculations keep a tiny error. `math.isclose(a, b)` (after `import math`) compares with a tolerance
- `or` returns its first true operand (the last one if there is none), `and` its first false operand (the last one if they are all true): the result isn't necessarily a `bool`. False values are `0`, `0.0`, `""`, `None` and `False`
- `and` and `or` stop as soon as the result is known: the second operand is then not computed
- `&` is the bitwise "and": 6 is `110` in binary, 3 is `011`

A skeleton to copy into the exercise cell, replacing each `...`:

```python

print(...)
...
print(math.isclose(..., ...))
```

### Solution

````python
import math

print(1 < 2 < 3)  # True: 1 < 2 and 2 < 3
print("abc" < "abd")  # True: "c" < "d"
print("Zebra" < "apple")  # True: "Z" (90) < "a" (97)
print("10" < "9")  # True: "1" < "9", characters, not numbers
print(1.1 + 2.2 == 3.3)  # False: 1.1 + 2.2 is 3.3000000000000003
print(math.isclose(1.1 + 2.2, 3.3))  # True
print("" or "default")  # default: "" is false, or returns the other operand
print(0 or "")  # an empty line: no operand is true, or returns the last one, ""
print(6 and 3)  # 3: 6 is true, and returns the second operand
print(6 & 3)  # 2: 110 and 011 bitwise give 010
print(0 and 1 / 0)  # 0: and stops at 0, 1 / 0 is never computed
````
