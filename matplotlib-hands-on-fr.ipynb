## Un premier plot simple

Créez un premier plot qui trace la fonction sinus et cosinus.
* Créez une liste d'indice `x` avec [`np.linspace`](https://numpy.org/doc/stable/reference/generated/numpy.linspace.html) qui représentera votre axe des abscisses
* Créez les valeurs associées à `x` dans un tableau `y_cos` et `y_sin` en utilisant [`np.cos`](https://numpy.org/doc/stable/reference/generated/numpy.cos.html) et [`np.sin`](https://numpy.org/doc/stable/reference/generated/numpy.sin.html)
* Ajoutez une coloration entre les deux plots cosinus et sinus en utilisant [`ax.fill_between`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.fill_between.html?highlight=fill_between#matplotlib.axes.Axes.fill_between) et changez la couleur par défaut avec la couleur `#B8DDF9`
* Ajoutez un titre avec [`ax.set_title`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_title.html?highlight=title#matplotlib.axes.Axes.set_title)
* Ajoutez une légende avec [`ax.legend`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.legend.html?highlight=legend#matplotlib.axes.Axes.legend)

**Attention prenez de bonnes habitudes.** Évitez d'utiliser directement `plt.plot` et forcez-vous à créer un objet figure et un objet axes avec [`plt.subplots`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html), ou une combinaison de [`plt.figure`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.figure.html?highlight=figure#matplotlib.pyplot.figure) et [`plt.subplot`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplot.html?highlight=subplot#matplotlib.pyplot.subplot) ou [`fig.add_subplot`](https://matplotlib.org/stable/api/_as_gen/matplotlib.figure.Figure.add_subplot.html), puis d'utiliser `ax.plot`.



```python
# # Création des données de plot
# x = ...
# y_cos = ...
# y_sin = ...

# # Création figure et axe
# fig, ax = ...

# # Création du plot
# # Cosinus
# ax.??
# # Sinus
# ax.??
# Coloration entre les deux courbes. Couleur #B8DDF9
# ax.??

# # Ajout d'un titre et d'une légende
# ax.??
# ax.??

# plt.show()
```

### Solution

```python
# Création des données de plot
x = np.linspace(0, np.pi * 2)
y_cos = np.cos(x)
y_sin = np.sin(x)

# Création figure et axe
fig, ax = plt.subplots(dpi=100)

# Création du plot
# Cosinus
ax.plot(x, y_cos, label="cos")
# Sinus
ax.plot(x, y_sin, label="sin")
# Coloration entre les deux courbes Couleur #B8DDF9
ax.fill_between(x, y_cos, y_sin, color="#b8ddf9")


# Ajouter un titre et une légende
ax.set_title("Fonction cosinus et sinus")
ax.legend()

plt.show()
```

## Styles et annotations

Tracez la courbe $f(x) = e^{-x/3} \sin(x)$ sur $[0, 10]$ et signalez son maximum :
* Créez `x` avec `np.linspace` et calculez `y`, puis trouvez l'indice du maximum avec [`np.argmax`](https://numpy.org/doc/stable/reference/generated/numpy.argmax.html) et ses coordonnées `x_max` et `y_max`
* Appliquez un style de `plt.style.available` (la [galerie des styles](https://matplotlib.org/stable/gallery/style_sheets/style_sheets_reference.html) les montre) avec [`plt.style.context`](https://matplotlib.org/stable/api/style_api.html#matplotlib.style.context), dans un bloc `with` : contrairement à `plt.style.use`, qui change le style de toutes les figures suivantes du notebook, il ne vaut que pour les figures créées dans le bloc
* Dans ce bloc, tracez la courbe et marquez le maximum d'un point (`ax.plot(x_max, y_max, "o")`)
* Annotez le maximum avec [`ax.annotate`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.annotate.html) : le texte (la valeur du maximum), `xy` le point visé, `xytext` la position du texte, et `arrowprops=dict(arrowstyle="->")` pour la flèche qui les relie
* Ajoutez un titre qui donne la formule de la courbe

Certains styles (`ggplot`, `bmh`, `fivethirtyeight`) dessinent la flèche presque de la couleur du fond : ajoutez-lui alors une couleur, `arrowprops=dict(arrowstyle="->", color="black")`.

```python
# # Données de la courbe et coordonnées de son maximum
# x = ...
# y = ...
# i_max = ...
# x_max, y_max = ...

# # Le style ne vaut que dans le bloc with
# with plt.style.context(...):
#   fig, ax = ...
#   ax.??
#   ax.??
#   # Annotation du maximum, avec une flèche du texte vers le point
#   ax.annotate(...)
#   ax.??
#   plt.show()
```

### Solution

```python
# Données de la courbe et coordonnées de son maximum
x = np.linspace(0, 10, 500)
y = np.exp(-x / 3) * np.sin(x)
i_max = np.argmax(y)
x_max, y_max = x[i_max], y[i_max]

# Le style ne vaut que dans le bloc with
with plt.style.context("seaborn-v0_8-whitegrid"):
  fig, ax = plt.subplots(dpi=100)
  ax.plot(x, y)
  ax.plot(x_max, y_max, "o")
  # Annotation du maximum, avec une flèche du texte vers le point
  ax.annotate(
    f"maximum : {y_max:.2f}",
    xy=(x_max, y_max),
    xytext=(x_max + 2, y_max - 0.1),
    arrowprops=dict(arrowstyle="->"),
  )
  ax.set_title(r"$f(x) = e^{-x/3} \sin(x)$")
  plt.show()
```

## `imshow` et `contour`

Nous allons visualiser une fonction $f$ à deux paramètres $x$ et $y$ définie comme suit :

$$
  f(x,y) = z = \frac{\sin(x^2 + y^2)}{x^2 + y^2}
$$

Pour pouvoir l'afficher en deux dimensions, nous allons utiliser [`imshow`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.imshow.html) et [`contour`](https://matplotlib.org/stable/api/contour_api.html) :
* Créez les variables `x` et `y` nécessaires à l'affichage. Pour cela vous aurez besoin de [`np.linspace`](https://numpy.org/doc/stable/reference/generated/numpy.linspace.html) et [`np.meshgrid`](https://numpy.org/doc/stable/reference/generated/numpy.meshgrid.html) pour créer une valeur sur chaque pixel du résultat à afficher.
* Calculez `z`
* Affichez `z` avec [`imshow`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.imshow.html) en utilisant la transparence à 0.5 (`alpha=0.5`), `extent=[-5, 5, -5, 5]` pour que l'image couvre les mêmes coordonnées que `contour` (sans lui, ses axes sont gradués en numéros de pixels), et `origin="lower"` pour que la première ligne de `z`, qui correspond à $y = -5$, soit dessinée en bas, comme `contour` le fait (par défaut, `imshow` la dessine en haut)
* Affichez `z` avec [`contour`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.contour.html)

```python
# # Création des variables x et y
# # dans l'intervalle [-5,5]
# x = np.linspace(-5, 5, 1000)
# y = np.linspace(-5, 5, 1000)

# # Création des données en meshgrid
# xx, yy = ???

# # Calcul de z à partir de xx et yy
# z = ???

# # Création de fig et ax
# fig, ax = ???

# # Création du plot avec imshow et contour
# ax.???
# ax.???

# plt.show()
```

### Solution

```python
# Création des variables x et y
# dans l'intervalle [-5,5]
x = np.linspace(-5, 5, 1000)
y = np.linspace(-5, 5, 1000)

# Création des données en meshgrid
xx, yy = np.meshgrid(x, y)

# Calcul de z à partir de xx et yy
z = np.sin(xx**2 + yy**2) / (xx**2 + yy**2)

# Création de fig et ax
fig, ax = plt.subplots(dpi=100)

# Création du plot avec imshow et contour
ax.imshow(z, alpha=0.5, extent=[-5, 5, -5, 5], origin="lower")
ax.contour(xx, yy, z)

plt.show()
```

Une autre solution consiste évidemment à faire directement un affichage en 3D

```python
# Création des variables x et y
# dans l'intervalle [-5,5]
x = np.linspace(-5, 5, 1000)
y = np.linspace(-5, 5, 1000)

# Création des données en meshgrid
xx, yy = np.meshgrid(x, y)

# Calcul de z à partir de xx et yy
z = np.sin(xx**2 + yy**2) / (xx**2 + yy**2)

# Création de fig et ax
fig, ax = plt.subplots(dpi=150, subplot_kw=dict(projection="3d"))

# Création du plot en 3D avec plot_surface
g = ax.plot_surface(xx, yy, z, cmap="coolwarm")
fig.colorbar(g)
plt.show()
```

## Plots multiples et projections


Vous allez créer 4 plots différents des trois fonctions suivantes :
* $f(x) = 10^x$
* $f(x) = x$
* $f(x) = \log_{10}(x)$

Vous allez afficher ces trois fonctions sur 4 plots avec différentes échelles sur $x$ et $y$ :
1.   Axe $x$ linéaire et axe $y$ linéaire
2.   Axe $x$ logarithmique et axe $y$ linéaire
3.   Axe $x$ linéaire et axe $y$ logarithmique
4.   Axe $x$ logarithmique et axe $y$ logarithmique

en suivant la disposition suivante :

![](https://raw.githubusercontent.com/rougier/matplotlib-tutorial/master/figures/subplot-grid.png)

Afin de faire la visualisation :
* Utilisez [`subplots`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html?highlight=subplots#matplotlib.pyplot.subplots) pour créer une figure contenant 4 plots sur 2 lignes et 2 colonnes
* Utilisez [`set_xscale`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_xscale.html?highlight=set_xscale#matplotlib.axes.Axes.set_xscale) et [`set_yscale`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_yscale.html?highlight=set_yscale#matplotlib.axes.Axes.set_yscale) pour changer l'échelle des axes
* Utilisez [`set_xlim`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_xlim.html?highlight=set_xlim#matplotlib.axes.Axes.set_xlim) et [`set_ylim`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_ylim.html?highlight=set%20ylim#matplotlib.axes.Axes.set_ylim) pour définir les bornes d'affichages



```python
# x = np.linspace(0.001, 90, 5000)
# fig, ax = plt.subplots(2, 2, figsize=(8, 8), dpi=100)
```

### Solution

```python
x = np.linspace(0.001, 90, 5000)
fig, ax = plt.subplots(2, 2, figsize=(8, 8), dpi=100)


def subplot(ax: mpl.axes.Axes, xlog: bool = False, ylog: bool = False) -> None:
  if xlog:
    ax.set_xscale("log")
    ax.set_xlim(0.01, 100)
  else:
    ax.set_xlim(0, 10)
  if ylog:
    ax.set_yscale("log")
    ax.set_ylim(0.01, 100)
  else:
    ax.set_ylim(0, 10)
  ax.plot(x, 10**x, label="$f(x) = 10^x$")
  ax.plot(x, x, label="$f(x) = x$")
  ax.plot(x, np.log10(x), label=r"$f(x) = \log_{10}(x)$")
  ax.grid(visible=True, which="both", axis="both")
  x_title = f"$x$ {'logarithmique' if xlog else 'linéaire'}"
  y_title = f"$y$ {'logarithmique' if ylog else 'linéaire'}"
  ax.set_title(f"{x_title} - {y_title}")
  ax.legend()


# x linéaire, y linéaire
subplot(ax[0, 0])
ax[0, 0].set_ylabel("Linéaire")

# x log, y linéaire
subplot(ax[0, 1], xlog=True)

# x linéaire, y log
subplot(ax[1, 0], ylog=True)
ax[1, 0].set_xlabel("Linéaire")
ax[1, 0].set_ylabel("Log")

# x log, y log
subplot(ax[1, 1], xlog=True, ylog=True)
ax[1, 1].set_xlabel("Log")

plt.show()
```

## Projection polaire

Il est possible de faire des visualisations polaires en spécifiant une projection polaire.


```python
rng = np.random.default_rng()
x = rng.uniform(0, 2 * np.pi, size=15)

fig = plt.figure(dpi=100)
ax = fig.add_subplot(1, 2, 1)
ax.hist(x)
ax = fig.add_subplot(1, 2, 2, projection="polar")
ax.hist(x)

plt.show()
```
