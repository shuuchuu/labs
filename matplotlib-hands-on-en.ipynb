## A simple plot

Create a first plot representing the sine and cosine functions.
* Create an index list `x` with [`np.linspace`](https://numpy.org/doc/stable/reference/generated/numpy.linspace.html) that will represent your x-axis
* Create the values associated with `x` in an array `y_cos` and `y_sin` using [`np.cos`](https://numpy.org/doc/stable/reference/generated/numpy.cos.html) and [`np.sin`](https://numpy.org/doc/stable/reference/generated/numpy.sin.html)
* Add a colored space between the two plots cosine and sine using [`ax.fill_between`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.fill_between.html?highlight=fill_between#matplotlib.axes.Axes.fill_between) and change the default color with the color `#B8DDF9`
* Add a title with [`ax.set_title`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_title.html?highlight=title#matplotlib.axes.Axes.set_title)
* Add a legend with [`ax.legend`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.legend.html?highlight=legend#matplotlib.axes.Axes.legend)

**Best practices** Avoid using `plt.plot` directly and force yourself to create a figure object and an axes objects with [`plt.subplots`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html), or a combination of [`plt.figure`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.figure.html) and [`plt.subplot`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplot.html) or [`fig.add_subplot`](https://matplotlib.org/stable/api/figure_api.html), then use `ax.plot`.

```python
# # Create the data for the plot
# x = ...
# y_cos = ...
# y_sin = ...

# # Create axes and figure
# fig, ax = ...

# # Plotting
# # Cosine
# ax.??
# # Sine
# ax.??
# In between space coloration. Color #B8DDF9
# ax.??

# # Add a title and a legend
# ax.??
# ax.??

# fig.show()
```

### Solution

```python
# Create the data for the plot
x = np.linspace(0, np.pi * 2)
y_cos = np.cos(x)
y_sin = np.sin(x)

# Create axes and figure
fig, ax = plt.subplots(dpi=100)

# Plotting
# Cosine
ax.plot(x, y_cos, label="cos")
# Sine
ax.plot(x, y_sin, label="sin")
# In between space coloration. Color #B8DDF9
ax.fill_between(x, y_cos, y_sin, color="#b8ddf9")


# Add a title and a legend
ax.set_title("Cosine and sine functions")
ax.legend()

fig.show()
```

## `imshow` and `contour`

We will visualize a function $f$ with two parameters $x$ and $y$ defined as follows:

$$
  f(x,y) = z = \frac{sin(x^2 + y^2)}{x^2 + y^2}
$$

In order to display it in two dimensions, we will use [`imshow`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.imshow.html) and [`contour`](https://matplotlib.org/stable/api/contour_api.html):
* Create the `x` and `y` variables required for the display. For this you will need [`np.linspace`](https://numpy.org/doc/stable/reference/generated/numpy.linspace.html) and [`np.meshgrid`](https://numpy.org/doc/stable/reference/generated/numpy.meshgrid.html) to create a value associate to each pixel of the displayed result.
* Calculate `z`
* Display `z` with [`imshow`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.imshow.html) using transparency at 0.5 (`alpha=0.5`)
* Display `z` with [`contour`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.contour.html)

```python
# # Variables creation x et y
# # in range [-5,5]
# x = np.linspace(-5, 5, 1000)
# y = np.linspace(-5, 5, 1000)

# # Visualization data using meshgrid
# xx, yy = ???

# # Compute z from xx and yy
# z = ???

# # Create fig and ax
# fig, ax = ???

# # Plotting  with imshow et contour
# ax.???
# ax.???

# fig.show()
```

### Solution

```python
# Create the x and y variables
# in the range [-5,5]
x = np.linspace(-5, 5, 1000)
y = np.linspace(-5, 5, 1000)

# Create the meshgrid data
xx, yy = np.meshgrid(x, y)

# Compute z from xx and yy
z = np.sin(xx**2 + yy**2) / (xx**2 + yy**2)

# Create fig and ax
fig, ax = plt.subplots(dpi=100)

# Create the plot with imshow and contour
ax.imshow(z, alpha=0.5, extent=[xx.min(), xx.max(), yy.min(), yy.max()])
ax.contour(xx, yy, z)

fig.show()
```

Another solution is obviously to make a 3D display

```python
# Variables creation x et y
# in range [-5,5]
x = np.linspace(-5, 5, 1000)
y = np.linspace(-5, 5, 1000)

# Visualization data using meshgrid
xx, yy = np.meshgrid(x, y)

# Compute z from xx and yy
z = np.sin(xx**2 + yy**2) / (xx**2 + yy**2)

# Create fig and ax
fig, ax = plt.subplots(dpi=150, subplot_kw=dict(projection="3d"))

# Plotting  with plot_surface
g = ax.plot_surface(xx, yy, z, cmap="coolwarm")
fig.colorbar(g)
plt.show()
```

## Multiple plots and projections


You will create 4 different plots of the following three functions:
* $f(x) = 10^x$
* $f(x) = x$
* $f(x) = \log(x)$

You will display these three functions on 4 plots with different scales on $x$ and $y$ :
1.   Linear $x$ axis and linear $y$ axis
2.   Logarithmic $x$ axis and linear $y$ axis
3.   Linear $x$ axis and logarithmic $y$ axis
4.   Logarithmic $x$ axis and logarithmic $y$ axis

using the following arrangement:

![](https://raw.githubusercontent.com/rougier/matplotlib-tutorial/master/figures/subplot-grid.png)

In order to do the visualization:
* Use [`subplots`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html?highlight=subplots#matplotlib.pyplot.subplots) to create a figure containing 4 plots in 2 rows and 2 columns
* Use [`set_xscale`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_xscale.html?highlight=set_xscale#matplotlib.axes.Axes.set_xscale) and [`set_yscale`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_yscale.html?highlight=set_yscale#matplotlib.axes.Axes.set_yscale) to change the scale of the axes
* Use [`set_xlim`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_xlim.html?highlight=set_xlim#matplotlib.axes.Axes.set_xlim) and [`set_ylim`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_ylim.html?highlight=set%20ylim#matplotlib.axes.Axes.set_ylim) to set the display bounds



```python
# X = np.linspace(0.001, 90, 5000)
# fig, ax = plt.subplots(2, 2, figsize=(8, 8), dpi=100)
```

### Solution

```python
x = np.linspace(0.001, 90, 5000)
fig, ax = plt.subplots(2, 2, figsize=(8, 8), dpi=100)


def subplot(ax: mpl.axes.Axes, xlog: bool = False, ylog: bool = False) -> None:
  if xlog:
    ax.set_xscale("log")
    ax.set_xlim(0.01, 100)
  else:
    ax.set_xlim(0, 10)
  if ylog:
    ax.set_yscale("log")
    ax.set_ylim(0.01, 100)
  else:
    ax.set_ylim(0, 10)
  ax.plot(x, 10**x, label="$f(x) = 10^x$")
  ax.plot(x, x, label="$f(x) = x$")
  ax.plot(x, np.log10(x), label=r"$f(x) = \log(x)$")
  ax.grid(visible=True, which="both", axis="both")
  x_title = f"$x$ {'logarithmic' if xlog else 'linear'}"
  y_title = f"$y$ {'logarithmic' if ylog else 'linear'}"
  ax.set_title(f"{x_title} - {y_title}")
  ax.legend()


# x linear, y linear
subplot(ax[0, 0])
ax[0, 0].set_ylabel("Linear")

# x log, y linear
subplot(ax[0, 1], xlog=True)

# x linear, y log
subplot(ax[1, 0], ylog=True)
ax[1, 0].set_xlabel("Linear")
ax[1, 0].set_ylabel("Log")

# x log, y log
subplot(ax[1, 1], xlog=True, ylog=True)
ax[1, 1].set_xlabel("Log")

fig.show()
```

## Polar projection

It is possible to make polar visualizations by specifying a polar projection.


```python
rng = np.random.default_rng()
x = rng.integers(0, 10, size=15)

fig = plt.figure(dpi=100)
ax = fig.add_subplot(1, 2, 1)
ax.hist(x)
ax = fig.add_subplot(1, 2, 2, projection="polar")
ax.hist(x)

fig.show()
```
