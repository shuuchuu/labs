### Conseils supplémentaires

Une boucle `for` sur le fichier ouvert donne ses lignes une à une, chacune avec son `"\n"` final. `ligne.strip()` retire les espaces et le saut de ligne des bords, une chaîne vide est fausse dans un `if`, `startswith("#")` repère les commentaires et `split()` sépare la date de la température. `releves.values()` donne les températures, dont `min`, `max` et `sum` calculent le minimum, le maximum et la somme. Vous devez trouver 7 relevés, un minimum de -3.5, un maximum de 4.5 et une moyenne de 0.5.

```python
releves = {}
with open("releves.txt", encoding="utf-8") as fichier:
  for ligne in fichier:
    ligne = ligne.strip()
    if ...:
      continue
    date, temperature = ...
    releves[date] = ...

temperatures = releves.values()
print(f"{len(releves)} relevés")
print(f"Minimum : {...:.1f} °C, maximum : {...:.1f} °C")
print(f"Moyenne : {...:.1f} °C")
```

### Solution

````python
releves = {}
with open("releves.txt", encoding="utf-8") as fichier:
  for ligne in fichier:
    ligne = ligne.strip()
    if not ligne or ligne.startswith("#"):
      continue
    date, temperature = ligne.split()
    releves[date] = float(temperature)

temperatures = releves.values()
print(f"{len(releves)} relevés")
print(f"Minimum : {min(temperatures):.1f} °C, maximum : {max(temperatures):.1f} °C")
print(f"Moyenne : {sum(temperatures) / len(temperatures):.1f} °C")
````

### Conseils supplémentaires

`json.dump(objet, fichier)` écrit dans un fichier ouvert, `json.dumps(objet)` renvoie une chaîne. Leur paramètre `indent=2` indente le résultat et `ensure_ascii=False` garde les accents au lieu de les écrire en séquences d'échappement (`\u00e9` pour « é »). `json.load(fichier)` renvoie l'objet relu. `films.jsonl` doit compter 6 lignes.

```python
import csv
import json

films = []
with open("films.csv", encoding="utf-8", newline="") as fichier:
  for film in csv.DictReader(fichier):
    film["annee"] = ...
    film["duree"] = ...
    films.append(film)

with open("films.json", mode="w", encoding="utf-8") as fichier:
  json.dump(films, fichier, ...)

with open("films.jsonl", mode="w", encoding="utf-8") as fichier:
  for film in films:
    fichier.write(...)

with open("films.json", encoding="utf-8") as fichier:
  relus = json.load(fichier)
print(relus[1]["titre"], type(relus[1]["annee"]))  # Prédiction : ...
```

### Solution

````python
import csv
import json

films = []
with open("films.csv", encoding="utf-8", newline="") as fichier:
  for film in csv.DictReader(fichier):
    film["annee"] = int(film["annee"])
    film["duree"] = int(film["duree"])
    films.append(film)

with open("films.json", mode="w", encoding="utf-8") as fichier:
  json.dump(films, fichier, indent=2, ensure_ascii=False)

with open("films.jsonl", mode="w", encoding="utf-8") as fichier:
  for film in films:
    fichier.write(json.dumps(film, ensure_ascii=False) + "\n")

with open("films.json", encoding="utf-8") as fichier:
  relus = json.load(fichier)
print(relus[1]["titre"], type(relus[1]["annee"]))  # <class 'int'> : JSON garde le type

with open("films.jsonl", encoding="utf-8") as fichier:
  print(fichier.read())
````
