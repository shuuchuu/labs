## Compréhension simple

La boucle ci-dessous range dans `longueurs` le nombre d'éléments de chaque liste de `d`. Obtenez le même dictionnaire, `{'a': 3, 'b': 5}`, avec une compréhension de dictionnaire.

````python
d = {"a": [1, 2, 3], "b": [1, 2, 3, 4, 5]}

longueurs = {}
for k, v in d.items():
  longueurs[k] = len(v)
print(longueurs)
````

````python
# Votre code ici
````

### Solution

````python
longueurs = {k: len(v) for k, v in d.items()}
print(longueurs)
````

## Fichier

La boucle ci-dessous compte, ligne à ligne, les caractères du code source du module `string` de la bibliothèque standard. Calculez ce total avec `sum`, sans construire de liste des longueurs : sur un fichier de plusieurs millions de lignes, elle occuperait la mémoire pour rien. Vous devez retrouver le nombre que la boucle affiche.

````python
import pathlib
import string

# Le code source du module string de la bibliothèque standard
source = pathlib.Path(string.__file__)
with source.open(encoding="utf8") as fh:
  total = 0
  for ligne in fh:
    total += len(ligne)
print(total)
````

````python
# Votre code ici
````

### Solution

````python
with source.open(encoding="utf8") as fh:
  total = sum(len(ligne) for ligne in fh)
print(total)
````

## Compréhensions à itérations multiples

````python
nombres = [1, 2, 3]
lettres = ["a", "b", "c"]
produit_cartesien = []
for a in nombres:
  for b in lettres:
    produit_cartesien.append((a, b))
print(produit_cartesien)
````

````python
# Votre code ici
````

### Solution

````python
produit_cartesien = [(a, b) for a in nombres for b in lettres]
print(produit_cartesien)
````

Pour aplatir, écrivez les deux `for` dans une même compréhension, dans l'ordre des boucles imbriquées : d'abord chaque `ligne` de `matrice`, puis chaque `valeur` de `ligne`. ```python


### Conseils supplémentaires

`capitales.items()` donne les couples `(pays, capitale)` ; une compréhension de dictionnaire les remet dans l'autre sens, la capitale en clé. Si deux pays avaient la même capitale, la même clé serait écrite deux fois, et seul le dernier pays resterait.

Un `if` à la fin d'une compréhension filtre : il retire les éléments qui ne remplissent pas la condition. ### Conseils supplémentaires

`timeit.timeit(fonction, number=1000)` appelle `fonction` 1000 fois et renvoie la durée totale, en secondes : divisée par 1000 puis multipliée par `1e6`, elle donne des microsecondes par appel. Une boucle sur la liste `[avec_boucle, avec_comprehension, avec_map]` évite d'écrire trois fois la mesure, et `fonction.__name__` donne le nom à afficher.

```python
def avec_boucle() -> list[int]:
  carres = []
  for x in valeurs:
    carres.append(...)
  return carres


def avec_comprehension() -> list[int]:
  return [...]


def avec_map() -> list[int]:
  return list(map(lambda x: ..., valeurs))


print(avec_boucle() == avec_comprehension() == avec_map())
for fonction in [avec_boucle, avec_comprehension, avec_map]:
  duree = timeit.timeit(fonction, number=...)
  print(f"{fonction.__name__} : {duree / 1000 * 1e6:.0f} µs par appel")
```

### Solution

````python
def avec_boucle() -> list[int]:
  carres = []
  for x in valeurs:
    carres.append(x * x)
  return carres


def avec_comprehension() -> list[int]:
  return [x * x for x in valeurs]


def avec_map() -> list[int]:
  return list(map(lambda x: x * x, valeurs))


print(avec_boucle() == avec_comprehension() == avec_map())
for fonction in [avec_boucle, avec_comprehension, avec_map]:
  duree = timeit.timeit(fonction, number=1000)
  print(f"{fonction.__name__} : {duree / 1000 * 1e6:.0f} µs par appel")

# Boucle et compréhension : même ordre de grandeur, la compréhension un peu devant
# ou à égalité selon la machine et la version de Python. `map` + `lambda` est le
# plus lent : un appel de fonction par valeur. La compréhension se choisit pour sa
# lisibilité, pas pour un gain de vitesse.
````
