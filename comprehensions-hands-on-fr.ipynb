## Compréhension simple

```python
import pathlib
import string

d = {"a": [1, 2, 3], "b": [1, 2, 3, 4, 5]}

longueurs = {}
for k, v in d.items():
  longueurs[k] = len(v)
print(longueurs)
```

```python
# Votre code ici
```

### Solution

```python
longueurs = {k: len(v) for k, v in d.items()}
print(longueurs)
```

## Compréhensions à itérations multiples

```python
nombres = [1, 2, 3]
lettres = ["a", "b", "c"]
produit_cartesien = []
for a in nombres:
  for b in lettres:
    produit_cartesien.append((a, b))
print(produit_cartesien)
```

```python
# Votre code ici
```

### Solution

```python
produit_cartesien = [(a, b) for a in nombres for b in lettres]
print(produit_cartesien)
```

## Fichier

```python
# Le code source du module string de la bibliothèque standard
source = pathlib.Path(string.__file__)
with source.open(encoding="utf8") as fh:
  total = 0
  for ligne in fh:
    total += len(ligne)
print(total)
```

### Solution

```python
with source.open(encoding="utf8") as fh:
  total = sum(len(ligne) for ligne in fh)
print(total)
```

## Compréhensions à itérations multiples avec conditions

```python
nombres = [1, 2, 3]
lettres = ["a", "b", "c"]
produit_cartesien = []
for i, a in enumerate(nombres):
  if i % 2 == 0:
    for j, b in enumerate(lettres):
      if j % 2 == 0:
        produit_cartesien.append((a, b))
print(produit_cartesien)
```

```python
# Votre code ici
```

### Solution

```python
produit_cartesien = [
  (a, b)
  for i, a in enumerate(nombres)
  if i % 2 == 0
  for j, b in enumerate(lettres)
  if j % 2 == 0
]
print(produit_cartesien)
```
